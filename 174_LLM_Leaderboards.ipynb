# LLM Leaderboards
*What LLM leaderboards are and why they exist, the four types of leaderboards, seven reasons not to trust them blindly, and a five-step practical guide to using them for model selection*

## What a leaderboard is

A benchmark is essentially an exam: a standardized test that measures one specific capability. Once a model takes that exam, the result needs to go somewhere visible. An **LLM leaderboard** is exactly that — a public ranking and comparison table showing how different models perform on a common set of evaluations, all in one place, so it's immediately clear which model currently leads on a given benchmark and how the rest compare.

## Why leaderboards exist

**A common reference across labs.** Every model listed took the same exam under the same conditions, which makes the comparison genuinely apples-to-apples — a leaderboard is what turns individual benchmark scores into a real ranking rather than a set of disconnected claims.

**Trust.** A lab self-reporting its own benchmark score has an obvious incentive to present its model favorably — the earlier caution about lab-published numbers applies directly here. A third-party leaderboard, with no comparable stake in any one model looking better than another, carries meaningfully more credibility than the same numbers reported by the model's own creator.

**Guiding model selection without running every evaluation yourself.** Suppose an application needs a model with strong math capability. Personally running every available model through a full evaluation isn't realistic — there are hundreds of models in circulation, and testing all of them would cost far more time and money than most teams can spend. A leaderboard does that work already: narrowing from hundreds of models down to a handful of serious candidates, which can then be tested more thoroughly with a smaller, more targeted evaluation.

**Detecting benchmark saturation.** When the top models on a given leaderboard cluster tightly together — several models all scoring within a couple of points of each other near the top — that clustering is itself a visible signal that the underlying benchmark is approaching saturation, the same pattern covered earlier for MMLU, GPQA, and others. A leaderboard makes that saturation obvious at a glance, well before anyone needs to dig into the raw numbers to notice it.

## Who uses leaderboards, and why

**AI engineers.** The most direct use case: shortlisting. Rather than selecting a single model straight off a leaderboard, the typical pattern is narrowing a large field down to a handful of serious candidates — say, from a hundred available models to five — using the leaderboard specific to the capability that matters most for the application (a math-focused leaderboard, a coding-focused one, and so on). Those candidates then get evaluated more specifically, using the same custom-eval approach covered earlier, before a final model gets chosen.

**Frontier labs.** Labs track leaderboards closely for two reasons: understanding exactly where they currently stand relative to competitors, and deciding when a new model is actually ready to release. A lab that trains a new iteration only to find it doesn't clearly beat a rival's current model has a strong incentive to hold that release back rather than ship something that visibly underperforms the competition — releasing a weaker follow-up invites exactly the kind of unflattering comparison a lab wants to avoid, so labs often wait for a version that leads meaningfully before making it public. This dynamic has also produced "stealth" releases — a new model quietly entered onto a leaderboard under an unfamiliar, sometimes deliberately playful codename, with no public statement about which lab built it, specifically to gauge how it performs and how it's received before formally attaching a company's name to it. If the reception is strong enough, the codename can even stick around permanently once the model is officially unveiled, simply because it already built up recognition and goodwill under that name.

**Researchers.** Leaderboards give the field visibility into what's actually happening in aggregate — which benchmarks have saturated, where genuine progress is still being made, and where a gap in the field might be worth a new research direction. Watching leaderboards over time is itself a way of tracking the state of the field.

**Policymakers and safety institutes.** These stakeholders monitor leaderboards to track what capability level current models have reached, watching in particular for any model that jumps unusually far ahead of the rest. If a new model's capabilities look like a significant, fast leap forward, that's a trigger for closer scrutiny — regulatory attention, safety review, or other intervention — precisely because leaderboards make an unusual jump immediately visible rather than something that would otherwise only surface later.

**The open-source community.** Leaderboards are also a genuine discovery mechanism. A small, relatively unknown lab that manages to place a new model near the top of even one specific leaderboard often gets discovered and talked about almost immediately as a direct result — this is exactly how a number of newer, smaller labs have gone from obscure to widely known in a short period, purely on the strength of a strong leaderboard placement drawing attention to what they'd built.

## Four types of leaderboards

Leaderboards aren't all built the same way. Four distinct types show up across the field, ranging from narrow and less useful to broad and genuinely practical for real decision-making.

### 1. Benchmark-specific leaderboards

The simplest kind: a leaderboard that ranks models using the results of exactly one benchmark — an MMLU leaderboard, a GSM8K leaderboard, a GPQA leaderboard, each showing which model performs best on that specific test. Most well-known benchmarks have a dedicated leaderboard like this, typically built and maintained directly by the research team behind the benchmark itself.

The problem with this type is the narrowness of what it actually tells you. Knowing which model tops one specific benchmark says nothing about how that model performs anywhere else — strong knowledge-benchmark performance says nothing about coding ability, and vice versa. This is the least generally useful of the four types, precisely because a single number on a single axis rarely reflects what matters for a real application.

### 2. Multi-benchmark leaderboards

This is where leaderboards become genuinely useful. Rather than relying on one test, a multi-benchmark leaderboard combines results across several benchmarks and evaluation dimensions — knowledge, reasoning, mathematics, coding, instruction following, data analysis, and more — into a single, cumulative comparison.

LiveBench is a good example: a benchmark suite spanning 23 objective tasks across seven categories, explicitly designed to resist contamination, reporting both category-level scores and one overall combined score per model.

Some organizations build this out even further. Artificial Analysis, for instance, is a company whose entire product is producing leaderboards like this — not just one combined capability leaderboard, but dedicated leaderboards for cost per task, latency, output speed, context window size, and further breakdowns by domain (coding agents, speech, image, audio, hardware), on top of specific per-benchmark leaderboards (an HLE leaderboard, a GPQA Diamond leaderboard, and so on) and one overall aggregate ranking pulling all of it together.

This type answers the question that actually matters most in practice: which model offers the strongest overall combination of capability, cost, and performance — which is exactly why it tends to be the most frequently used type of leaderboard for real model-selection decisions.

### 3. Human-preference-based leaderboards

This type doesn't rank models on any benchmark score at all — it ranks them on human votes. A user submits a question, which is sent anonymously to two different models (labeled simply "A" and "B," with no indication of which model is which), and after reading both responses, the user votes for the one they preferred — based on helpfulness, clarity, writing quality, creativity, or whatever else feels like the better answer. Once enough votes accumulate across enough users, a ranking emerges directly from that aggregate preference data.

LM Arena is the best-known example of this — its "battle mode" runs exactly this blind, head-to-head comparison at scale, across many thousands of users worldwide, covering categories like normal chat, coding, image generation, and video generation separately.

The clear limitation here is that a response a human prefers isn't necessarily the more *correct* one. People are genuinely susceptible to formatting, tone, and style — a nicely structured but less accurate answer can beat a plainer but more correct one purely on presentation. That said, this happens at a large enough scale, across enough independent voters, that the aggregate signal still tends to be broadly trustworthy — which is part of why the models that lead on more rigorous benchmark-based leaderboards also tend to lead on human-preference leaderboards like this one.

### 4. Application-specific leaderboards

The last type is built around one particular domain or task rather than general capability. A dedicated coding leaderboard might combine several coding-specific benchmarks into one score; other examples include leaderboards focused specifically on agentic task completion, generating SQL queries, or answering medical questions.

The Berkeley Function-Calling Leaderboard is a concrete real example — it specifically measures how well a model handles tool and function calling, ranking models on that one capability alone, even though it draws on multiple underlying benchmarks to do so. It operates entirely within a single domain, unlike the broad, general-purpose multi-benchmark leaderboards covered above.

## Summary: which type to reach for

Across these four types, usefulness scales roughly with breadth. A single-benchmark leaderboard is the least useful on its own, since it says nothing beyond one narrow capability. A multi-benchmark, general-capability leaderboard — especially one that also reports cost, latency, and other operational details — tends to be the most practically useful for real model-selection decisions. Human-preference-based leaderboards are popular and carry real marketing weight (being "top of the arena" is a widely cited claim), but come with a known bias toward style over correctness. And application-specific leaderboards are the right tool specifically when the decision at hand is scoped to one domain — coding, tool use, or any other single, well-defined task.

## Why leaderboards shouldn't be trusted blindly

Leaderboards genuinely are useful — a single, summarized view of the entire LLM landscape at a glance. But for someone working as an AI engineer, model selection is typically the very first step of building an application, and checking a leaderboard is typically the first step *within* model selection. That makes it easy to develop an unconscious bias: assuming that a model topping a leaderboard must be the right choice, simply because it's visibly on top. That assumption doesn't hold up well under scrutiny, for several distinct reasons.

### 1. Benchmark performance may not transfer to real applications

A model scoring 80–90% on a leaderboard doesn't guarantee equally strong results in an actual application built on it. This mirrors a piece of well-known advice from data science: winning competitions on a platform like Kaggle doesn't automatically make someone a strong real-world data scientist, because competition datasets are unusually clean, with a clearly defined problem statement — real-world work is messy in ways a competition rarely is. The same logic applies to benchmarks: their data tends to be clean and well-specified, so models naturally perform well on it. Real-world usage brings ambiguous requests, missing information, company-specific data the model has never seen, tool failures, and unusual edge cases — none of which a clean benchmark score says anything about.

### 2. Contamination inflates leaderboard scores too

Benchmarks are prone to contamination, as covered earlier — and a contaminated benchmark produces an inflated, unreliable leaderboard score by direct extension. A high score might be entirely genuine, or it might partly reflect a model having already seen the exact questions (and their answers) during training. A leaderboard position alone gives no way to tell which of the two is actually happening.

### 3. Models can be over-optimized for the leaderboard itself

This is a genuine, increasingly discussed problem: once a specific leaderboard becomes popular enough — LM Arena is the clearest example — being ranked highly on it starts to matter for its own sake, independent of what the ranking is actually supposed to measure. Winning votes on a human-preference leaderboard specifically rewards responses that are well-formatted, agreeable, and confident-sounding, whether or not the underlying answer is actually the most accurate or useful one. A company noticing this can start fine-tuning its models specifically to perform well in that setting — feeding training data that optimizes for the kind of response that wins votes, rather than for genuine capability. The model's leaderboard score climbs, but its real-world capability doesn't improve to match.

This is a direct instance of **Goodhart's Law**: when a measure becomes a target, it stops being a good measure. A useful way to picture this: imagine a car manufacturer that knows fuel efficiency is the single number most buyers fixate on, so the entire engineering team redirects its effort toward improving mileage specifically. Mileage climbs — but acceleration, handling, and overall driving quality quietly suffer, because none of the engineering effort went there. The same dynamic plays out when a model gets built specifically to top a leaderboard rather than to be genuinely capable — the leaderboard number improves while real-world performance doesn't necessarily follow.

### 4. Multi-benchmark leaderboards can hide their own methodology

For leaderboards that combine several benchmarks into one overall score, two important details are often left unclear: exactly which benchmarks were included or excluded from the combined score, and how results across different capabilities were normalized and weighted against each other. A leaderboard that quietly weights coding benchmarks more heavily than knowledge benchmarks, say, produces a very different "overall" ranking than one that weights them evenly — and that choice isn't always disclosed. The more transparent a leaderboard is about its own methodology, the more its ranking can actually be trusted.

### 5. Small score differences don't mean much

A model scoring 84.3 landing in 3rd place and one scoring 84.1 landing in 5th looks like a meaningful gap purely because of the rank labels attached — third place sounds notably better than fifth. But the actual difference in score is only 0.2 points, which is close enough that the two models are functionally similar for most purposes, and the fifth-ranked one might even be the better fit for a specific application. Fixating on exact rank position, rather than the size of the actual gap in the underlying numbers, is its own kind of bias. The same logic shows up in any large-scale ranked exam: the difference between a top rank and one a few dozen places behind often comes down to a question or two, not a meaningful gap in overall ability.

### 6. Human-preference leaderboards carry human bias

This was touched on earlier and is worth restating directly: the top model on a human-preference leaderboard like LM Arena isn't necessarily the objectively strongest model. People tend to prefer longer answers, more confident-sounding answers, better-formatted answers, and more entertaining answers — none of which is the same thing as correctness. A genuinely stronger model that happens to answer more plainly, more concisely, or less confidently can lose head-to-head votes to a flashier but less accurate one, purely on presentation.

### 7. Leaderboard scores can be stale, incomplete, or self-reported

Leaderboards don't always stay current — a newly released model might simply not be listed yet on a given leaderboard, because whoever maintains it hasn't gotten around to testing and adding it, leaving the visible ranking populated with slightly outdated model versions rather than the current state of the art. Some leaderboards, on top of that, allow companies to submit their own results directly for their own models — which undercuts exactly the third-party trust benefit that makes leaderboards valuable in the first place.

## The takeaway

None of this makes leaderboards useless — they remain a genuinely valuable, high-level view of the field. But they deserve the same careful, skeptical reading as any individual benchmark score: useful as a starting point for narrowing down candidates, not as a substitute for actually testing a shortlisted model against the specific task it needs to handle.

## A practical guide to reading leaderboards

Putting everything above together, here's how model selection using leaderboards should actually work in practice, as an AI engineer building a real application.

### Step 1: Clarify requirements before looking at any leaderboard

Before opening a single leaderboard, get clear on: what kind of application is being built, how much latency it can tolerate, what the cost budget looks like, what context-length needs it has, and whether there are deployment constraints — can this run on a publicly available hosted model, or does it need to run on-premise?

Getting these answers straight first is what prevents the bias toward automatically favoring whichever model sits at rank one. If on-premise deployment is a hard requirement, for instance, a proprietary closed model is off the table regardless of how it ranks — that constraint naturally points toward open-source options instead, and knowing that upfront keeps the search objective rather than starting from "whatever's ranked highest."

### Step 2: Go to the leaderboard that actually matches the task

Rather than checking whichever leaderboard happens to be popular, go to the one that measures what the application actually needs:

- Building an agent → an agentic-task leaderboard.
- Building a conversational chatbot → a human-preference leaderboard like LM Arena, since that gives a direct measurement of conversational quality specifically.
- Building a RAG system → **MTEB** (the Massive Text Embedding Benchmark leaderboard) — RAG depends on an embedding model for retrieval, and MTEB is built specifically to rank those.
- Working under a tight budget → a cost- and speed-focused leaderboard, like the ones covered earlier that report price per task and latency directly alongside capability.

### Step 3: Read the leaderboard correctly, not just the top number

Once on the right leaderboard, actually understand what's behind the ranking: what's being scored, and how exactly it's scored; who ran the evaluation; what inference budget or settings were used; whether reasoning was enabled; how old the underlying evaluation dataset is and whether it's still being updated; whether a private, uncontaminated test set is maintained; whether the benchmarks behind it are already saturated; and whether confidence intervals are reported at all — a small evaluation dataset produces a wide confidence interval, which means two closely scored models may be statistically indistinguishable rather than genuinely different. For a composite leaderboard combining multiple benchmarks, it's also worth checking what weighting was given to each capability in the combined score.

None of this is usually front and center — it's the footnotes, the fine print, the key definitions, and the FAQ sections that carry this information. Reading those is what actually puts someone in a position to judge a leaderboard properly, rather than reacting to a single headline number.

### Step 4: Shortlist, don't select

Using the criteria from step 1 and a properly read leaderboard from steps 2 and 3, narrow the field down to three to five serious candidate models — not a single final choice yet.

### Step 5: Run a custom evaluation on the shortlist

This is the step that actually produces a final decision. The shortlisted candidates get run through a custom evaluation built around the actual application's data and task — the same custom-eval approach covered earlier in this notebook series — and whichever model performs best on that specific evaluation becomes the one actually used.

## The single most important line to take from this

**Leaderboards are a filtering tool, not a decision tool.** They're what narrows a field of hundreds of models down to a manageable shortlist — they were never meant to make the final call. That final call comes from running an actual custom evaluation on the shortlisted candidates against the real task at hand, not from reading off whichever model happens to sit at the top of a leaderboard.
