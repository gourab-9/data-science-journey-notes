# 🖼️ ImageNet Dataset

**ImageNet** is a large-sca
le, hierarchical image dataset developed by researchers at Stanford and Princeton (led by Fei-Fei Li) to advance research in **visual recognition** and **machine learning**.

- **Launched**: 2009  
- **Purpose**: To provide a comprehensive labeled image resource for training and evaluating models in **image classification**, **object detection**, and **image segmentation**.

---

## 🔑 Key Features

### 1. **Large-Scale Dataset**
- Contains **14+ million images**, each manually labeled with object categories.
- Over **22,000 categories** (called **synsets**) based on the **WordNet hierarchy**.
- Example: “Dog” synset contains sub-synsets like **Labrador**, **Bulldog**, etc.

### 2. **Diverse Categories**
- Covers a wide range: animals, objects, scenes, plants, tools, vehicles, etc.

### 3. **High-Quality Annotations**
- All labels are **human-annotated** for reliability and accuracy.

### 4. **Hierarchical Structure**
- Based on the **WordNet hierarchy**.
- Each node = **synset** (a set of synonymous words representing a concept).

### 5. **ILSVRC (ImageNet Large Scale Visual Recognition Challenge)**
- Annual competition (2010–2017) based on a subset of ImageNet.
- Contains:
  - **1,000 categories**
  - **1.2M training images**
  - **50k validation**
  - **100k test images**
- Catalyzed breakthroughs in image classification and object detection.

---

## 🚀 Impact on Deep Learning

### 1. **Catalyst for CNN Revolution**
- **AlexNet (2012)** won ILSVRC with a deep CNN, reducing top-5 error from 25% to **15%**.
- Sparked the deep learning boom in computer vision.

### 3. **Transfer Learning**
- Pretrained ImageNet models are widely used as **feature extractors**.
- Fine-tuned on smaller datasets (e.g. **medical**, **satellite**, **custom datasets**).

# 🧠 LeNet-5 – Pioneering Convolutional Neural Network

---

## 📌 What is LeNet-5?

LeNet-5 is one of the earliest and most influential convolutional neural network (CNN) architectures, developed by **Yann LeCun et al. in 1998**. It was designed to recognize **handwritten digits** from the MNIST dataset. This model laid the groundwork for modern deep learning in computer vision.

---

## 🌟 Features of LeNet-5

- Introduced the idea of combining **convolution + pooling + fully connected layers**.
- Uses **tanh activation functions** (ReLU wasn’t common then).
- Inspired architectures like **AlexNet, VGGNet, ResNet**.
- Low computational cost, designed for devices with limited power in the 90s.
- Introduced **sparse connectivity** between layers to reduce complexity.

---

## 🧱 Architecture of LeNet-5

### 1. **Input Layer**
- **Input Size**: 32×32×1 (Grayscale)
- MNIST images (28×28) were zero-padded to 32×32.

---

### 2. **Convolutional Layer (C1)**
- **Filters**: 6 filters of size 5×5
- **Output**: 28×28×6
- **Stride**: 1
- **Activation**: `tanh`

---

### 3. **Subsampling Layer (S2)** – Average Pooling
- **Filter**: 2×2
- **Stride**: 2
- **Output**: 14×14×6
- **Activation**: `tanh`

---

### 4. **Convolutional Layer (C3)**
- **Filters**: 16 filters of size 5×5
- **Output**: 10×10×16
- **Stride**: 1
- **Activation**: `tanh`
- **Connectivity**: Sparse – not every input map is connected to every output map

---

### 5. **Subsampling Layer (S4)** – Average Pooling
- **Filter**: 2×2
- **Stride**: 2
- **Output**: 5×5×16
- **Activation**: `tanh`

---

### 6. **Fully Connected Layer (C5)**
- **Input**: Flattened 5×5×16 = 400
- **Output**: 120
- **Activation**: `tanh`

---

### 7. **Fully Connected Layer (F6)**
- **Input**: 120
- **Output**: 84
- **Activation**: `tanh`

---

### 8. **Output Layer**
- **Input**: 84
- **Output**: 10 classes (digits 0–9)
- **Activation**: `softmax`

---

## 🧾 Summary of LeNet-5 Architecture

| Layer Type        | Feature Maps | Output Size | Kernel Size | Stride | Activation |
|------------------|--------------|-------------|-------------|--------|------------|
| **Input**         | 1            | 32×32       | -           | -      | -          |
| **Conv (C1)**     | 6            | 28×28       | 5×5         | 1      | tanh       |
| **AvgPool (S2)**  | 6            | 14×14       | 2×2         | 2      | tanh       |
| **Conv (C3)**     | 16           | 10×10       | 5×5         | 1      | tanh       |
| **AvgPool (S4)**  | 16           | 5×5         | 2×2         | 2      | tanh       |
| **FC (C5)**       | 120          | 1×1         | -           | -      | tanh       |
| **FC (F6)**       | 84           | 1×1         | -           | -      | tanh       |
| **Output**        | 10           | 1×1         | -           | -      | softmax    |

---

## 📜 Development History & Timeline

| Year | Milestone                                          |
|------|----------------------------------------------------|
| 1988 | First neural network used on ZIP code digits       |
| 1989 | First LeNet versions for digit recognition         |
| 1990 | LeNet using backpropagation on simple networks     |
| 1994 | LeNet-4 introduced sparse connections               |
| 1998 | **LeNet-5 officially published** (most cited)      |
| ~2005 | Replaced in production by deeper CNNs like AlexNet |

---

## 🧩 Applications

- Digit recognition (e.g., ZIP codes, bank checks)
- Early OCR systems
- Academic learning and baseline model for CNNs
- Embedded systems with low compute power

---

## 🔁 Influence & Subsequent Work

- Inspired deep architectures like **AlexNet, VGG, GoogLeNet, ResNet**
- Basis for CNN design patterns: Conv → Pool → FC
- Introduced **sparse connections** and **parameter sharing**
- Proof that CNNs could outperform traditional ML for vision

# 🧠 AlexNet – The Breakthrough CNN Model (2012)

---

## 📌 What is AlexNet?

**AlexNet** is a deep convolutional neural network developed by **Alex Krizhevsky, Ilya Sutskever, and Geoffrey Hinton**, which won the **ImageNet Large Scale Visual Recognition Challenge (ILSVRC)** in **2012** by a **huge margin**. It was the first model to prove the power of deep learning on large-scale image classification, marking a turning point in computer vision.

---

## 🌟 Features of AlexNet

- Consists of **8 layers**: 5 convolutional + 3 fully connected.
- First to use **ReLU activation** instead of tanh or sigmoid (improves training speed).
- Applied **dropout** to reduce overfitting.
- Used **data augmentation** and **GPU acceleration**.
- **Overlapped max pooling** instead of traditional non-overlapping pooling.
- Trained on **ImageNet dataset** with 1.2 million images across 1000 classes.

---

## 🧱 Architecture of AlexNet

### 1. **Input Layer**
- **Input Size**: 227×227×3 (RGB image)
- MNIST images were resized and normalized to fit this dimension.

---

### 2. **Convolutional Layer (Conv1)**
- **Filters**: 96
- **Filter Size**: 11×11
- **Stride**: 4
- **Padding**: 0
- **Output Size**: 55×55×96
- **Activation**: ReLU

---

### 3. **Max Pooling Layer (Pool1)**
- **Filter Size**: 3×3
- **Stride**: 2
- **Output Size**: 27×27×96

---

### 4. **Convolutional Layer (Conv2)**
- **Filters**: 256
- **Filter Size**: 5×5
- **Stride**: 1
- **Padding**: 2 (Same padding)
- **Output Size**: 27×27×256
- **Activation**: ReLU

---

### 5. **Max Pooling Layer (Pool2)**
- **Filter Size**: 3×3
- **Stride**: 2
- **Output Size**: 13×13×256

---

### 6. **Convolutional Layer (Conv3)**
- **Filters**: 384
- **Filter Size**: 3×3
- **Stride**: 1
- **Padding**: 1
- **Output Size**: 13×13×384
- **Activation**: ReLU

---

### 7. **Convolutional Layer (Conv4)**
- **Filters**: 384
- **Filter Size**: 3×3
- **Stride**: 1
- **Padding**: 1
- **Output Size**: 13×13×384
- **Activation**: ReLU

---

### 8. **Convolutional Layer (Conv5)**
- **Filters**: 256
- **Filter Size**: 3×3
- **Stride**: 1
- **Padding**: 1
- **Output Size**: 13×13×256
- **Activation**: ReLU

---

### 9. **Max Pooling Layer (Pool3)**
- **Filter Size**: 3×3
- **Stride**: 2
- **Output Size**: 6×6×256

---

### 10. **Fully Connected Layer (FC6)**
- **Input Size**: 6×6×256 = 9216 (flattened)
- **Output Size**: 4096
- **Activation**: ReLU

---

### 11. **Fully Connected Layer (FC7)**
- **Input Size**: 4096
- **Output Size**: 4096
- **Activation**: ReLU

---

### 12. **Fully Connected Layer (FC8) – Output Layer**
- **Input Size**: 4096
- **Output Size**: 1000 (ImageNet classes)
- **Activation**: Softmax

---

## 🧾 Summary of AlexNet Architecture

| Layer Type        | Feature Maps | Output Size   | Kernel Size | Stride | Padding | Activation |
|------------------|--------------|----------------|-------------|--------|---------|------------|
| **Input**         | 3            | 227×227        | -           | -      | -       | -          |
| **Conv1**         | 96           | 55×55          | 11×11       | 4      | 0       | ReLU       |
| **Pool1**         | 96           | 27×27          | 3×3         | 2      | -       | -          |
| **Conv2**         | 256          | 27×27          | 5×5         | 1      | 2       | ReLU       |
| **Pool2**         | 256          | 13×13          | 3×3         | 2      | -       | -          |
| **Conv3**         | 384          | 13×13          | 3×3         | 1      | 1       | ReLU       |
| **Conv4**         | 384          | 13×13          | 3×3         | 1      | 1       | ReLU       |
| **Conv5**         | 256          | 13×13          | 3×3         | 1      | 1       | ReLU       |
| **Pool3**         | 256          | 6×6            | 3×3         | 2      | -       | -          |
| **FC6**           | 4096         | 1×1            | -           | -      | -       | ReLU       |
| **FC7**           | 4096         | 1×1            | -           | -      | -       | ReLU       |
| **FC8 (Output)**  | 1000         | 1×1            | -           | -      | -       | Softmax    |

## 🔑 Key Innovations in AlexNet

### 1. **ReLU Activation Function**
- ReLU (Rectified Linear Unit) introduced non-linearity without saturating gradients.
- ReLU accelerates convergence during training compared to traditional activation functions like **sigmoid** or **tanh**.
- Helped the network train faster and more efficiently.

### 2. **Dropout Regularization**
- Applied to fully connected layers.
- Randomly sets a fraction of neuron outputs to zero during training.
- Prevents overfitting by discouraging complex co-adaptations on training data.

### 3. **Data Augmentation**
- Artificially enlarged the training dataset using:
  - **Image translation**
  - **Horizontal flipping**
  - **Patch extraction**
- Helped in reducing overfitting and improving generalization.

### 4. **Use of GPUs**
- AlexNet utilized **two Nvidia GPUs** for training.
- The model was split across GPUs to enable **data and model parallelism**.
- This significantly accelerated training time on large datasets like ImageNet.

### 5. **Overlapping Pooling**
- Used **stride = 2** with **window size = 3**.
- Unlike non-overlapping pooling (stride = 2, window size = 2), overlapping pooling retains more spatial information.
- Resulted in reduced error rates.

---

## ⚙️ TensorFlow Optimizations

### 🔄 **Prefetch in TensorFlow**
- Prefetching overlaps **data preprocessing** and **model execution**.
- While the model is training on one batch, the **next batch is being prepared** in the background.
- This reduces model idle time and ensures **efficient use of CPU/GPU**.

#### 📌 When to Use:
- Ideal when image loading, decoding, and transformations are **CPU-intensive**.
- Keeps the GPU busy and prevents bottlenecks due to slow input pipelines.

---

### 🤖 **AUTOTUNE in TensorFlow**
- `tf.data.AUTOTUNE` lets TensorFlow **dynamically tune** the `prefetch` buffer size.
- It decides how many batches to load in advance **based on available system resources**:
  - CPU memory
  - I/O bandwidth
  - Hardware configuration

#### ✅ Benefits:
- **Reduced idle time** for training loops.
- **Automatic performance tuning** without manual buffer size tweaking.
- **Seamless and faster training**, especially for large datasets or heavy transformations.

---

These innovations not only made **AlexNet** a breakthrough in deep learning but also laid the groundwork for modern neural networks in terms of **performance, efficiency, and scalability**.

```python

# Set path to the dataset
data_dir = '/content/dogs_vs_cats/train'

# Hyperparameters
img_width, img_height = 224, 224  # AlexNet input size
batch_size = 32
epochs = 5
validation_split = 0.2  # 20% of data for validation
```

```python
# Load datasets
train_dataset = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=validation_split,
    subset="training",
    seed=123,
    image_size=(img_width, img_height),
    batch_size=batch_size
)
```

```python
validation_dataset = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=validation_split,
    subset="validation",
    seed=123,
    image_size=(img_width, img_height),
    batch_size=batch_size
)
```

```python
# Data Augmentation layer
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip('horizontal'),
    tf.keras.layers.RandomRotation(0.2),
    tf.keras.layers.RandomZoom(0.2),
])
```

```python
# Apply data augmentation only on the training dataset
train_dataset = train_dataset.map(lambda x, y: (data_augmentation(x, training=True), y))
```

32 images - Pass to model trainign.
load next batch of 32 images and - Pass to model traing.

```python
# Prefetch the datasets for performance
AUTOTUNE = tf.data.AUTOTUNE # Performance of resrouces
train_dataset = train_dataset.prefetch(buffer_size=AUTOTUNE)
validation_dataset = validation_dataset.prefetch(buffer_size=AUTOTUNE)
```

Before diving into the specifics of AUTOTUNE, it's important to understand the role of data pipelines:

Data Loading: The process of reading data from storage (e.g., disk).
Preprocessing: Operations like resizing, normalization, augmentation, etc., that are applied to the data before feeding it to the model.
Batching: Grouping individual examples into batches for efficient processing.
Prefetching: Loading the next batch of data while the current one is being processed by the model, to ensure that the GPU/CPU is always busy.
A well-optimized data pipeline ensures that the model training process is not waiting for data, which would slow down the training.

# Building Alexnet Model

```python
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense, Dropout

# AlexNet model
alexnet_model = Sequential([
    Input(shape=(224, 224, 3)),

    Conv2D(96, (11, 11), strides=(4, 4), activation='relu'),
    MaxPooling2D((3, 3), strides=(2, 2)),

    Conv2D(256, (5, 5), padding='same', activation='relu'),
    MaxPooling2D((3, 3), strides=(2, 2)),

    Conv2D(384, (3, 3), padding='same', activation='relu'),
    Conv2D(384, (3, 3), padding='same', activation='relu'),
    Conv2D(256, (3, 3), padding='same', activation='relu'),

    MaxPooling2D((3, 3), strides=(2, 2)),

    Flatten(),

    Dense(4096, activation='relu'),
    Dropout(0.5),
    Dense(4096, activation='relu'),
    Dropout(0.5),

    Dense(1, activation='sigmoid')
])
```

```python
alexnet_model.summary()
```

# Model Compile

```python
# Compile the model
alexnet_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

# Model Training

```python
history = alexnet_model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=1
)
```

```python
# Evaluate the model on validation data
val_loss, val_accuracy = alexnet_model.evaluate(validation_dataset)
print(f'Validation Accuracy: {val_accuracy * 100:.2f}%')
```
