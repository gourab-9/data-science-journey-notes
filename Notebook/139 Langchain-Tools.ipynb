---

## What Can LLMs Actually Do?

To appreciate why tools exist, you first need to be honest about what LLMs can and can't do.

Modern LLMs are genuinely impressive at two things:

**Thinking** — given a question or a problem, an LLM can reason through it, break it down logically, consider different angles, and arrive at a well-structured answer. This is the "brain" part.

**Speaking** — once the LLM has figured out what to say, it generates the response fluently, word by word. This is the "mouth" part.

That combination — think + speak — is what makes LLMs feel so capable in conversation. You can ask them almost anything and get a coherent, detailed response.

---

## But Here's the Problem

As powerful as LLMs are at thinking and speaking, they have a hard boundary: **they cannot take action in the real world.**

Let's make this concrete with some examples:

- You ask an LLM *"book me a train ticket to Mumbai for Friday"* → it will explain how to do it, suggest websites, maybe even write the steps — but it cannot actually go and book the ticket
- You ask *"what's the weather like in Bangalore right now?"* → it will tell you it doesn't have real-time data and suggest you check a weather app
- You ask *"run this Python script and tell me the output"* → it will predict what the output might be, but it cannot actually execute the code
- You ask *"send an email to my team about tomorrow's meeting"* → it can draft a beautiful email, but it cannot hit send
- You ask *"what's the current stock price of Infosys?"* → same problem, no real-time data access

See the pattern? LLMs are brilliant advisors. They can tell you *what* to do, *how* to do it, and even *why* — but they cannot *do* it themselves.

> 💡 The best analogy here is a person with a brilliant mind and excellent communication skills, but no hands and no legs. They can think through any problem and explain the solution perfectly. But when it comes to actually executing something in the real world — booking, searching, calculating, sending — they're stuck. They need someone or something else to do the physical work for them.

This is the fundamental limitation of LLMs, and it's exactly the gap that **tools** are designed to fill.

---

## So What is a Tool?

A tool is the mechanism that gives an LLM the ability to **act** — not just think and speak.

In the simplest possible terms: **a tool is just a Python function**, wrapped and packaged in a specific way so that an LLM can understand what it does and call it when needed.
```python
def book_train_ticket(source: str, destination: str, date: str) -> str:
    """Books a train ticket from source to destination on the given date."""
    # logic that hits the railway booking API
    ...
    return "Ticket booked successfully! PNR: 4521839"
```

That function above? That's the core of a tool. The "tool" part is just the packaging around it — a name, a description written in plain English, and a schema that tells the LLM exactly what inputs the function needs and what it returns.

Once you package a function as a tool and connect it to an LLM:

1. The LLM reads the tool's description and understands what it is capable of doing
2. When a user sends a message that requires that capability, the LLM recognizes it
3. The LLM decides to call the tool and figures out the correct inputs from the conversation
4. The tool function executes — it does the actual real-world work
5. The result comes back to the LLM
6. The LLM uses that result to form a natural language response for the user

The user never sees any of this machinery. From their perspective, they asked for something and it got done.

---

## A Full Example — Ticket Booking

Let's walk through the ticket booking example end to end to make this feel real.

**Without a tool:**
```
User:  Book me a train ticket from Delhi to Mumbai on Friday.
LLM:   Sure! Here's how you can book a train ticket:
       1. Go to IRCTC website
       2. Create an account if you don't have one
       3. Search for trains between Delhi and Mumbai...
```
Helpful? Sure. But the user wanted the ticket booked, not a tutorial.

**With a tool:**
```
User:  Book me a train ticket from Delhi to Mumbai on Friday.
LLM:   [internally: I have a book_train_ticket tool, let me use it]
       [calls book_train_ticket(source="Delhi", destination="Mumbai", date="Friday")]
       [tool returns: "Ticket booked! PNR: 4521839, Train: Rajdhani Express, Seat: 34B"]
LLM:   Done! Your ticket has been booked on Rajdhani Express.
       PNR: 4521839, Seat: 34B. Have a great journey! 🚂
```

Same LLM. Same intelligence. But now it can *act* — because it has a tool.

---

## The More Tools, The More Powerful

There's really no ceiling here. Anything a computer can do can be wrapped as a tool:

- **Web search** — let the LLM look up current information
- **Code execution** — let it run Python and return real outputs
- **Database queries** — let it fetch or write data
- **Email/calendar** — let it schedule meetings and send messages
- **Weather/stock APIs** — let it pull real-time data
- **File operations** — let it read and write files

The more tools you give an LLM, the wider its capabilities become. Connect enough tools and you have something that can genuinely handle complex, multi-step real-world tasks on its own — which is exactly what we call an **agent**.

---

## Simple Definition to Remember

> 📌 *A tool is a function packaged in a way that an LLM can understand and call when required. The LLM decides when to use it, which inputs to pass, and what to do with the result.*

# Creating a Custom Tool in LangChain (Simplest Method)

Let us now see **how to create your own tool in LangChain**.

We will start with a **very simple example**, but it will make the entire concept clear.

LangChain provides **multiple ways** to create custom tools.  
The method discussed here is:
- The **most simple**
- The **most straightforward**

Going forward, other approaches will also be covered.

---

## Tool Creation: High-Level Overview

Creating a custom tool in LangChain is a **three-step process**.

Before that, remember:
- A tool is essentially a function
- But a **special function** that can communicate with an LLM

---

## Step 1: Create a Function (Tool Logic)

First, you need to create a **normal function** for your tool.

### Example Use Case

Assume:
- The LLM is not good at mathematical operations
- Specifically, it cannot reliably perform multiplication

So we:
- Create an **external function**
- This function takes two numbers
- Returns their product

This function is written just like a **regular Python function**.

### Important: Docstring

Inside this function, we add a **docstring** that explains:
- What the function does

Although:
- The docstring is not mandatory

It is **highly recommended**, because:
- The LLM uses this description
- To understand what the tool does

So:
- The docstring becomes the **tool description**
- It directly helps the LLM during tool selection

---

## Step 2: Add Type Hinting

Next, you add **type hints** to the function.

This means:
- Specify the data type of each input
- Specify the return type of the function

Example idea:
- Inputs `A` and `B` are integers
- Output is also an integer

Type hinting is:
- Not mandatory
- But strongly recommended

Why?

Because:
- It tells the LLM what type of input is expected
- It tells the LLM what type of output it will receive

This improves:
- Reliability
- Correct tool usage

---

## Step 3: Add the Tool Decorator

This is the **final and most important step**.

You add a **tool decorator** on top of the function.

### What Does the Tool Decorator Do?

- It converts a normal function into a **LangChain tool**
- It enables the LLM to interact with this function

All the **magic happens inside this decorator**.

---

## Summary of Tool Creation

To create a custom tool, you only need to do three things:

1. Write the function logic
2. Add type hints
3. Add the tool decorator

That’s it.

Your custom tool is now ready.

---

## Using the Custom Tool

Once the tool is created:

- The **function name becomes the tool name**
- Since it is a tool, it is also **runnable**
- It supports invocation with inputs

Inputs are passed:
- As key–value pairs
- Where each key matches a function argument

The output:
- Is returned just like a normal function

Even though:
- It looks like a simple function call

It is actually:
- A **tool**
- Capable of being used by an LLM

---

## Why Is This Function Special?

This tool has additional capabilities.

Every LangChain tool exposes three important attributes:

---

## Tool Attributes

### 1. Name
- Usually the same as the function name
- Used by the LLM to identify the tool

### 2. Description
- Comes directly from the function docstring
- Explains what the tool does

### 3. Arguments
- Lists all required inputs
- Includes argument names
- Includes argument types
- Comes from type hinting

These three attributes are **always present** for any tool.

---

## Built-in Tools Follow the Same Pattern

Built-in tools in LangChain:
- Also have a name
- Also have a description
- Also have arguments

So whether a tool is:
- Built-in
- Or custom

The structure remains **exactly the same**.

---

## What the LLM Actually Sees

An important concept:

- The LLM does **not** see the function code
- The LLM does **not** see the tool logic

Instead:
- The LLM sees a **tool schema**

This schema is:
- A structured representation
- Includes:
  - Tool name
  - Description
  - Arguments
  - Required fields
  - Data types

---

## Tool Schema (Key Insight)

Whenever:
- You connect a tool to an LLM

You are actually:
- Sending the **tool schema**
- Not the function itself

This schema is:
- Human-readable
- Machine-readable
- Easy for the LLM to understand

The same applies to:
- Custom tools
- Built-in tools

---

## Final Takeaway

- Custom tools are created using simple functions
- A decorator turns them into LLM-compatible tools
- Docstrings become descriptions
- Type hints become argument definitions
- LLMs interact using tool schemas, not code

This forms the **foundation of tool calling and agent behavior** in LangChain.


````python
from langchain_core.tools import tool
````

````python
# Step 1 - create a function

def multiply(a, b):
    """Multiply two numbers"""
    return a*b
````

````python
# Step 2 - add type hints

def multiply(a: int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
````

````python
# Step 3 - add tool decorator

@tool
def multiply(a: int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
````

````python
result = multiply.invoke({"a":3, "b":5})
````

````python
print(result)
````

````python
print(multiply.name)
print(multiply.description)
print(multiply.args)
````

````python
print(multiply.args_schema.model_json_schema())
````

## Method 2 - Using StructuredTool

# Additional Ways to Create Custom Tools in LangChain

So far, we have already learned **one way** to create a custom tool in LangChain using the `@tool` decorator.  
However, this is **not the only way**.

LangChain provides **multiple approaches** for building custom tools.

In practice, you will mostly see **three main methods**:

1. Using the `@tool` decorator  
2. Using **StructuredTool with Pydantic**
3. Using the **BaseTool class**

There are a few other older approaches as well, but:
- They were used earlier
- They are not very common now

So, these **three methods** are the most important ones to understand.

In this section, we will focus on the **second method**:
**Creating custom tools using StructuredTool and Pydantic**.

---

## What Is a Structured Tool?

A **Structured Tool** in LangChain is a special type of tool where:

- The input to the tool follows a **strict structured schema**
- This schema is usually defined using a **Pydantic model**

In simple terms:
- The tool expects inputs in a **well-defined format**
- The input structure is strongly validated

---

## Why Use Structured Tools?

Let us compare this with the `@tool` decorator approach.

### `@tool` Decorator Method (Loose Validation)

- You define a function
- You optionally add type hints
- Input validation is **relatively loose**
- The constraints are not strictly enforced

This works perfectly fine in:
- Most simple use cases
- Many real-world scenarios

---

### Structured Tool Method (Strict Validation)

The Structured Tool approach is **more strict and controlled**.

Here:
- Input constraints are **strongly enforced**
- You define inputs using a **Pydantic model**
- The model specifies:
  - Required fields
  - Data types
  - Field descriptions
  - Validation rules

This makes the tool:
- More predictable
- Safer
- Better suited for production systems

---

## High-Level Flow of Structured Tool Creation

When using StructuredTool with Pydantic, the process looks like this:

1. Create a normal function (tool logic)
2. Create a Pydantic model for input validation
3. Combine both using the StructuredTool class

Unlike the decorator approach:
- Everything is **not handled inside the function**
- Tool configuration is done **explicitly and clearly**

---

## Step 1: Create the Tool Function

Just like before:
- You write a normal function
- Example: a multiplication function that takes two numbers and returns their product

Important point:
- Type hints inside the function are **optional**
- Because validation will be handled by the Pydantic model

---

## Step 2: Create a Pydantic Input Schema

Next, you create a **Pydantic model** that defines:

- All input fields
- Their data types
- Whether they are required
- Their descriptions

For example:
- Two inputs: `A` and `B`
- Both are integers
- Both are required
- Each has a clear description

This Pydantic model becomes:
- The **single source of truth** for tool inputs

---

## Step 3: Create the Structured Tool

Now comes the most important part.

You create the tool using a **factory-style method**, where you explicitly provide:

- The function to be used
- The tool name
- The tool description
- The argument schema (Pydantic model)

At this stage, you are clearly defining:
- Which function powers the tool
- How the tool should be described to the LLM
- What exact input structure the tool accepts

This separation makes the design:
- Cleaner
- More readable
- Easier to maintain

---

## Key Difference from `@tool` Decorator

The **core difference** is:

- With `@tool`, everything is inferred from the function
- With StructuredTool, everything is **explicitly defined**

In StructuredTool:
- Input constraints are enforced more strictly
- Schema is richer and more descriptive
- Validation errors are clearer

---

## Tool Behavior Remains the Same

Even though the creation process is different:

- The tool can still be invoked the same way
- It still has:
  - Name
  - Description
  - Arguments
- It still exposes a schema to the LLM

You may also notice:
- Additional metadata such as `required = true`
- Field-level descriptions
- Stronger schema enforcement

All of this comes directly from the **Pydantic model**.

---

## When Should You Use Structured Tools?

Structured tools are especially useful when:

- You are building **production-ready agents**
- Input validation is critical
- You want strict control over tool usage
- You want fewer runtime errors
- You want predictable LLM behavior

---

## When Is the `@tool` Decorator Enough?

In most cases:
- The `@tool` decorator method is sufficient
- It is simpler and faster to implement
- It covers a large number of real-world use cases

That is why:
- The decorator-based approach is more commonly used
- Structured tools are chosen when **extra strictness** is required

---

## Final Takeaway

- LangChain supports multiple ways to create custom tools
- StructuredTool + Pydantic provides:
  - Strong input validation
  - Explicit schemas
  - Better production readiness
- The main difference is **how strictly inputs are enforced**
- Functionality remains the same from the LLM’s perspective

That is the **only real difference** between the two approaches.


````python
from langchain.tools import StructuredTool
from pydantic import BaseModel, Field
````

````python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
````

````python
def multiply_func(a: int, b: int) -> int:
    return a * b
````

````python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
````

````python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
````

## Method 3 - Using BaseTool Class

# Third Way to Create Custom Tools in LangChain: Using BaseTool Class

Now let us discuss the **third method** of creating custom tools in LangChain.  
This method is based on the **BaseTool class**.

To understand this approach properly, we first need to understand **what the BaseTool class actually is**.

---

## What Is BaseTool?

**BaseTool** is the **abstract base class** for all tools in LangChain.

It defines:
- The **core structure**
- The **standard interface**

that **every tool in LangChain must follow**, whether:
- It is a very simple one-line tool, or
- It is a fully customized and complex tool

In simple terms:
- BaseTool decides **how a tool should behave inside LangChain**

---

## Relationship Between BaseTool and Other Tool Types

All other tool types in LangChain are built **on top of BaseTool**, such as:
- Tools created using the `@tool` decorator
- Structured tools created using `StructuredTool`

Even:
- Built-in tools
- Custom tools

All of them **internally inherit from the BaseTool class**.

This means:
- No matter how a tool is created
- Eventually, it follows the rules defined by BaseTool

---

## Why BaseTool Is Important

Since BaseTool is an abstract class:
- It enforces a **common behavior**
- It ensures consistency across all tools

This guarantees that:
- LangChain knows how to invoke tools
- LangChain knows how to pass inputs and receive outputs
- LangChain knows how tools interact with LLMs

Without BaseTool:
- Tools would behave inconsistently
- Agent–tool interaction would break

---

## What We Are Doing in This Third Approach

In the first two approaches:
- We indirectly used BaseTool
- LangChain handled inheritance automatically for us

Now, in the **third approach**:
- We **directly inherit** from the BaseTool class
- We manually define our custom tool

So instead of:
- Using decorators
- Using helper classes

We:
- Take full control
- Build the tool from the ground up using BaseTool

---

## Key Idea of This Method

The core idea is simple:

- BaseTool already defines how a tool should behave
- Every tool must follow that structure
- So we directly extend BaseTool and create our own tool

This gives:
- Maximum flexibility
- Full customization
- Complete control over tool behavior

---

## Summary

- BaseTool is the **foundation** of all tools in LangChain
- Every tool, built-in or custom, ultimately inherits from BaseTool
- `@tool` and `StructuredTool` are just **convenience layers** built on top of it
- In the third method, we:
  - Directly inherit from BaseTool
  - Create a fully custom tool
- This approach is useful when:
  - You need advanced customization
  - You want full control over tool behavior

This is why BaseTool is considered the **most fundamental and powerful** way to create tools in LangChain.


````python
from langchain.tools import BaseTool
from typing import Type
````

````python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
````

````python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
````

````python
multiply_tool = MultiplyTool()
````

# Toolkits in LangChain

In this video, the **last topic** we cover is **Toolkits**.

---

## What Is a Toolkit?

A **toolkit** is used when you have **multiple tools that are related to each other** and serve a **common purpose**.

Instead of managing each tool separately, you can **club related tools together** and treat them as a single unit called a **Toolkit**.

### Simple Idea
- Multiple related tools  
- Grouped together  
- For convenience and reuse  

---

## Example to Understand Toolkits

Imagine you are building tools related to **Google Drive**:

- Tool to upload a file to Google Drive  
- Tool to search files in Google Drive  
- Tool to read a file from Google Drive  

All these tools:
- Are different
- But belong to the **same domain (Google Drive)**

So instead of handling them separately, you can group them into a:
**Google Drive Toolkit**

---

## Definition of Toolkit

A **Toolkit** is:
> A collection of related tools that serve a common purpose, packaged together for convenience and reusability.

---

## Why Toolkits Are Useful

The **biggest benefit** of toolkits is **reusability**.

- You create a toolkit once  
- You can reuse it across:
  - Multiple agents
  - Multiple applications
  - Multiple projects  

This avoids duplication and keeps your system organized.

---

## Built-in and Custom Toolkits

In LangChain:
- You already get **built-in toolkits**
- You can also create **custom toolkits** as per your use case

So you have flexibility:
- Use ready-made toolkits
- Or define your own

---

## How a Custom Toolkit Is Structured (Conceptually)

To create a custom toolkit:

1. First, create **two or more related tools**
   - Example:
     - Add tool
     - Multiply tool  
   - Both belong to **arithmetic operations**

2. Create a **Toolkit class**
   - The class name becomes the toolkit name  
   - Example: `MathToolkit`

3. Inside the class, define a method called:
   - `get_tools`

4. This method simply returns:
   - The list of tools that belong to the toolkit

That’s it.

---

## How Toolkits Are Used

Once the toolkit is created:
- You create an object of the toolkit
- Call `get_tools`
- You receive all tools belonging to that toolkit

Then you can:
- Loop through tools
- Access each tool’s:
  - Name
  - Description
  - Arguments
- Run or inspect them as needed

All tools are now available **together**, instead of being managed one by one.

---

## Key Advantages of Toolkits

- Cleaner organization of tools
- Easy access to related tools
- High reusability across projects
- Better structure for large agent-based systems

````python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
````

````python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
````

````python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
````
