## 2. What is a Workflow?

### Definition
> A **workflow** is a clearly defined sequence of tasks and actions arranged in a logical order to achieve a specific outcome.

In simpler terms: **a workflow is a series of steps you follow in a set order to accomplish a goal.**

### Key Characteristics of a Workflow
- Defines **what happens first**
- Specifies **what comes next** after each step
- Sets **conditions** under which each step is executed

### Real-World Examples

| Context | Workflow Example |
|---|---|
| Content Creation | Draft → Review → Edit → Approve → Publish |
| App Development | Requirements → Design → Build → Test → Deploy |
| Everyday Life | Making chapati: Sift flour → Knead dough → Shape → Cook |

### Example: Content Approval Workflow
1. Content creator submits a draft to the editor
2. Editor adds comments and returns the draft
3. Writer revises and resubmits the final draft
4. Draft goes for approval
   - ✅ **Approved** → Content is published
   - ❌ **Rejected** → Writer edits again → Returns for review

## 3. What is Business Automation?

### Definition
> **Business Automation** means automating your workflows using technology — so that software or digital tools execute your workflow steps *on your behalf*.

### How it Works
- You define the workflow (the steps, rules, and conditions)
- A digital platform or tool runs that workflow automatically
- The human worker is freed from repetitive, manual tasks

### Benefits of Business Automation

| Before Automation | After Automation |
|---|---|
| Manual, repetitive task execution | Automated execution by digital tools |
| Human time spent on boring tasks | Human time freed for high-value work |
| Prone to errors and delays | Consistent, fast, and reliable |

**High-value work humans can now focus on:**
- Strategy formulation
- Complex problem solving
- Decision making
- Idea generation

### Everyday Example
When you place an order on Amazon:
- An order confirmation is automatically sent to your email
- A WhatsApp notification is triggered
- All of this happens without any manual effort from Amazon's team

This is business automation — the entire workflow is automated end-to-end.

## 4. What is AI in Automation?

### Definition
> **AI in Automation** means adding *intelligence* to your workflow — so the workflow itself can decide which path to follow, which decisions to execute, and how to respond to different inputs — all on its own.

Traditional business automation follows fixed, rule-based paths. AI in Automation introduces **dynamic decision-making** powered by Large Language Models (LLMs).

### Example: Intelligent Email Routing System

**Problem Setup:**
- A company receives all customer queries on a single email address
- Queries can be: Billing-related, Technical, or Feedback/Suggestions
- Three separate departments handle each type

**Without AI:**
- A human agent manually reads every email
- Manually routes each email to the correct department
- Time-consuming, error-prone, and not scalable

**With AI in Automation:**
1. Email arrives at the company inbox
2. An **LLM** (e.g., Gemini, ChatGPT, Claude) reads the email
3. The LLM classifies the email based on its content (trained via system prompts)
4. The email is automatically routed to the correct department


```mermaid
graph TD
    A[📧 Email Received] --> B[🤖 LLM Classifies Email]
    B --> C[Billing Query]
    B --> D[Tech Query]
    B --> E[Feedback & Suggestions]
    C --> F[Full Billing Workflow Runs]
    D --> G[Full Tech Workflow Runs]
    E --> H[Full Feedback Workflow Runs]
```



**Result:** The entire routing and handling process is automated with zero human intervention at the classification step.

## 3 Make.com — Key Terminology

Before building workflows in Make.com, it is essential to understand the core terms and concepts used within the platform. These terms will appear repeatedly during hands-on work.

---

### 3.1 Scenario

> A **Scenario** is simply Make.com's name for a **Workflow**.

The complete flow from start to finish — every step, every connection — is called a **Scenario** in Make.com terminology.

**Example:**
```
Email Received → Extract Data → Save to Google Drive → Send Acknowledgement Email
```
This entire flow = **One Scenario**

| General Term | Make.com Term |
|---|---|
| Workflow | Scenario |
| Pipeline / Process | Scenario |

---

### 3.2 Module

> A **Module** is a single step or individual task inside a Scenario.

A scenario is made up of multiple modules connected together. Each module performs one specific task.

**Example — Same 3-step workflow:**

```mermaid
graph LR
    A[📧 Module 1\nFetch Email Data] --> B[💾 Module 2\nSave to Google Drive] --> C[📤 Module 3\nSend Acknowledgement Email]
```

| Module | Task Performed |
|---|---|
| Module 1 | Fetch/read incoming email |
| Module 2 | Save email data to Google Drive |
| Module 3 | Send acknowledgement email back to customer |

> **One Scenario = Multiple Modules**
> **One Module = One Individual Task/Step**

---

### 3.3 Bundle

> A **Bundle** is one complete unit of data that passes through a Scenario.

Think of a bundle as a **single record or item** being processed by your workflow — regardless of how much data is inside it.

**Examples:**

| Data Source | One Bundle Equals |
|---|---|
| Gmail | One single email (can contain text, images, PDFs, attachments — all together = 1 bundle) |
| Google Sheets | One single row (regardless of how many columns it has) |
| API Response | One single response object |

**Practical Example:**
- One email arrives containing: plain text + an image + a PDF
- All of that together = **1 Bundle**
- That 1 bundle flows through all modules of the scenario

---

### 3.4 Operation

> An **Operation** is the task performed on **one bundle** by **one module**.

This is the most important concept to understand because **Make.com billing is based on Operations consumed.**

**Formula:**
```
1 Bundle processed by 1 Module = 1 Operation
```

**Worked Example — 1 Email arrives:**

```mermaid
graph LR
    A[📧 Fetch Email\n1 Operation] --> B[💾 Save to Drive\n1 Operation] --> C[📤 Send Email\n1 Operation]
```
- 1 email → 1 bundle → processed through 3 modules = **3 Operations total**

**Worked Example — 10 Emails arrive:**

| Module | Bundles Processed | Operations |
|---|---|---|
| Fetch Email | 10 emails | 10 |
| Save to Drive | 10 records | 10 |
| Send Acknowledgement | 10 emails | 10 |
| **Total** | | **30 Operations** |

> **Important Note:**
> - Normal workflows → **1 bundle × 1 module = 1 operation**
> - Agentic workflows → **1 bundle × 1 module can cost MORE than 1 operation** (due to the complexity of AI decision-making steps)

---

**Credit Rules:**

| Scenario Type | Cost |

### 5.7 The Concept of a Trigger — Most Important Rule

> **The first module in every scenario MUST be a Trigger.**

#### What is a Trigger?
A **Trigger** is a stimulus or event that **starts the entire scenario**. Without a trigger, the workflow has no starting point and will not run automatically.

**How triggers work:**
- The first module watches for a specific event
- When that event occurs, the first module fires
- This causes the entire chain of modules to execute in sequence

**Example:**
- First module: **Gmail → "Watch Emails"** (triggers when a new email arrives in inbox)
- As soon as an email arrives → Module 1 fires → Module 2 runs → Module 3 runs → and so on

### 6.2 The Business Problem — Company ABC

**Situation:**
- Company ABC handles all customer communication through a **single official email address**

  - 💰 Billing-related queries

| Step | What Happens |

| 7 | If Feedback → Feedback is **noted** and acknowledgement email is sent |

**Problems with the Current Approach:**
- Entirely manual — error-prone and slow
- Not scalable as query volume grows
- Multiple bottlenecks at every step
- Human effort wasted on repetitive, low-value tasks

## 8. Flow Control in Make.com

After the LLM classifies the email, the scenario needs to **split into different paths** based on the classification result. This is handled by a special category of modules in Make.com called **Flow Control**.

---

### 8.1 What is Flow Control?

> **Flow Control** is a special built-in module in Make.com that acts as the **traffic police of your scenario.**

Unlike other modules (Gmail, Google Sheets, Gemini) which interact with external applications, the **Flow Control module does not connect to any external app**. Its only job is to **control the direction and behavior of data flow** within your scenario.

To access it: search for **"Flow Control"** in the module search panel.

---

### 8.2 The Four Types of Flow Control

Make.com provides **four flow control options**, each serving a different purpose:

---

#### 8.2.1 Router — Branching

> Use when you need to **treat a single input in different ways based on conditions.**

The Router splits your scenario into **multiple branches**, and each branch executes based on a defined condition.

**Current scenario use case:**
- If Gemini output = `feedback` → follow the Feedback path (no ticket)
- If Gemini output = `billing` or `product` → follow the Non-Feedback path (generate ticket)

- This prevents the scenario from failing on a single transient error

## 13. Error Handling in Make.com

---

### 13.1 Why Error Handling Matters

When scenarios run in production, things can go wrong. Common errors include:

| Error Type | Example |
|---|---|
| API error | A module returns a `401 Unauthorized` error |
| Missing record | A customer record does not exist in the database |
| Expired authorization | The connection token for an app expires mid-run |
| Corrupted data | A bundle contains malformed or incomplete data |

Without error handling, a single error stops the entire scenario. Make.com provides **five error handlers** to control exactly what happens when something goes wrong.

> **Important:** Error handlers are applied at the **module level**, not the scenario level. Right-click any module → **Add Error Handler** to attach one.

---

### 13.2 The Five Error Handlers

---

#### 13.2.1 Ignore

> **Skips the problematic bundle and continues processing all remaining bundles.**

- The module encounters an error on a specific bundle
- That bundle is **skipped entirely**
- All other bundles continue processing normally
- The scenario never stops

**Example:**
- 100 emails arrive for processing
- Email #25 has corrupted data
- With **Ignore** applied: emails 1–24 process fine → email 25 is skipped → emails 26–100 process fine

**Best for:** Non-critical workflows where losing one record occasionally is acceptable.

---

#### 13.2.2 Resume

> **Replaces the missing/broken data with a predefined default value and continues the scenario.**

- When a bundle causes an error, instead of stopping or skipping, Resume **substitutes default values** for the missing information
- The scenario continues with that default data as if no error occurred

**Example:**
- Scraping product data from a website
- Some products have no reviews → the "Summarize Reviews" module would fail
- With **Resume** applied: a default value like `"No reviews available"` is used in place of the missing review data
- The scenario continues without interruption

**How to configure:**
- When adding the Resume handler, you must define a **default value for every output field** of that module
- Make.com will use these defaults whenever the actual data is missing or causes an error

**Best for:** Workflows where you want to guarantee completion even with incomplete data.

---

#### 13.2.3 Break

> **Stops the scenario but does NOT end it — keeps it in a paused state for manual resolution.**

- When an error occurs on a bundle, that bundle's processing **halts at the error point**
- All subsequent bundles after the errored one go into a **pending state**
- The scenario remains open and paused — not terminated
- Once you manually fix the error (or it resolves automatically), the scenario **resumes from exactly where it stopped**

**Example:**
- A 100-module scenario with 100 bundles
- Bundle #36 hits an error at Module 55
- With **Break** applied: Bundle 36 pauses at Module 55 → Bundles 37–100 enter pending state → Once error is resolved, all pending bundles resume processing

**Break Configuration Options:**

| Setting | Description |
|---|---|
| Number of retries | How many times to retry the failing module (e.g., 3 times) |
| Retry interval | Time gap between retries (e.g., every 15 minutes) |
| Auto-complete after retries | If retries are exhausted, auto-complete the scenario (Yes/No) |

> If **Auto-complete = No**, you must manually intervene to resolve the error before the scenario can continue.

**Best for:** Critical workflows where no data should be lost and human oversight is required for errors.

---

#### 13.2.4 Commit

> **Stops AND permanently ends the scenario, committing all changes made before the error.**

- Similar to a **database transaction commit**
- When an error occurs, the scenario stops and terminates completely
- All changes made by modules **before** the error are **permanently saved**
- The errored module and everything after it are not executed

**Key difference from Break:**

| | Break | Commit |
|---|---|---|
| Stops scenario | ✅ Yes | ✅ Yes |
| Ends scenario | ❌ No | ✅ Yes |
| Data before error | Preserved | Permanently committed |
| Can resume | ✅ Yes | ❌ No |

**Special behaviour with database apps (MySQL, Data Store, etc.):**
- All changes in modules before the error are **committed (permanently written)** to the database
- This mirrors SQL transaction behaviour

**Configuration:** No additional configuration needed — simply select Commit and save.

**Best for:** Scenarios involving database writes where partial completion is acceptable and you want clean transaction boundaries.

---

#### 13.2.5 Rollback

> **Reverts all changes made by previous modules when an error occurs.**

- When an error is detected, Make.com goes back through all previously executed modules and **undoes their changes**
- Condition: the previous modules must **support rollback** (i.e., their actions can be reversed)

**Important limitation:**
- Some actions **cannot be rolled back** — for example, a sent email cannot be unsent
- Rollback only works for modules/apps that support reversible operations (e.g., database writes, file creation)

**Best for:** Scenarios where all-or-nothing execution is critical — either everything succeeds or nothing is saved.

---

### 13.3 Error Handler Comparison Summary

| Error Handler | Stops Scenario | Ends Scenario | What Happens to Errored Bundle | Best Use Case |
|---|---|---|---|---|
| **Ignore** | ❌ No | ❌ No | Skipped entirely | Non-critical data, bulk processing |
| **Resume** | ❌ No | ❌ No | Replaced with default value | Must-complete workflows with optional data |
| **Break** | ✅ Yes | ❌ No | Paused, pending resolution | Critical workflows needing human oversight |
| **Commit** | ✅ Yes | ✅ Yes | Error point and beyond not executed | Database transactions, partial completion ok |
| **Rollback** | ✅ Yes | ✅ Yes | All previous changes undone | All-or-nothing critical operations |

## 14. Introduction to AI Agents in Make.com

---

### 14.1 What is an AI Agent?

> An **AI Agent** is a smart software assistant that understands a goal, figures out on its own what steps need to be taken to achieve it, and executes those steps — with minimal human involvement.

In simple terms:
- You give the agent a **goal**
- The agent **thinks** about what needs to be done
- The agent **acts** to achieve it — choosing tools, collecting information, and executing tasks autonomously

---

### 14.2 Real-World Example — Flight Booking Agent

**Agent Name:** Alpha Flight

**Goal given:** "Book a flight ticket to Hyderabad."

**What the agent does autonomously:**

1. Understands the goal: book a flight
2. Identifies what information is needed: date, time, airline, class, payment method
3. Asks the user for these details
4. Once information is collected → calls the **Book Flight** tool
5. If the booking fails (e.g., server is down on one platform) → automatically tries another platform
6. Completes the booking without further human instruction

> The human only stated the goal — the agent handled all intermediate decisions and actions.

---

### 14.3 How AI Agents Work — Two Core Components

Every AI agent is built from two components:

<table style="border-collapse:separate; border-spacing:8px; font-family:sans-serif; font-size:13px; width:100%;">
  <tr>
    <td style="background:#6c63ff; color:white; padding:12px; border-radius:8px; text-align:center; width:50%;">

      Issues commands to tools

**How they work together:**

1. LLM interacts with the user → understands the goal
2. LLM identifies which tool is needed
3. LLM checks what inputs the tool requires
4. LLM collects those inputs from the user
5. LLM calls the tool with the collected inputs
6. Tool executes the task and returns output
7. LLM communicates the result to the user

### 14.4 Tools in Make.com — No Code Required

In Make.com, **tools are simply scenarios** — built the same way as any other workflow, but structured with:

| Component | What to Define |
|---|---|
| **Input** | What data does this tool need to run? |
| **Output** | What does this tool return when done? |
| **Flow** | How does the tool go from input to output? (the modules in between) |

> You build tools the same way you build scenarios — using drag and drop, no coding required. Make.com handles everything else.

**Example — A flight booking agent's toolset:**

| Tool | Capability |
|---|---|
| `book_flight` | Books a flight given date, time, airline, class, payment |
| `cancel_flight` | Cancels an existing booking given booking ID |
| `reschedule_flight` | Changes the date/time of an existing booking |

The LLM (brain) decides which tool to call based on the user's request. The tool (function/scenario) does the actual work.

### 15.3 Step 2 — Adding a Knowledge Base (RAG)

A Word document containing all café information is uploaded as the agent's **context/knowledge base**.

**Knowledge base contents:**
- Café address and location
- Opening and closing timings
- Holidays and weekly off days
- Dining options (dine-in, takeaway, home delivery)
- Delivery areas and delivery charges
- Minimum order value
- Full menu
- Order cancellation policy
- Customer support policy

**How to add:**
- In the agent builder → Click **Context** section → Upload the document

**Result:** The agent becomes a **RAG-based chatbot** — it answers questions by retrieving relevant information from the uploaded document rather than hallucinating.

**Test:**

| Query | Response |
|---|---|
| "What are the working hours?" | Returns exact timings from the document |
| "What are the delivery charges?" | Returns exact charges from the document |
| "In which areas can you deliver?" | Returns exact delivery zones from the document |

> **RAG in Make.com vs. coding:** Building RAG traditionally requires creating a vector database, generating embeddings, and writing retrieval logic. Make.com handles all of this automatically when you upload a document as context.

### 15.5 Step 4 — Connecting Agent to Telegram Bot

#### 15.5.1 Create a Telegram Bot (via BotFather)

1. Open **Telegram Desktop** (recommended over web version)
2. Search for **BotFather**
3. Send the command: `/newbot`
4. Give the bot a display name: e.g., `TechLearn Café`
5. Give the bot a username (must end in `bot`): e.g., `techlearncafe_bot`
6. BotFather returns an **HTTP API Token** — copy this

#### 15.5.2 Build the Make.com Scenario to Connect Telegram to the Agent

**Module 1 — Telegram Bot Trigger (Instant Trigger)**

| Setting | Value |
|---|---|
| App | Telegram Bot |
| Task | Watch Updates (triggers instantly when a new message arrives) |
| Connection | Create webhook using the BotFather API token |

> This uses an **instant trigger** (⚡ lightning icon) — the scenario fires immediately when a message arrives, without needing a schedule.

**Module 2 — Make AI Agent: Run Agent**

| Setting | Value |
|---|---|
| App | Make AI Agents |
| Task | Run an Agent |
| Agent | TechLearn Café Agent |
| Thread ID | Chat ID from Module 1 (maintains conversation history per user) |
| Message/Prompt | Text message from Module 1 (the user's Telegram message) |

**Module 3 — Telegram Bot: Send Reply**

| Setting | Value |
|---|---|
| App | Telegram Bot |
| Task | Send a Text Message (Reply) |
| Chat ID | Chat ID from Module 1 |
| Text | Agent response output from Module 2 |

### 15.6 What is a Webhook?

> A **Webhook** is a bridge that allows two different applications to communicate and share information in real time.

In this project, the webhook connects the Telegram bot to Make.com — whenever a message arrives in Telegram, it instantly notifies Make.com to start the scenario.

### 15.7 Instant Trigger vs Scheduled Trigger

| Trigger Type | Icon | How it works | Example |
|---|---|---|---|
| **Instant Trigger** | ⚡ | Fires immediately when an event occurs | Telegram message arrives → scenario runs instantly |
| **Scheduled Trigger** | 🕐 | Runs at fixed intervals (e.g., every 15 min) | Gmail inbox checked every 15 minutes |

### 15.9 Key Learnings — AI Agent Architecture Summary

| Component | Role | In Make.com |
|---|---|---|
| **LLM (Brain)** | Decides what to do, asks clarifying questions, chooses tools | Built-in model selection in AI Agent builder |
| **Knowledge Base** | Provides static information for RAG-based Q&A | Uploaded document in Context section |
| **Tools** | Perform actual tasks (order taking, data storage, email) | Scenarios with defined inputs and outputs |
| **Webhook** | Connects external apps (Telegram) to Make.com in real time | Created via Telegram BotFather + Make.com connection |
| **Thread ID** | Maintains conversation history per user session | Chat ID from Telegram passed to agent |
