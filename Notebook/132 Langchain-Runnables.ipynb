## Runnables in LangChain – Conceptual Understanding

Before understanding **what Runnables are**, it is very important to understand **why Runnables exist**.  
To understand this properly, let’s go a little back in time.

---

## Background: Why LangChain Was Created

Around **November 2022**, ChatGPT was released to the public.  
At the same time, **OpenAI opened its APIs** for developers, which made it possible to build custom **LLM-based applications**.

Soon, it became clear that:
- LLMs are extremely powerful
- They can understand human language
- They can generate relevant and useful responses

The LangChain team realized that in the future:
- Most applications would **use LLMs internally**
- Chatbots, PDF readers, AI assistants, and agents would become common

And this prediction turned out to be **absolutely correct**.

Today:
- Almost every company has its own chatbot
- PDF readers allow users to ask questions
- AI agents perform tasks using LLMs

---

## The First Problem LangChain Solved

Initially, LangChain focused on **interacting with LLM APIs**.

The problem was:
- OpenAI had one API style
- Anthropic had another
- Google had a different one
- Other companies followed different patterns

This made development messy and inconsistent.

### LangChain’s First Solution
LangChain introduced **standardized interfaces** so that:
- You could switch between LLM providers
- With **minimal code changes**

This abstraction layer made LangChain popular very quickly.

---

## The Bigger Problem: LLM Is Only One Part of the Application

Over time, the LangChain team noticed something important:

> **Interacting with an LLM is only a small part of building an LLM-based application.**

Consider a **PDF Question Answering app**:

Steps involved:
1. Load a PDF (possibly from cloud storage)
2. Split it into chunks (pages, paragraphs, sections)
3. Generate embeddings for each chunk
4. Store embeddings in a vector database
5. Perform semantic search using a retriever
6. Send relevant chunks to the LLM
7. Receive the response
8. Process and display the output to the user

In this entire flow:
- Calling the LLM is **just one step**
- Many other components are equally important

---

## LangChain’s Evolution into a Full Framework

The LangChain team decided to support **every major building block** of an LLM application.

They introduced dedicated components such as:
- **Document Loaders** – for loading PDFs, text files, web data
- **Text Splitters** – for breaking large data into chunks
- **Embedding Models** – for vector representations
- **Vector Stores & Retrievers** – for semantic search
- **Output Parsers** – for structured responses
- **Memory Components** – for conversational context

Over time, LangChain became a **complete ecosystem**, not just an LLM wrapper.

---

## The Need for a Unifying Abstraction

Now another challenge appeared:

- All these components had **different input and output formats**
- Connecting them cleanly was becoming complex
- Developers needed a **consistent way to compose components**

This is where **Runnables** come into the picture.

---

## Intuition Behind Runnables

At a high level:

- A **Runnable** is any unit of work that:
  - Takes an input
  - Performs an operation
  - Produces an output

Examples of operations:
- Prompt formatting
- Calling an LLM
- Parsing output
- Running retrieval
- Transforming data

The idea was to make **every LangChain component behave consistently**, so they can be:
- Connected easily
- Composed into pipelines
- Executed sequentially, in parallel, or conditionally

---

## Why Runnables Are Important

Runnables allow:
- Clean chaining of components
- Pipeline-style execution
- Reusability
- Flexibility across different application types

Instead of manually wiring:
- Prompts
- Models
- Parsers
- Retrievers

You can treat everything as a **Runnable** and connect them smoothly.

## Key Takeaway

## What are Runnables in LangChain?

### 1. Runnable = Unit of Work
- A **Runnable** is a **single unit of work** in LangChain.
- Every runnable has **one clear purpose**.
- It:
  - Takes an **input**
  - Processes it
  - Returns an **output**

This makes each runnable small, focused, and reusable.

---

### 2. Common Interface (Most Important Idea)
- **All runnables follow the same standard interface**
- This means every runnable exposes the **same set of methods**

Key methods:
- **`invoke(input)`** → process one input and return one output
- **`batch(inputs)`** → process multiple inputs at once
- **`stream(input)`** → return output in streaming mode

Because method names and behavior are consistent, all runnables behave predictably.

---

### 3. Runnables Can Be Connected Seamlessly
- Since all runnables follow the same interface:
  - You can **connect runnables together**
- Output of one runnable automatically becomes:
  - Input of the next runnable

Example flow:
R1 → R2 → R3

- Output of `R1` → input of `R2`
- Output of `R2` → input of `R3`

This allows building **complex workflows** easily.

---

### 4. A Workflow is Also a Runnable (Composable Design)
- When you connect multiple runnables:
  - The **entire workflow itself becomes a runnable**
- This means:
  - Small workflows can be combined into bigger workflows
  - No limit to complexity

Example:
Workflow A (R1 → R2 → R3)
Workflow B (R4 → R5)

Workflow A → Workflow B


Everything remains composable and reusable.

---

### 5. Lego Blocks Analogy (Perfect Mental Model)
Runnables work exactly like **LEGO blocks**:

#### (a) Unit of Work
- Each LEGO block has a specific shape and purpose
- Each runnable has a specific task

#### (b) Common Interface
- All LEGO blocks have compatible connectors
- All runnables share the same interface (`invoke`, `batch`, `stream`)

#### (c) Easy Connection
- LEGO blocks can snap together
- Runnables can connect output → input

#### (d) Structures Are Also Blocks
- A structure made from LEGO blocks can connect to others
- A workflow made from runnables is itself a runnable

---

### 6. Core Benefit of Runnables
- **Standardization**
- **Seamless composition**
- **No custom glue code**
- **Infinite scalability of workflows**

> Runnables solve the exact problem caused by non-standard components and chain explosion by enforcing a common interface and true composability.

## Problem with Traditional Chains

Earlier, we explored the perspective where the LangChain team thought of creating multiple chain functions to simplify AI engineers’ workflows. However, they realized that approach wasn’t ideal. The chain classes they had created were **not flexible**.

For example, consider a **two-step chain**: first, generate a joke, then generate an explanation for that joke. This workflow requires **two separate LLM calls**, but the existing chain structure cannot handle multiple dynamic calls efficiently. Even if you tried to write code inside `run` to make it flexible, implementing a solution to support an arbitrary number of LLM calls would be extremely difficult.  

This limitation shows why the simple LLM chain we saw earlier **cannot support flexible workflows**. The main issue comes from the way the **PromptTemplate class** interacts via `format` and the **LLM class** interacts via `predict`. Without a standard interface between components, creating flexible chains is challenging.

### Solution Preview: Standardized Components with Runnables

To solve this, we need to **standardize components** using **Runnables**. Every runnable follows a **common interface**, allowing it to process input and return output in a predictable way. Once we have standardized components, we can **chain them together** to create flexible and reusable workflows.  

Before building these chains, we first need these **standardized components** ready. Using Runnables, we can then construct complex workflows that were previously hard to implement with traditional LLM chains.


````python
from abc import ABC, abstractmethod
class Runnable(ABC):

  @abstractmethod
  def invoke(input_data):
    pass

import random

class NakliLLM(Runnable):

  def __init__(self):
    print('LLM created')

  def invoke(self, prompt):
    response_list = [
        'Delhi is the capital of India',
        'IPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}


  def predict(self, prompt):

    response_list = [
        'Delhi is the capital of India',
        'IPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}


class NakliPromptTemplate(Runnable):

  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def invoke(self, input_dict):
    return self.template.format(**input_dict)

  def format(self, input_dict):
    return self.template.format(**input_dict)

class NakliStrOutputParser(Runnable):

  def __init__(self):
    pass

  def invoke(self, input_data):
    return input_data['response']

class RunnableConnector(Runnable):

  def __init__(self, runnable_list):
    self.runnable_list = runnable_list

  def invoke(self, input_data):

    for runnable in self.runnable_list:
      input_data = runnable.invoke(input_data)

    return input_data

template = NakliPromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length', 'topic']
)


llm = NakliLLM()

parser = NakliStrOutputParser()

chain = RunnableConnector([template, llm, parser])

chain.invoke({'length':'long', 'topic':'india'})

template1 = NakliPromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)

template2 = NakliPromptTemplate(
    template='Explain the following joke {response}',
    input_variables=['response']
)

llm = NakliLLM()

parser = NakliStrOutputParser()

chain1 = RunnableConnector([template1, llm])

chain2 = RunnableConnector([template2, llm, parser])

final_chain = RunnableConnector([chain1, chain2])

final_chain.invoke({'topic':'cricket'})
````

````python
from abc import ABC, abstractmethod
import random

# -------------------------------
# Base class for all runnables
# -------------------------------
class Runnable(ABC):
    @abstractmethod
    def invoke(self, input_data):
        pass
````

````python
# -------------------------------
# Mock LLM class
# -------------------------------
class CustomLLM(Runnable):
    def __init__(self):
        print('Custom LLM initialized')

    def invoke(self, prompt):
        responses = [
            'Paris is the capital of France',
            'NBA is a basketball league',
            'ML stands for Machine Learning'
        ]
        return {'response': random.choice(responses)}

    def predict(self, prompt):
        responses = [
            'Paris is the capital of France',
            'NBA is a basketball league',
            'ML stands for Machine Learning'
        ]
        return {'response': random.choice(responses)}
````

````python
# -------------------------------
# Prompt Template class
# -------------------------------
class CustomPromptTemplate(Runnable):
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def invoke(self, input_dict):
        return self.template.format(**input_dict)

    def format(self, input_dict):
        return self.template.format(**input_dict)
````

````python
# -------------------------------
# Output Parser class
# -------------------------------
class CustomStrOutputParser(Runnable):
    def invoke(self, input_data):
        return input_data['response']
````

````python
# -------------------------------
# Connector to chain multiple task units
# -------------------------------
class RunnableChain(Runnable):
    def __init__(self, task_list):
        self.task_list = task_list

    def invoke(self, input_data):
        for task in self.task_list:
            input_data = task.invoke(input_data)
        return input_data
````

````python
# -------------------------------
# Example: Single-step chain
# -------------------------------
poem_template = CustomPromptTemplate(
    template='Write a {length} story about {theme}',
    input_variables=['length', 'theme']
)
mock_llm = CustomLLM()
````

````python
parser = CustomStrOutputParser()
````

````python
single_chain = RunnableChain([poem_template, mock_llm, parser])
````

````python
print(single_chain.invoke({'length':'short', 'theme':'space exploration'}))
````

````python
# -------------------------------
# Example: Two-step chain (joke + explanation)
# -------------------------------
joke_template = CustomPromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)
explain_template = CustomPromptTemplate(
    template='Explain the following joke: {response}',
    input_variables=['response']
)
````

````python
mock_llm = CustomLLM()
````

````python
parser = CustomStrOutputParser()
````

````python
joke_chain = RunnableChain([joke_template, mock_llm])
````

````python
explanation_chain = RunnableChain([explain_template, mock_llm, parser])
````

````python
final_chain = RunnableChain([joke_chain, explanation_chain])
````

````python
print(final_chain.invoke({'topic':'astronomy'}))
````

## From Rigid Chains to Flexible Workflows

Initially, one approach was to create multiple **chain classes** to simplify the work of AI engineers. Over time, it became clear that this approach does not scale well. These chain classes are **not flexible enough** to handle real-world workflows.

### Why Simple Chains Are Limited

Consider a simple two-step workflow:
1. Generate a joke  
2. Generate an explanation for that joke  

This requires **two sequential calls to the LLM**. With a rigid chain abstraction, supporting multiple LLM calls inside the same workflow becomes difficult. As the number of steps increases, the design quickly becomes:
- Hard to extend  
- Difficult to reuse  
- Uncomfortable to maintain  

The core issue comes from **inconsistent interaction patterns**:
- `PromptTemplate` exposes a `build` method  
- `LLM` exposes a `predict` method  

Because each component follows a different interface, composing them into flexible pipelines is not straightforward.

---

## Standardization Using Runnables

To address this, we introduced **standardized components** using a common `Runnable` interface.

Using:
- A dummy LLM  
- Custom prompt templates  
- A string output parser  

each component was converted into a **Runnable**.  
Once standardized, all components could be executed in the same way using `invoke`.

This enables:
- Uniform execution semantics  
- Automatic data flow between components  
- Easy composition of multi-step workflows  

---

## Composing Multi-Step Chains

With the help of `RunnableConnector`, we created:

- **Chain 1:** Generates a joke  
- **Chain 2:** Takes the generated joke and produces an explanation  

Instead of manually invoking each chain, both chains were combined into a **single final chain**.

When this final chain is invoked with an input such as:
```text

```

the entire workflow executes end-to-end, and the output of the last step is returned directly.

### **Key Takeaway**

Even though the LLM used here is a dummy implementation, the workflow design is realistic.

By:

1. Creating independent components

## Runnables in LangChain

LangChain introduces the concept of **Runnables** to standardize how different components of an AI system interact with each other.  
A Runnable represents any component that can take an input, perform some operation, and return an output through a common interface.

This standardization enables flexible composition of components into complex workflows.

---

## Classification of Runnables

Runnables in LangChain can be divided into two broad categories:

1. **Task-Specific Runnables**
2. **Runnable Primitives**

Each category serves a distinct role in building AI pipelines.

---

## Task-Specific Runnables

Task-Specific Runnables are components that perform a **single, clearly defined responsibility** within an AI workflow.

These components are focused on *what* operation needs to be performed.

### Common Examples

- **Prompt Templates**  
  Responsible for constructing prompts from input variables.

- **LLMs / Chat Models**  
  Responsible for generating text responses based on prompts.

- **Retrievers**  
  Responsible for fetching relevant documents or context.

- **Output Parsers**  
  Responsible for transforming raw model outputs into structured formats.

Each of these components is implemented as a Runnable so that they all expose a **common execution interface**. This allows them to be connected seamlessly without special handling.

---

## Why Task-Specific Runnables Matter

- They encapsulate domain-specific logic.
- Each component remains reusable and independently testable.
- They simplify reasoning about individual steps in a workflow.

However, on their own, task-specific Runnables do not define how multiple steps should be combined or executed together.

---

## Runnable Primitives

Runnable Primitives are Runnables that focus on **workflow orchestration** rather than task execution.

These primitives define *how* multiple task-specific Runnables interact with each other.

### Core Responsibilities

- Connecting Runnables together
- Managing execution order
- Controlling data flow between components
- Enabling advanced workflow patterns

Runnable Primitives act as **structural building blocks** for AI pipelines.

---

## Common Runnable Primitives

LangChain provides several built-in Runnable Primitives, including:

- **RunnableSequence**  
  Executes multiple Runnables sequentially.

- **RunnableParallel**  
  Executes multiple Runnables at the same time on the same input.

- **RunnableBranch**  
  Executes different Runnables based on conditions.

- **RunnableLambda**  
  Allows custom transformation logic inside a pipeline.

- **RunnablePassthrough**  
  Forwards input data without modification.

Each of these primitives enables a specific execution pattern.

---

## RunnableSequence (Sequential Execution)

`RunnableSequence` is one of the most commonly used Runnable Primitives.

### Behavior

- Runnables are executed in a fixed order.
- The output of one Runnable becomes the input of the next.
- There is no limit on the number of Runnables in the sequence.

### Conceptual Flow

Input → Runnable A → Runnable B → Runnable C → Output


### Typical Use Case

A standard LLM pipeline often follows this structure:

1. Prompt construction
2. Model invocation
3. Output parsing

Each step is independent, yet connected through a sequence.

---

## Benefits of Using Runnable Primitives

- Clear separation between **logic** and **orchestration**
- Highly flexible pipeline construction
- Easy extension of workflows without rewriting existing components
- Improved readability and maintainability of AI systems

---

## Key Takeaway

Task-Specific Runnables define **what** operations are performed, while Runnable Primitives define **how** those operations are combined.

This separation allows LangChain to support complex, scalable, and composable AI workflows using simple, standardized components.


````python
# Runnable Sequence

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain.schema.runnable import RunnableSequence
import os


load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")


prompt1 = PromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)

model = ChatOpenAI(api_key=api_key)

parser = StrOutputParser()

prompt2 = PromptTemplate(
    template='Explain the following joke - {text}',
    input_variables=['text']
)

chain = RunnableSequence(prompt1, model, parser, prompt2, model, parser)

print(chain.invoke({'topic':'AI'}))
````

## Runnable Parallel

The next runnable primitive we explore is **Runnable Parallel**.

Just as **Runnable Sequence** helps us build **sequential chains**,  
**Runnable Parallel** helps us build **parallel execution chains**.

---

## What is Runnable Parallel?

Runnable Parallel is a runnable primitive that allows **multiple runnables to execute at the same time**.

- All runnables receive **the same input**
- Each runnable processes the input **independently**
- The final output is returned as a **dictionary**, where each key represents one runnable’s result

---

## Conceptual Example

Assume we have a topic:

**Topic = "Artificial Intelligence"**


Now, we want to build an application where:

- **Model A** generates a **tweet** about the topic  
- **Model B** generates a **LinkedIn post** about the same topic  

This setup makes practical sense because:
- One model might be optimized or trained for short-form content (Twitter)
- Another model might be optimized for professional, long-form content (LinkedIn)

Both tasks:
- Use the **same input topic**
- Are **independent of each other**
- Can be executed **simultaneously**

---

## Execution Flow

1. The same input (`"Artificial Intelligence"`) is sent to multiple runnables
2. Each runnable executes in parallel
3. Outputs are collected together in a dictionary, for example:

```python
{
  "tweet": "AI is transforming the future, one model at a time.",
  "linkedin_post": "Artificial Intelligence is reshaping industries by enabling data-driven decision-making..."
}
```

These outputs can then be:

- Posted to different platforms  
- Stored  
- Further processed or combined  

---

### **Key Characteristics of Runnable Parallel**

- Execution happens in **parallel**, not sequentially  
- All runnables receive the **same input**  
- Outputs are returned together as a **key-value structure**  
- Ideal for workflows where multiple independent tasks must run on the same data  

---

### **Formal Definition**

>Runnable Parallel is a runnable primitive that allows multiple runnables to execute in parallel.  
Each runnable receives the same input and processes it independently, producing a dictionary of outputs.

---

### **When to Use Runnable Parallel**

Runnable Parallel is useful when:

- You want multiple outputs from the same input  
- Tasks do not depend on each other’s output  
- Performance matters and tasks can be executed simultaneously  

Examples include:

- Generating content for multiple platforms  
- Running multiple analyses on the same document  
- Producing different representations (summary, keywords, sentiment) from the same text


````python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain.schema.runnable import RunnableSequence, RunnableParallel
import os 

load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")

prompt1 = PromptTemplate(
    template='Generate a tweet about {topic}',
    input_variables=['topic']
)

prompt2 = PromptTemplate(
    template='Generate a Linkedin post about {topic}',
    input_variables=['topic']
)

model = ChatOpenAI(api_key=api_key)

parser = StrOutputParser()

parallel_chain = RunnableParallel({
    'tweet': RunnableSequence(prompt1, model, parser),
    'linkedin': RunnableSequence(prompt2, model, parser)
})

result = parallel_chain.invoke({'topic':'AI'})

print(result['tweet'])
print(result['linkedin'])
````

## Runnable Passthrough

The next runnable primitive is **Runnable Passthrough**.

As the name suggests, this is a **special runnable** whose behavior is very simple:
- Whatever input you provide
- It returns **exactly the same input as output**
- No transformation, no processing, no modification

---

## What Does Runnable Passthrough Do?

Runnable Passthrough acts like an **identity function**.

Examples:
- Input: `2` → Output: `2`
- Input: `"hello"` → Output: `"hello"`
- Input: `{ "joke": "some text" }` → Output: `{ "joke": "some text" }`

It does **nothing**, and that is precisely why it is useful.

---

## Why Do We Need Runnable Passthrough?

At first glance, Runnable Passthrough may seem unnecessary.  
However, it becomes extremely useful in **branching and parallel workflows**, where we want to **preserve intermediate outputs**.

---

## Problem Scenario

Consider a sequential workflow:

1. User provides a topic
2. A joke is generated using the LLM
3. The joke is sent again to the LLM to generate an explanation

In a **pure sequential chain**, the final output will only contain:
- The **explanation**
- The **original joke is lost**, because each step replaces the previous output

But what if the requirement is:
- Show the **joke**
- AND show the **explanation**
- Both together

---

## Solution Using Runnable Passthrough

To solve this, we introduce **Runnable Passthrough** along with **parallel execution**.

---

## Conceptual Flow

### Step 1: Joke Generation Chain (Sequential)
- Input topic → Prompt → LLM → Output Parser
- Output: `joke`

At this point, we have the joke available.

---

### Step 2: Parallel Execution

From the joke, we create **two parallel paths**:

#### Path 1: Explanation Generation
- Joke → Prompt (explain joke) → LLM → Output Parser
- Output: `explanation`

#### Path 2: Passthrough
- Joke → Runnable Passthrough
- Output: `joke` (unchanged)

---

### Step 3: Final Output

Now we have:
- The original **joke**
- The **explanation of the joke**

Both are available together as outputs.

---

## Why Runnable Passthrough Is Important

Runnable Passthrough allows us to:
- Preserve intermediate values
- Reuse outputs without modification
- Design workflows where multiple results are required
- Avoid losing data in sequential chains

---

## Key Takeaways

- Runnable Passthrough returns input **as-is**
- It is especially useful in **parallel and branching workflows**
- Helps retain intermediate outputs that would otherwise be overwritten
- Often used together with `RunnableParallel` or branching logic

---

Runnable Passthrough may look simple, but it plays a crucial role in building **flexible and expressive AI workflows**.


````python
# -------------------------------
# Imports and setup
# -------------------------------
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain.schema.runnable import (
    RunnableSequence,
    RunnableParallel,
    RunnablePassthrough
)
import os


# -------------------------------
# Environment configuration
# -------------------------------
# Load environment variables (OPENAI_API_KEY)
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")


# -------------------------------
# Prompt 1: Joke generation
# -------------------------------
# This prompt takes a topic and asks the LLM to generate a joke
prompt1 = PromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)


# -------------------------------
# LLM and Output Parser
# -------------------------------
# Chat-based LLM
model = ChatOpenAI(api_key=api_key)

# Parses LLM output into a plain string
parser = StrOutputParser()


# -------------------------------
# Prompt 2: Joke explanation
# -------------------------------
# This prompt takes a joke and asks the LLM to explain it
prompt2 = PromptTemplate(
    template='Explain the following joke - {text}',
    input_variables=['text']
)


# -------------------------------
# Chain 1: Joke Generator Chain
# -------------------------------
# Sequential flow:
# Prompt (topic) → LLM → Parsed joke
joke_gen_chain = RunnableSequence(
    prompt1,
    model,
    parser
)


# -------------------------------
# Parallel Chain: Joke + Explanation
# -------------------------------
# Both paths receive the SAME input (the joke)
parallel_chain = RunnableParallel({
    # Passes the joke forward unchanged
    'joke': RunnablePassthrough(),

    # Generates an explanation for the joke
    'explanation': RunnableSequence(
        prompt2,
        model,
        parser
    )
})


# -------------------------------
# Final Chain: Complete Workflow
# -------------------------------
# Step 1: Generate a joke from the topic
# Step 2: In parallel, return the joke and its explanation
final_chain = RunnableSequence(
    joke_gen_chain,
    parallel_chain
)


# -------------------------------
# Execute the chain
# -------------------------------
# Input: topic
# Output: {
#   'joke': <generated joke>,
#   'explanation': <joke explanation>
# }
print(final_chain.invoke({'topic': 'cricket'}))
````

## Runnable Lambda

The next runnable primitive we are going to study is **Runnable Lambda**.  
This is one of the most practical and commonly used runnable primitives.

---

## What is Runnable Lambda?

**Runnable Lambda** allows you to convert **any Python function into a Runnable**.

Once a Python function is converted into a runnable:
- It follows the same standardized interface as other runnables
- It can be connected with prompt templates, LLMs, parsers, and other runnables
- It can become part of any chain or workflow

In simple terms, Runnable Lambda makes normal Python logic **chain-compatible**.

---

## Why Do We Need Runnable Lambda?

In real-world AI applications, not every task should be handled by an LLM.

Some operations are:
- More accurate in Python
- Faster to execute
- Cheaper than calling an LLM

Runnable Lambda allows us to insert such logic **directly into the chain**, instead of handling it outside the workflow.

---

## Practical Scenario: Text Preprocessing Before an LLM

Consider a scenario where:
- Customer reviews are loaded from a database
- These reviews are sent to an LLM for sentiment analysis

The problem is that the raw reviews are often noisy:
- HTML tags
- Punctuation
- Emojis and smileys
- Inconsistent casing

LLMs usually perform poorly on such unclean text.

---

## Ideal Approach

Before sending the text to the LLM, we should:
- Clean the text
- Normalize it
- Apply NLP preprocessing steps (lowercasing, punctuation removal, etc.)

This preprocessing logic is **pure Python**, not an LLM task.

---

## Where Runnable Lambda Fits

1. Write a Python function that performs preprocessing
2. Convert that function into a runnable using **Runnable Lambda**
3. Place this runnable inside the chain
4. The cleaned output flows directly into the LLM runnable

Because the function is now a runnable, preprocessing becomes an **automatic part of the workflow**.

---

## Core Idea

> Runnable Lambda allows any Python function to participate in a LangChain workflow as a first-class runnable component.

---

## Example: Joke Generation with Word Count

Let’s understand Runnable Lambda using a simple example.

---

## Requirement

- Take a topic from the user
- Generate a joke using an LLM
- Display the joke
- Also calculate the **number of words** in the joke
- The word count should **not** be generated by the LLM

Why not the LLM?
- LLMs are not reliable for exact counting
- Python is better suited for this task

---

## Workflow Design

### Step 1: Sequential Chain
- Prompt → LLM → Output Parser  
This step generates the joke.

---

### Step 2: Parallel Chain

The joke is sent to two paths simultaneously:

- **Path 1 – Runnable Passthrough**  
  Returns the joke exactly as it is.

- **Path 2 – Runnable Lambda**  
  - Receives the joke text  
  - Counts the number of words using Python logic  
  - Returns the word count

---

### Step 3: Final Output

Both outputs are combined, resulting in:
- The generated joke
- The total number of words in the joke

---

## Key Takeaways

- Runnable Lambda converts Python functions into runnables
- These runnables can be connected with any other LangChain component
- Python logic can be seamlessly embedded inside AI workflows
- This leads to more accurate, efficient, and flexible chains


````python
# -------------------------------
# Import required modules
# -------------------------------
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain.schema.runnable import RunnableSequence, RunnableLambda, RunnablePassthrough, RunnableParallel
import os 

# -------------------------------
# Load environment variables
# -------------------------------
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")

# -------------------------------
# Define a Python function for RunnableLambda
# -------------------------------
# This function calculates the number of words in a text
def word_count(text):
    return len(text.split())

# -------------------------------
# Prompt Template
# -------------------------------
# Defines the input format for the LLM
prompt = PromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)

# -------------------------------
# LLM Model
# -------------------------------
# Initialize the OpenAI Chat model
model = ChatOpenAI(api_key=api_key)

# -------------------------------
# Output Parser
# -------------------------------
# Converts LLM output to a string
parser = StrOutputParser()

# -------------------------------
# Step 1: Sequential Chain for Joke Generation
# -------------------------------
# This chain performs the sequential execution:
# Prompt -> LLM -> Parser
joke_gen_chain = RunnableSequence(prompt, model, parser)

# -------------------------------
# Step 2: Parallel Chain
# -------------------------------
# This chain takes the joke generated from joke_gen_chain and performs two operations in parallel:
# 1. Returns the joke as it is (RunnablePassthrough)
# 2. Counts the number of words in the joke (RunnableLambda)
parallel_chain = RunnableParallel({
    'joke': RunnablePassthrough(),
    'word_count': RunnableLambda(word_count)
})

# -------------------------------
# Step 3: Final Chain
# -------------------------------
# Combines the sequential joke generation chain with the parallel chain
# This ensures the joke is generated first, then processed in parallel for both outputs
final_chain = RunnableSequence(joke_gen_chain, parallel_chain)

# -------------------------------
# Step 4: Invoke the Final Chain
# -------------------------------
# Provide the topic input
result = final_chain.invoke({'topic':'AI'})

# -------------------------------
# Step 5: Format and Print Final Result
# -------------------------------
# Combines the joke and word count into a formatted string for output
final_result = """{} \n word count - {}""".format(result['joke'], result['word_count'])

print(final_result)
````

## Runnable Branch

The last Runnable primitive we will discuss is **RunnableBranch**.  
This primitive is used to create **conditional chains**, similar to an `if-else` statement in programming.

---

### Concept

- **Purpose:** RunnableBranch allows you to define multiple branches and execute a specific branch based on a condition.
- **Use Case:** Whenever you have **conditional logic** in your workflow—where the next step depends on some value or output—RunnableBranch is ideal.
- **Analogy:** Think of it as a **decision-making step** in a workflow, similar to `if-else` statements in Python.

---

### Example Use Case

Imagine you receive customer emails or messages that could be:
1. A complaint
2. A refund request
3. A general query

You want to handle each type differently:
- **Complaint:** Forward to customer support team
- **Refund:** Trigger a database workflow
- **General query:** Reply via a chatbot

---

### How to Implement

1. **Analyze the Email**
   - First, create a prompt for the LLM asking it to categorize the email:
     > "Analyze this email and classify it as either a complaint, refund request, or general query."
   - Send this prompt to the LLM.
   - LLM outputs the category: `complaint`, `refund`, or `general query`.

2. **Create RunnableBranch**
   - Use RunnableBranch to create **three branches**, one for each category.
   - Each branch defines the workflow for that category:
     - **Complaint branch:** Forward email
     - **Refund branch:** Trigger database workflow
     - **General query branch:** Generate chatbot response
   - RunnableBranch executes **only the branch corresponding to the LLM's output**.

---

### Another Example: Conditional Report Summarization

Suppose you want to generate a report from a topic provided by the user:

1. **Generate Report**
   - Ask the LLM to generate a report on a topic.
   - Parse the output.

2. **Check Report Length**
   - If the report is **greater than 500 words**, send it back to the LLM to **summarize**.
   - If the report is **less than 500 words**, print it **as it is**.

3. **RunnableBranch Implementation**
   - RunnableBranch allows you to conditionally select the workflow:
     - `> 500 words → Summarize`
     - `<= 500 words → Print as-is`

---

### Key Takeaways

- RunnableBranch is essential for **conditional execution** in workflows.
- It allows chaining of tasks **based on dynamic outputs**.
- It can handle any scenario where the next step depends on a decision or condition.


````python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain.schema.runnable import RunnableSequence, RunnableParallel, RunnablePassthrough, RunnableBranch, RunnableLambda
import os 

# -------------------------------
# Load environment variables
# -------------------------------
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")

# -------------------------------
# Prompt to generate a detailed report
# -------------------------------
prompt1 = PromptTemplate(
    template='Write a detailed report on {topic}',
    input_variables=['topic']
)

# -------------------------------
# Prompt to summarize a given text
# -------------------------------
prompt2 = PromptTemplate(
    template='Summarize the following text \n {text}',
    input_variables=['text']
)

# -------------------------------
# Initialize the LLM
# -------------------------------
model = ChatOpenAI(api_key=api_key)

# -------------------------------
# Output parser to handle string outputs from the LLM
# -------------------------------
parser = StrOutputParser()

# -------------------------------
# Step 1: Generate the report
# Using RunnableSequence to chain prompt -> model -> parser
# -------------------------------
report_gen_chain = prompt1 | model | parser

# -------------------------------
# Step 2: Conditional logic using RunnableBranch
# - If the report has more than 300 words, summarize it using prompt2 -> model -> parser
# - Otherwise, pass the report as-is using RunnablePassthrough
# -------------------------------
branch_chain = RunnableBranch(
    (lambda x: len(x.split())>300, prompt2 | model | parser),
    RunnablePassthrough()
)

# -------------------------------
# Step 3: Final workflow
# Sequence the report generation and branch logic
# -------------------------------
final_chain = RunnableSequence(report_gen_chain, branch_chain)

# -------------------------------
# Execute the chain with a given topic
# -------------------------------
print(final_chain.invoke({'topic':'Russia vs Ukraine'}))
````

## LangChain Expression Language (LCEL)

LangChain Expression Language (LCEL) is a **declarative syntax** introduced to simplify the creation of **sequential chains** using Runnables.  

So far, we have learned five Runnable primitives:
- `RunnableSequence`
- `RunnableParallel`
- `RunnablePassthrough`
- `RunnableLambda`
- `RunnableBranch`

Among these, the **most commonly used primitive is `RunnableSequence`**.  
This is because sequential chaining is a **typical use case** in almost every workflow:
- `RunnableSequence` is used directly to connect steps sequentially.
- It is also used **within parallel chains**, `Passthrough`, `Lambda`, and `Branch` examples.

---

### The Problem LCEL Solves

Previously, to create a sequential chain, we had to manually create a `RunnableSequence` instance:

```python
final_chain = RunnableSequence(runnable1, runnable2, runnable3)
```

While this works, it is **less readable** for larger chains.

LCEL introduces a **simpler and more declarative** syntax:

```python
final_chain = runnable1 | runnable2 | runnable3
```

- Here, the pipe (`|`) operator connects multiple Runnables sequentially

- This is easier to read, write, and maintain, especially in longer workflows

### How LCEL Works

1. LCEL is currently designed for sequential chains (RunnableSequence).
2. By using the pipe operator, you can chain multiple Runnables together without explicitly creating a RunnableSequence object.
3. This declarative approach improves code clarity and makes workflows easier to build.

**Example**

Instead of:

```python
final_chain = RunnableSequence(report_gen_chain, branch_chain)
```
With LCEL, you can write:

```python
final_chain = report_gen_chain | branch_chain
```

Both versions behave the same, but the LCEL syntax is cleaner and easier to scale.
