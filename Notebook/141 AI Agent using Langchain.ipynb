# AI Agents — Conceptual & Technical Overview (LangChain Context)

## What Problem Do AI Agents Solve?
- Traditional web-based workflows (booking, planning, research) are **multi-step, fragmented, and effort-intensive**
- Users must:
  - Visit multiple platforms
  - Perform repeated form-filling
  - Do manual comparison, decision-making, and execution
- This creates **high cognitive load**, especially for:
  - Non-technical users
  - Elderly users
  - Users unfamiliar with digital systems

**AI Agents simplify complex, multi-step tasks into a seamless, goal-driven experience.**

---

## Intuitive Use Case: Trip Planning
**Goal given by user**  
> “Create a budget travel itinerary from Delhi to Goa from May 1 to May 7.”

### Traditional Approach
- Manually book:
  - Travel tickets (train/flight)
  - Hotel stays
  - Local transport
  - Sightseeing & activities
- Requires days of research, comparisons, coordination, and execution

### AI Agent-Based Approach
- User provides **one high-level goal**
- Agent autonomously:
  - Understands intent & constraints (dates, budget, preferences)
  - Plans step-by-step execution
  - Uses tools/APIs to fetch real-time data
  - Asks for confirmation only where required
  - Executes bookings
  - Summarizes total cost
  - Sets reminders & calendar events

**Result:** Minimal user effort, maximum automation, improved experience

---

## Formal Definition of an AI Agent
> **An AI Agent is an intelligent system that receives a high-level goal from a user and autonomously plans, decides, and executes a sequence of actions using external tools, APIs, and knowledge sources, while maintaining context, reasoning over multiple steps, adapting to new information, and optimizing for the intended outcome.**

---

## Core Capabilities of an AI Agent
- Accepts **high-level goals**, not step-by-step instructions
- Breaks problems into **subtasks**
- Plans and executes tasks **sequentially**
- Uses **external tools & APIs**
- Maintains **memory & context**
- Adapts dynamically to new information or failures
- Optimizes outcomes (cost, speed, efficiency)

---

## LLM vs AI Agent — Key Difference

### Large Language Model (LLM)
- Strong at:
  - Reasoning
  - Language understanding
  - Generating responses
- Limitations:
  - Cannot take real-world actions
  - No direct access to APIs, databases, or systems

### AI Agent
- Combines:
  - **LLM** → reasoning & decision-making engine
  - **Tools** → APIs, databases, search, booking systems
- Can:
  - Perform actions
  - Execute workflows
  - Interact with external systems

> **AI Agent = LLM (Reasoning Brain) + Tools (Action Layer)**

---

## How an AI Agent Works (Behind the Scenes)
1. Receives user goal
2. Interprets intent using LLM
3. Creates an internal plan
4. Selects appropriate tools
5. Executes actions step-by-step
6. Maintains memory of past steps & preferences
7. Re-plans if new information or failures occur
8. Produces a final optimized outcome

# ReAct Design Pattern in AI Agents (Reasoning + Acting)

## What is ReAct?
**ReAct** is a **design pattern used in AI agents**.

- **Re** → Reasoning  
- **Act** → Acting (using tools)

> **ReAct allows a language model to interleave internal reasoning with external actions in a structured, multi-step process.**

---

## How ReAct Differs from Standard LLM Usage

### Traditional LLM Interaction
- Single-turn interaction
- User asks a question
- LLM directly generates an answer

**Limitation:**  
- No step-by-step reasoning
- No tool usage
- Cannot handle complex or multi-step tasks reliably

---

## ReAct Approach (Multi-Step)
- Problems are solved using **multiple steps**
- Each step may involve:
  - Thinking
  - Acting using tools
  - Observing results
- Continues until the final answer is reached

---

## Core ReAct Loop
ReAct works using a **repeating loop**:

1. **Thought**
   - Agent reasons about what to do next
2. **Action**
   - Agent selects and invokes a tool
3. **Observation**
   - Tool returns output (data/result)

This loop repeats until:
- Agent determines it has enough information
- Final answer is produced
- Loop stops

---

## Concrete Example: Population of Capital of France

### User Question
> “What is the population of the capital of France?”

---

### Iteration 1
**Thought:**  
- To answer this, first find the capital of France

**Action:**  
- Use search tool → query: *“capital of France”*

**Observation:**  
- Result: **Paris**

---

### Iteration 2
**Thought:**  
- Now find the population of Paris

**Action:**  
- Use search tool → query: *“population of Paris”*

**Observation:**  
- Result: **~2.1 million**

---

### Iteration 3 (Final)
**Thought:**  
- I now have all required information

**Final Answer:**  
> Paris is the capital of France, and it has a population of approximately 2.1 million.

**Loop Ends**

---

## What ReAct is Doing Internally
- **Reasoning** → handled by the LLM
- **Actions** → handled by tools (search, APIs, DBs)
- ReAct **combines both philosophies** into one workflow

> **ReAct = LLM reasoning + Tool-based execution**

---

## Key Advantages of ReAct

### 1. Handles Multi-Step Problems
- Ideal when intermediate information is required
- Example:
  - Find X → then compute Y → then answer

### 2. Tool Integration
- Supports:
  - Web search
  - Database lookup
  - API calls
- Enables real-world actions

### 3. Transparency & Auditability
- Agent’s:
  - Thoughts
  - Actions
  - Observations  
  are visible during execution

**This makes debugging and trust much easier.**

---

## When to Use ReAct
ReAct is highly effective when:
- Tasks require **multiple steps**
- External tools are needed
- Reasoning must adapt based on new information
- Transparency is important

---

## Origin of ReAct
- Introduced in the 2022 research paper:
  **“Synergizing Reasoning and Acting in Language Models”**
- Recommended reading:
  - Abstract
  - Introduction  
  for architectural understanding

---

## High-Level Summary
- ReAct operates on a **Thought → Action → Observation loop**
- Continues until a final answer is reached
- Merges reasoning and action in a clean, structured way
- Forms the foundation of many modern AI agents

---



# How Agent and Agent Executor Implement the ReAct Pattern

## Key Insight (Very Important)
> **Agent Executor is the main orchestrator in ReAct.**  
It performs most of the heavy lifting and controls the entire Thought–Action–Observation loop.

---

## What Needs to Be Implemented in ReAct
- ReAct requires a **continuous loop** of:
  - Thought
  - Action
  - Observation
- This loop runs **until a final answer is produced**
- **Agent Executor** is responsible for running and managing this loop

---

## Roles & Responsibilities

### Agent
- Focuses on:
  - Reasoning
  - Planning next steps
  - Deciding **which tool** to use and **with what input**
- Produces:
  - Thought
  - Action
  - Action Input

### Agent Executor
- Focuses on:
  - Orchestrating the loop
  - Executing actions
  - Managing state (thought trace)
- Produces:
  - Observations
  - Updated context for next iteration

---

## Step-by-Step ReAct Execution Flow

### Step 1: Loop Initialization (Executor → Agent)
Agent Executor sends:
- User query  
- Current **Thought Trace** (initially empty)

> **Thought Trace** = history of all previous  
Thought → Action → Observation steps

---

### Step 2: Reasoning & Planning (Agent)
Agent receives:
- User query
- Thought trace so far

Agent performs:
- **Thought** → What needs to be done next?
- **Action planning** → Which tool to use?
- **Action input** → Tool input parameters

Agent sends back:
- Action name
- Action input

---

### Step 3: Tool Execution (Executor)
Agent Executor:
- Invokes the selected tool
- Passes the action input
- Receives tool output

Tool output becomes:
- **Observation**

---

### Step 4: Update State (Executor)
Agent Executor:
- Appends observation to the thought trace
- Maintains full execution history

This updated thought trace is now the **context** for the next iteration.

---

### Step 5: Loop Continues
- Executor sends:
  - User query
  - Updated thought trace
- Agent reasons again
- New action planned or final answer produced

Loop repeats until:
- Agent decides no further action is required

---

## Example (Capital of France)

### Iteration 1
- **Thought:** Find capital of France
- **Action:** Search tool
- **Observation:** Paris

### Iteration 2
- **Thought:** Find population of Paris
- **Action:** Search tool
- **Observation:** ~2.1 million

### Iteration 3
- **Thought:** Final answer ready
- **Final Output:** Returned to user
- **Loop stops**

---

## Why Agent Executor is Critical
- Maintains execution state
- Controls loop lifecycle
- Executes tools safely
- Ensures deterministic flow

> Without Agent Executor, the ReAct loop cannot function.

---

## Mental Model (Easy to Remember)
- **Agent** → Thinks & decides  
- **Agent Executor** → Executes & manages memory  

---

## High-Level Summary
- ReAct is implemented by **collaboration**
- Agent plans, Executor executes
- Thought trace is continuously updated
- Loop ends only when final answer is produced

---

```python
# =========================
# Step 1: Import Libraries
# =========================

# LLM wrapper for OpenAI chat models
from langchain_openai import ChatOpenAI

# Tool decorator to convert Python functions into LangChain tools
from langchain_core.tools import tool

# For making HTTP API calls
import requests

# Built-in community tool for web search
from langchain_community.tools import DuckDuckGoSearchRun
```

```python
# =========================
# Step 2: Create Tools
# =========================

# Tool 1: DuckDuckGo Search Tool
# Allows the agent to search the internet
search_tool = DuckDuckGoSearchRun()


# Tool 2: Weather API Tool
@tool
def get_weather_data(city: str) -> str:
    """
    Fetches current weather data for a given city using Weatherstack API.
    This tool will be called by the agent when weather information is required.
    """
    url = (
        f"https://api.weatherstack.com/current"
        f"?access_key=936b245b27dfb08b546a7b9b30b0dd88"
        f"&query={city}"
    )

    response = requests.get(url)

    # Return raw JSON response (agent will interpret it)
    return response.json()
```

```python
# =========================
# Step 3: Load Environment Variables
# =========================

from dotenv import load_dotenv
import os

# Load variables from .env file
load_dotenv(override=True)

# Fetch OpenAI API key securely
api_key = os.getenv("Open_AI_API_key")
```

```python
# =========================
# Step 4: Initialize LLM
# =========================

# ChatOpenAI acts as the reasoning engine for the agent
llm = ChatOpenAI(api_key=api_key)
```

```python
# =========================
# Step 5: Import Agent Utilities
# =========================

from langchain.agents import create_react_agent, AgentExecutor
from langchain import hub
```

```python
# =========================
# Step 6: Pull ReAct Prompt
# =========================

# Pulls a predefined ReAct prompt from LangChain Hub
# This prompt enforces the Thought → Action → Observation loop
prompt = hub.pull("hwchase17/react")
```

```python
# =========================
# Step 7: Create ReAct Agent
# =========================

# create_react_agent combines:
# - LLM (reasoning)
# - Tools (actions)
# - Prompt (behavior rules)
agent = create_react_agent(
    llm=llm,
    tools=[search_tool, get_weather_data],
    prompt=prompt
)
```

```python
# =========================
# Step 8: Create Agent Executor
# =========================

# AgentExecutor:
# - Orchestrates the ReAct loop
# - Executes tools
# - Maintains scratchpad (thought trace)
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool, get_weather_data],
    verbose=True  # Shows Thought, Action, Observation steps
)
```

```python
# =========================
# Step 9: Invoke the Agent
# =========================

# User provides a single high-level query
response = agent_executor.invoke({
    "input": "Find the capital of New Delhi, then find its current weather condition and what is the AQI of New Delhi is that aqi is in survval range dangeor in danger zone?"
})

# Print full response object (includes logs + output)
print(response)
```

```python
# =========================
# Step 10: Access Final Answer
# =========================

# Extract only the final answer generated by the agent
response["output"]
```
