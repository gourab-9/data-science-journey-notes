## Introduction to Vector Stores  
### Why Do We Need Vector Stores?

Before understanding **what vector stores are**, it is important to understand **why vector stores are needed**.  
To explain this clearly, let’s walk through a **real-world use case**.

---

## Example: Movie Recommendation System

Imagine you want to build a website like **IMDb**, where:
- Every movie in the world is listed
- Users can search for a movie
- Users can see complete information about that movie

### Basic Website Architecture

To build this system, you typically need:
1. **Database**
2. **Backend (e.g., Python)**
3. **Frontend (UI)**

---

## Step 1: Traditional Movie Database

Your database might contain fields like:
- Movie ID
- Movie Name
- Director
- Actors
- Genre
- Release Year
- Hit / Flop Status

This data can be collected using:
- Public APIs
- Web scraping

Once stored, the backend fetches this data and sends it to the frontend for display.

---

## Step 2: Adding a Movie Recommendation System

Now you want to improve user engagement by adding:
> **"Similar Movies" recommendations**

### Example
If a user is viewing **Spider-Man**, you might recommend:
- Iron Man
- Captain America

This increases:
- Time spent on the website
- User engagement
- Ad revenue

---

## First Attempt: Keyword-Based Recommendation

A simple approach is **keyword matching**, where you compare movies based on:
- Director
- Actor
- Genre
- Release period

### Logic
- If many keywords match → Movies are similar
- If few keywords match → Movies are dissimilar

This system:
- Is easy to build
- Works reasonably well in some cases

---

## Major Problems with Keyword Matching

### Problem 1: Illogical Recommendations

**Example**
- User watched: *My Name Is Khan*
- Recommended movie: *Kabhi Alvida Naa Kehna*

Why?
- Same director (Karan Johar)
- Same actor (Shah Rukh Khan)
- Similar genre (Drama)

But in reality:
- Storylines are completely different
- Emotional theme is different

➡️ **Bad recommendation**

---

### Problem 2: Missing Truly Similar Movies

**Example**
- *Taare Zameen Par*
- *A Beautiful Mind*

These movies are:
- Thematically very similar
- Both about brilliance mixed with personal struggle

But:
- Directors are different
- Actors are different
- Release years are different

➡️ Keyword matching fails to detect similarity

---

## Root Cause of the Problem

Keyword matching is:
- Shallow
- Surface-level
- Does not understand **meaning**

To build a **high-quality recommendation system**, we must compare:
> **The story (plot) of movies, not metadata**

---

## Improved Idea: Plot-Based Similarity

New approach:
- Compare movie **plots**
- If stories are similar → movies are similar

### New Requirement
You now need:
- Full plot text for every movie  
  (often 2000–3000 words)

So you:
- Collect plots using APIs or web scraping
- Store them in your database

---

## New Challenge: Comparing Movie Plots

Now the task becomes:
> **How do we compare two large pieces of text based on meaning?**

This is a very difficult NLP problem using traditional methods.

---

## Solution: Text Embeddings

### What Are Embeddings?

Embeddings are a technique to:
- Convert text into **numbers (vectors)**
- Preserve **semantic meaning**

### How It Works
1. Text is passed through a neural network
2. The model understands the meaning
3. Output is a high-dimensional vector (e.g., 512 dimensions)

Each vector represents:
> **The meaning of the text**

---

## Embedding-Based Movie Representation

Now:
- Each movie plot → One embedding vector
- All movies → Set of vectors

Example:
- M1 → 3 Idiots
- M2 → Inception
- M3 → Chennai Express
- M4 → Stree

---

## Measuring Similarity Between Movies

These vectors can be:
- Plotted in a high-dimensional space
- Compared using **cosine similarity** or **angular distance**

### Key Idea
- Smaller angle → Higher similarity
- Larger angle → Lower similarity

So if:
- M4 is closest to M1  
➡️ M1 is the most similar movie to M4

---

## Result: Semantic Movie Recommendation System

Now recommendations are based on:
- Story similarity
- Semantic meaning
- Not just keywords

This produces:
- Much better recommendations
- Higher user satisfaction

---

## Practical Challenges in Building This System

When implementing this at scale, you face **three major challenges**:

---

### Challenge 1: Embedding Generation

- You may have **millions of movies**
- You must generate embeddings for each one

---

### Challenge 2: Storage of Embeddings

- Embeddings are reused many times
- They must be stored efficiently
- **Relational databases (MySQL, Oracle)** are not suitable
  - They cannot perform vector similarity search

➡️ You need a **special type of database**

---

### Challenge 3: Fast Semantic Search

Problem:
- Find top 5 similar movies
- From **10 lakh vectors**

Brute-force comparison:
- Too slow
- Computationally expensive
- Bad user experience

You need:
- Intelligent indexing
- Fast similarity search
- Approximate nearest neighbor algorithms

---

## Final Answer: Vector Stores

**Vector Stores solve all three problems:**
1. Efficient storage of embeddings
2. Fast semantic similarity search
3. Scalable retrieval over millions of vectors

---

## Key Takeaway

Whenever your application needs:
- Semantic search
- Similarity-based recommendation
- Embedding storage and retrieval

➡️ **Vector Stores are the correct solution**

## Vector Store — Definition

A **vector store** is a system designed to **store and retrieve data represented as numerical vectors**.

In simple terms:
- If your application generates vectors (for example, embeddings)
- And you need to **store them**, **retrieve them**, and **search among them efficiently**
- Then the system used for this purpose is called a **vector store**

---

## Core Features of a Vector Store

Every vector store provides a few **essential features**.  
Let’s go through them one by one.

---

## 1. Storage

### What does storage mean in a vector store?

The most fundamental feature of a vector store is the ability to **store vectors**.

- These vectors usually come from:
  - Text embeddings
  - Image embeddings
  - Audio embeddings
  - Any numerical vector representation

---

### Associated Metadata

Along with vectors, vector stores also allow you to store **metadata**.

**Example (Movie Use Case):**
- Vector → Embedding of the movie plot
- Metadata → 
  - Movie ID
  - Movie name
  - Genre
  - Release year

This metadata:
- Helps in filtering
- Helps in contextual retrieval
- Makes results more meaningful

---

### In-Memory vs On-Disk Storage

Vector stores usually offer **two storage modes**:

#### In-Memory Storage
- Stored in RAM
- Very fast access
- Data is lost when the application stops
- Best for:
  - Small projects
  - Experiments
  - Prototypes

#### On-Disk Storage
- Stored on disk / database
- Persistent storage
- Data remains even after application restarts
- Best for:
  - Production systems
  - Enterprise-scale applications

➡️ Most vector stores support **both options**, and you choose based on your use case.

---

## 2. Similarity Search

This is a **core capability** of any vector store.

### What does similarity search do?

Given:
- A **query vector**
- A collection of stored vectors

The vector store:
- Compares the query vector with stored vectors
- Computes similarity scores
- Returns the **most similar vectors**

Common similarity metrics:
- Cosine similarity
- Dot product
- Euclidean distance

This feature enables:
- Semantic search
- Recommendation systems
- Retrieval-Augmented Generation (RAG)

---

## 3. Indexing (For Fast Search)

### Why indexing is needed?

Without indexing:
- If you have 10 lakh vectors
- And one query vector
- You would need to compare against **all 10 lakh vectors**

This is:
- Linear search
- Computationally expensive
- Very slow

---

### What indexing does

Indexing introduces **smart data structures** that:
- Reduce the number of comparisons
- Make similarity search much faster
- Work efficiently even for high-dimensional vectors

---

### Simple Indexing Example (Clustering-Based Intuition)

Assume:
- 10 lakh vectors
- Each vector has 784 dimensions

#### Step 1: Clustering
- Group vectors into clusters (say 10 clusters)
- Each cluster has ~1 lakh vectors

#### Step 2: Compute Centroids
- For each cluster, compute an average vector (centroid)
- Now you have 10 centroid vectors

#### Step 3: Query Processing
- Compare the query vector with only 10 centroids
- Find the closest centroid
- Search **only inside that cluster**

Result:
- Instead of 10 lakh comparisons
- You do ~1 lakh + a few centroid comparisons

➡️ Massive performance improvement with minimal accuracy loss

---

### Approximate Nearest Neighbor (ANN)

Most vector stores internally use:
- Approximate Nearest Neighbor algorithms

These:
- Trade exactness for speed
- Are extremely fast
- Are suitable for large-scale systems

Indexing is a **deep research topic**, but the key idea is:
> Vector stores use intelligent indexing to make similarity search scalable.

---

## 4. CRUD Operations

Vector stores also support standard database operations:

- **Create** → Add new vectors
- **Read** → Retrieve vectors
- **Update** → Modify existing vectors or metadata
- **Delete** → Remove outdated vectors

This makes vector stores:
- Easy to integrate
- Practical for real-world applications

---

## Summary: Key Features of a Vector Store

A vector store provides:

- Vector storage (with metadata)
- Similarity search
- Indexing for fast retrieval
- Full CRUD operations

Vector stores are used wherever vectors are central to the application:

## Final Takeaway

If your application:

➡️ **Vector stores are the correct and scalable solution**

Vector stores are:

# Vector Stores vs Vector Databases (LangChain Context)

Before understanding **how vector stores are used in LangChain**, it is important to clearly understand a commonly confusing topic:

> **Vector Store vs Vector Database**

These two terms are often used interchangeably on YouTube, blogs, and documentation, but they are **not exactly the same**.

---

## What is a Vector Store?

A **vector store** is a system that primarily performs **two core functions**:

1. **Storage**
   - Stores numerical vectors (embeddings)
   - Stores associated **metadata** (IDs, text, labels, source info, etc.)

2. **Retrieval**
   - Performs **semantic similarity search**
   - Finds vectors most similar to a query vector

If a system provides **only these two capabilities**, it can be classified as a **vector store**.

### Key Characteristics of Vector Stores
- Lightweight
- Focused on embeddings + similarity search
- May **not** include traditional database features
- Best suited for:
  - Prototyping
  - Small-scale applications
  - Local or in-memory use cases

### Example
- **FAISS (Facebook AI Similarity Search)**
  - A popular vector store
  - Excellent for fast similarity search
  - Does not provide full database features

---

## What is a Vector Database?

A **vector database** is a **vector store plus database-like features**.

In other words:

> **Vector Database = Vector Store + Database Capabilities**

### Additional Features Provided by Vector Databases

Along with storage and retrieval, vector databases usually offer:

- **Distributed architecture**
  - Horizontal scaling
  - Handles very large datasets

- **Durability & persistence**
  - Data stored reliably on disk
  - Backup and restore support

- **Concurrency control**
  - Multiple users can query/update simultaneously

- **Transactions**
  - Atomic operations (ACID or near-ACID guarantees)

- **Authentication & authorization**
  - Secure access control
  - Role-based permissions

- **Advanced metadata handling**
  - Rich filtering and structured queries

### Where Vector Databases Are Used
- Production environments
- Enterprise-scale applications
- Large and continuously growing datasets

### Examples of Vector Databases
- **Milvus**
- **Weaviate**
- **Pinecone**
- **Qdrant**

---

## Core Difference (One-Line Rule)

> Every **vector database** is a **vector store**,  
> but **not every vector store** is a **vector database**.

---

## Quick Comparison Table

| Aspect | Vector Store | Vector Database |
|------|-------------|----------------|
| Vector storage | ✅ | ✅ |
| Similarity search | ✅ | ✅ |
| Indexing | ✅ | ✅ |
| Persistence | Optional | ✅ |
| Distributed system | ❌ | ✅ |
| Backup & restore | ❌ | ✅ |
| Transactions | ❌ | ✅ |
| Authentication | ❌ | ✅ |
| Production-ready | ❌ (mostly) | ✅ |

---

## Practical Takeaway for LangChain

- LangChain supports **both vector stores and vector databases**
- Choice depends on:
  - Scale of application
  - Persistence needs
  - Security requirements
  - Production readiness

### Rule of Thumb
- **Learning / small apps / experiments** → Vector Store (e.g., FAISS)
- **Production / enterprise / RAG systems** → Vector Database (e.g., Pinecone, Weaviate)

---

Now that the fundamentals of **vector stores vs vector databases** are clear, we can confidently move forward to **how LangChain uses vector stores internally**.


# Vector Store Implementation in LangChain

Now let us understand **how vector stores are implemented and used in LangChain**.

When **LangChain was originally designed as a framework**, its creators realized very early that **future LLM-based applications would heavily rely on embedding vectors**. This assumption turned out to be absolutely correct.

Any **RAG-based application** fundamentally works with:
- Text → Embeddings  
- Embeddings → Vector Stores / Vector Databases  

So it was obvious that **vector stores would play a central role** in LangChain’s ecosystem.

---

## Early and Strong Support for Vector Stores

Because of this foresight, LangChain provides **extensive built-in support** for a wide variety of popular vector stores and vector databases.

Some well-known examples supported by LangChain:
- FAISS
- Pinecone
- Chroma
- Qdrant
- Weaviate  
- and many more

For each of these, LangChain provides **ready-made wrapper components**.

---

## Core Design Philosophy: Common Interface

The **most important design idea** behind LangChain’s vector store integration is:

> **All vector store wrappers share the same method signatures**

This means:
- Method names are the same
- Method behavior is conceptually the same
- APIs feel consistent across different vector stores

### Why This Matters

If you write code using one vector store today (e.g., FAISS) and later decide to switch to another (e.g., Pinecone), you **do not need to rewrite your application logic**.

You simply replace the vector store backend.

---

## Common Methods Across Vector Stores

LangChain defines a **common set of methods** that almost every vector store supports:

### 1. Creating a Vector Store
- `from_documents()`
- `from_texts()`

Used to:
- Generate embeddings
- Store them inside the vector store

---

### 2. Adding New Data
- `add_documents()`
- `add_texts()`

Used to:
- Insert new embeddings into an existing vector store

---

### 3. Semantic Similarity Search
- `similarity_search()`

Used to:
- Compare a query embedding against stored embeddings
- Retrieve the most semantically similar results

---

### 4. Metadata-Based Filtering
- Allows filtering results using metadata
- Example use cases:
  - Filter by document source
  - Filter by category, tag, or ID

---

## Plug-and-Play Vector Store Switching

Because all vector stores follow the **same interface**, LangChain enables **easy backend switching**.

### Example Scenario
- You initially build a small application using **FAISS**
- Your app grows and needs:
  - Persistence
  - Scalability
  - Production-grade infrastructure
- You decide to move to **Pinecone**

➡️ You can replace FAISS with Pinecone **without major code changes**, because:
- Method names remain the same
- Function calls behave consistently

This makes LangChain **highly flexible and future-proof**.

---

## Summary of LangChain’s Vector Store Strategy

- Built-in components for all major vector stores
- Unified and consistent API
- Minimal coupling between application logic and storage backend
- Easy migration from prototype to production

```python
from langchain_openai import OpenAIEmbeddings
from langchain.vectorstores import Chroma
```

```python

from langchain.schema import Document

# Create LangChain documents for IPL players

doc1 = Document(
        page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and fitness, he has led the Royal Challengers Bangalore in multiple seasons.",
        metadata={"team": "Royal Challengers Bangalore"}
    )
doc2 = Document(
        page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm demeanor and ability to play big innings under pressure.",
        metadata={"team": "Mumbai Indians"}
    )
doc3 = Document(
        page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicketkeeping, and leadership are legendary.",
        metadata={"team": "Chennai Super Kings"}
    )
doc4 = Document(
        page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his yorkers and death-over expertise.",
        metadata={"team": "Mumbai Indians"}
    )
doc5 = Document(
        page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and match-winning performances make him a key player.",
        metadata={"team": "Chennai Super Kings"}
    )
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
from dotenv import load_dotenv
import os

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
```

```python
vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(api_key=api_key),
    persist_directory='my_chroma_db',
    collection_name='sample'
)
```

```python
# add documents
vector_store.add_documents(docs)
```

```python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
     
```

```python
# search documents
vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2
)
```

```python

# search with similarity score
vector_store.similarity_search_with_score(
    query='Who among these are a bowler?',
    k=2
)
    
```

```python
# meta-data filtering
vector_store.similarity_search_with_score(
    query="",
    filter={"team": "Chennai Super Kings"}
)
```

```python
# update documents
updated_doc1 = Document(
    page_content="Virat Kohli, the former captain of Royal Challengers Bangalore (RCB), is renowned for his aggressive leadership and consistent batting performances. He holds the record for the most runs in IPL history, including multiple centuries in a single season. Despite RCB not winning an IPL title under his captaincy, Kohli's passion and fitness set a benchmark for the league. His ability to chase targets and anchor innings has made him one of the most dependable players in T20 cricket.",
    metadata={"team": "Royal Challengers Bangalore"}
)

vector_store.update_document(document_id='97bacfd1-60df-404b-b229-0843b92bd6a3', document=updated_doc1)
```

```python

# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python

# delete document
vector_store.delete(ids=['97bacfd1-60df-404b-b229-0843b92bd6a3'])
```
