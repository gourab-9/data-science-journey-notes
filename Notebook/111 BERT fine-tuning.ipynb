# Fine-Tuning in Machine Learning

Fine-tuning is a process where we take a **pre-trained model** and adapt it to a **specific task** through some extra training.

You can imagine the pre-trained model as a **slab of marble**, and fine-tuning as the process of **chiseling it into a statue**.  
The marble (pre-trained model) already has structure and potential — we just shape it further to fit our desired purpose (the fine-tuned model).

---

## Pre-Trained Models and Self-Supervision

A pre-trained model is often created using **self-supervised learning**.  
This means the model **learns from data without needing manual labels**.  
The labels come naturally from the structure of the data itself.

### Example (Text Data)
In text, we can train a model to **predict the next word (or token)** in a sentence.  
For example, take the sentence:

> "Adapting a pre-trained model to a particular task."

The model might learn like this:
- “adapt” → predicts “ing”  
- “adapting” → predicts “a”  
- “adapting a” → predicts “pre”  
- “adapting a pre” → predicts “trained”  
- and so on...

Here, no human labels are needed — the data provides its own supervision.

---

## Fine-Tuned Models and Supervised Learning

Once the pre-trained model has learned general patterns, we **fine-tune** it on a smaller, specific dataset.  
This step usually involves **supervised learning**, meaning **humans define the correct answers (labels)**.

### Why This Two-Step Process Matters
- **Pre-training** doesn’t need manual labels → we can train on **massive datasets**.  
- **Fine-tuning** uses labeled examples → helps the model perform well on a **specific task**.

For example, **BERT** was pre-trained on about **3 billion words**, while a fine-tuning dataset might have only a **few thousand examples**.  
That’s a difference of **millions of times** more data in pre-training!

---

## The Modern Paradigm

This two-step method — **pre-training first, fine-tuning later** — is now the **standard approach** for building advanced machine learning models.

It allows:
- Models to **learn general knowledge** from huge datasets.
- Developers to **adapt** these models for particular applications easily.

---

## Democratization of AI

Training large models like **GPT-4, LLaMA, Mistral, Claude,** etc., needs enormous computing power and resources.  
But once these **foundation models** are built by big research labs (like OpenAI, Google, Meta, Anthropic, or Mistral), they can be **shared publicly**.

Anyone can then **fine-tune** them for specific use cases — this makes AI development more **accessible and inclusive**.

---

## Multiple Rounds of Fine-Tuning

Fine-tuning doesn’t have to stop after one round.  
A **fine-tuned model** can itself be **fine-tuned again** for even more specialized purposes.

For example, **ChatGPT** was built in three stages:
1. **Unsupervised pre-training**  
2. **Supervised fine-tuning**  
3. **Reinforcement learning** (to refine behavior further)

# Understanding BERT: The Foundation of Modern NLP Models

In this part, we’ll talk about **BERT**, one of the most important models in natural language processing (NLP).

Even though the idea of **pre-training** and **fine-tuning** became widely popular after OpenAI’s ChatGPT, this concept has existed for years.  
In fact, one of the earliest and most influential models to use this approach was **Google’s BERT**, released in **2019**.

---

## What is BERT?

**BERT** stands for **Bidirectional Encoder Representations from Transformers**.  
It was designed by Google with **fine-tuning** in mind — meaning it was built to be reused and adapted for many different NLP tasks.

Researchers at Google wanted to create a **foundation model** that others could easily **fine-tune** for specific applications such as:
- Sentiment analysis  
- Question answering  
- Text classification  
- Sentence similarity  

To make this possible, BERT was trained on **two main tasks**.

---

## 1. Masked Language Modeling (MLM)

In **masked language modeling**, a random word in a sentence is **hidden (masked)**, and the model learns to **predict that missing word** using the surrounding context.

### Example
Sentence:  
> “The cat ___ on the mat.”

BERT looks at the **words before and after** the blank — “The cat” and “on the mat” — to guess that the missing word is “sat”.

This is different from models like **GPT**, which use **causal language modeling**.  
GPT learns to predict the **next word** in a sequence (left-to-right), while BERT can use information from **both directions** — before and after the word.

### Why It Matters
Because BERT reads text **bidirectionally**, it has **more context** and can make **more accurate predictions**.  
It doesn’t just see what came before — it also understands what comes after.

---

## 2. Next Sentence Prediction (NSP)

The second task BERT was trained on is called **Next Sentence Prediction**.  
This task helps the model understand **relationships between two sentences**.

### Example
Sentence A:  
> “BERT is conceptually simple and empirically powerful.”  
Sentence B:  
> “It obtains new state-of-the-art results on 11 NLP tasks.”

In this case, Sentence B **is** the next logical sentence after Sentence A, so the correct label is **“Is Next.”**

However, if Sentence B were:  
> “The cat sat on the mat,”  
then the label would be **“Is Not Next.”**

### Why It Matters
Understanding how two sentences relate is crucial for tasks like:
- **Question answering**
- **Text matching**
- **Sentence similarity**
- **Reading comprehension**

By combining **masked language modeling** and **next sentence prediction**, BERT learns both **word-level understanding** and **sentence-level relationships**.

---

## Why BERT Was Revolutionary

Before BERT, most models could only look at text in one direction (like reading from left to right).  
BERT changed that by reading **both forward and backward**, giving it a **deeper understanding of context**.

As a result, it quickly achieved **state-of-the-art performance** on many NLP benchmarks and became the **foundation for modern transformer-based models**, including GPT and many others.

# Understanding Text Classification

Before jumping into example code, let’s look at one more important concept — **Text Classification**.

---

## What is Text Classification?

**Text classification** is the process of giving a **label** or **category** to a piece of text.  
In simple terms, it means telling the model:  
> “Here’s some text — now decide what type it is.”

For example, the **Next Sentence Prediction (NSP)** task we discussed earlier with **BERT** is actually a type of **text classification**.  
In NSP, the model reads two sentences and classifies them as either **“Is Next”** or **“Is Not Next.”**

---

## Common Examples of Text Classification

Here are a few real-world examples where text classification is very useful:

### 1. Spam Detection
- **Input:** An incoming email message.  
- **Task:** Predict whether the email is **Spam** or **Not Spam**.  
- This helps email systems automatically filter unwanted messages.

### 2. IT Ticket Categorization
- **Input:** A user’s IT support message or complaint.  
- **Task:** Classify the ticket into the right **category or priority level** (for example, “Network Issue,” “Software Bug,” or “Hardware Request”).  
- This helps automate and speed up customer support.

### 3. Sentiment Analysis
- **Input:** A product or service review from a customer.  
- **Task:** Identify the **emotion or sentiment** — such as **Positive**, **Negative**, or **Neutral**.  
- This helps companies understand customer satisfaction and feedback trends.

---

## Why Text Classification Matters

Text classification is one of the **most common NLP tasks**, and it forms the **basis for many real-world applications**, including:
- Email filtering  
- Customer service automation  
- Social media monitoring  
- Review and feedback analysis  
- Fake news and phishing detection

# Imports

```python
from datasets import DatasetDict, Dataset, load_dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer

import evaluate
import numpy as np
from transformers import DataCollatorWithPadding
```

# Load Data

```python
dataset_dict = load_dataset("shawhin/phishing-site-classification")
```

# Load Pre-trained Model

```python
# define pre-trained model path
model_path = "google-bert/bert-base-uncased"

# load model tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_path)

# load model with binary classification head
id2label = {0: "Safe", 1: "Not Safe"}
label2id = {"Safe": 0, "Not Safe": 1}
model = AutoModelForSequenceClassification.from_pretrained(model_path,
num_labels=2,
id2label=id2label,
label2id=label2id,)
```

# Set Trainable Parameters

```python
# freeze all base model parameters
for name, param in model.base_model.named_parameters():
  param.requires_grad = False

# unfreeze base model pooling layers
for name, param in model.base_model.named_parameters():
  if "pooler" in name:
    param. requires_grad = True
```

# Data Pre-processing

```python
# define text preprocessing
def preprocess_function(examples) :
  # return tokenized text with truncation
  return tokenizer(examples["text"], truncation=True)

# preprocess all datasets
tokenized_data = dataset_dict.map(preprocess_function, batched=True)

```

```python
# create data collator
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
```

# Define Evaluation Metrics

```python

# load metrics
accuracy = evaluate.load("accuracy")
auc_score = evaluate.load("roc_auc")

def compute_metrics(eval_pred) :
# get predictions
  predictions, labels = eval_pred

  # apply softmax to get probabilities
  probabilities = np.exp(predictions) / np.exp(predictions).sum(-1,
  keepdims=True)

  # use probabilities of the positive class for ROC AUC
  positive_class_probs = probabilities[:, 1]
  # compute auc
  auc = np.round(auc_score. compute(prediction_scores=positive_class_probs,
  references=labels) ['roc_auc'],3)

  # predict most probable class
  predicted_classes = np.argmax(predictions, axis=1)
  # compute accuracy
  acc = np.round(accuracy.compute(predictions=predicted_classes,
  references=labels) ['accuracy'],3)

  return {"Accuracy": acc, "AUC": auc}
```

# Training Parameters

```python
# hyperparameters
lr = 2e-4
batch_size = 8
num_epochs = 10

training_args = TrainingArguments(
  output_dir="bert-phishing-classifier_teacher",
  learning_rate=lr,
  per_device_train_batch_size=batch_size,
  per_device_eval_batch_size=batch_size,
  num_train_epochs=num_epochs,
  logging_strategy="epoch",
  eval_strategy="epoch",
  save_strategy="epoch",
  load_best_model_at_end=True,
)
```

# Fine-tune Model

```python
trainer = Trainer (
model=model,
args=training_args,
train_dataset=tokenized_data["train"],
eval_dataset=tokenized_data["test"],
tokenizer=tokenizer,
data_collator=data_collator,
compute_metrics=compute_metrics,

)

trainer.train()
```

# Validation Data

```python
# apply model to validation dataset
predictions = trainer.predict(tokenized_data["validation"])

# Extract the logits and labels from the predictions object
logits = predictions.predictions
labels = predictions. label_ids

# Use your compute_metrics function
metrics = compute_metrics((logits, labels))
print(metrics)

# >> {'Accuracy': 0.896, 'AUC': 0.945}
```
