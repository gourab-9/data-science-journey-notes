# Tool Binding in LangChain

Now we start with **Tool Binding**.

Tool Binding is the **process where a tool is connected to an LLM**.  
In other words, you can say that **a tool is registered with an LLM**.

---

## What Is Tool Binding?

**Tool Binding** is the step where:
- You register tools with a Large Language Model
- So that the LLM knows:
  - Which tools are available
  - What each tool does
  - What input format each tool expects

In simple terms, Tool Binding makes the LLM **aware of the tools it can use**.

---

## Why Tool Binding Is Required

From the previous notebook, we already learned that while creating a tool, we must define:

1. **Tool Name**  
   - So the LLM can identify the tool

2. **Tool Description**  
   - So the LLM understands *what the tool does*

3. **Input Schema**  
   - So the LLM knows *what input format the tool expects*

However, just creating a tool is **not enough**.

Unless the tool is **bound to the LLM**, the LLM:
- Does not know that the tool exists
- Cannot decide when to use it
- Cannot call it correctly

This is exactly what Tool Binding solves.

---

## What Happens During Tool Binding?

When you perform Tool Binding, three important things happen:

### 1. Tool Availability Awareness
The LLM now knows:
- Which tools are available
- Which tools it is allowed to use

### 2. Tool Purpose Understanding
Because each tool has a description:
- The LLM understands what each tool is meant for
- This helps the LLM decide *when* a tool should be used

### 3. Correct Tool Invocation Format
Since the input schema is shared:
- The LLM understands the exact input structure
- Whenever it calls a tool, it sends inputs **in the correct format**

This ensures:
- No mismatched inputs
- No incorrect tool usage

---

## Why Tool Binding Is a Critical Step

All of the following happens **only after tool binding**:
- The LLM can choose the right tool
- The LLM can call the tool correctly
- The LLM can perform real tasks using that tool

So, Tool Binding is the **bridge** between:
- An LLM that can think
- And tools that can act

# Tool Calling in LangChain

Now we come to the **most important third step**, where we understand **when and how an LLM calls a tool whenever required**.  
This entire process is called **Tool Calling**.

---

## What Is Tool Calling?

**Tool Calling** is a process where:
- The LLM **decides during a conversation or task**
- That it needs to use a **specific tool**
- And then **generates a structured output**
  - Containing the **tool name**
  - And the **arguments required to call that tool**

In simple words, Tool Calling is the moment when:
> The LLM realizes, *“I cannot do this task on my own, I need to use a tool.”*

---

## Understanding Tool Calling with an Example

Let us consider a simple scenario.

### Initial Setup
- You are chatting with an LLM
- The LLM has access to a **multiplication tool**
- This tool is already **bound** to the LLM

---

## Case 1: When Tool Calling Is NOT Required

Suppose you ask the LLM:

> **“Hi, how are you?”**

In this case:
- The LLM does **not** need any tool
- It already knows how to respond using its language and reasoning abilities

So the LLM will simply reply with a normal text response like:
> “Hi, I am good. How can I help you?”

Here:
- No tool is called
- No structured output is generated
- The response is purely conversational

---

## Case 2: When Tool Calling IS Required

Now suppose you ask the LLM:

> **“What is 8 multiplied by 7?”**

At this point, the LLM follows a decision-making process:

### Step 1: Tool Check
The LLM checks:
- What tools are available to me?

It finds that:
- A **multiplication tool** is available

---

### Step 2: Decision Making
The LLM understands:
- This task involves calculation
- A tool exists that is specifically designed to handle multiplication

So instead of answering directly, the LLM decides:
> “I should use the multiplication tool.”

---

### Step 3: Structured Output Generation

Instead of generating a normal text answer, the LLM now generates a **structured output** that contains:

1. **Name of the tool to be called**  
   - Example: `multiply`

2. **Arguments for the tool**  
   - Based on the question:
     - First number = 8
     - Second number = 7

So the output clearly specifies:
- Which tool should be executed
- What inputs should be passed to that tool

This structured response is **not meant for the user directly**,  
but for the system to understand and execute the tool correctly.

---

## Why This Process Is Called Tool Calling

This entire flow—from:
- Understanding the user query
- Deciding that a tool is needed
- Selecting the correct tool
- Generating structured arguments

is known as **Tool Calling**.

---

## Key Takeaway

- Tool Calling does **not happen for every question**
- It happens **only when the LLM realizes it cannot solve the task alone**
- The LLM does **not execute the tool itself**
- It only **decides and specifies**:
  - Which tool to use
  - How to call it

Once this structured output is generated, the system executes the tool and returns the result.

---

## Summary in Simple Terms

- **Tool Binding** → Makes the LLM aware of available tools  
- **Tool Calling** → The LLM decides *when* and *how* to use a tool  

This is what gives LLMs their **real-world action capability**.

# Tool Execution in LangChain

So far, we have learned **three very powerful concepts**:

1. How to **create tools**
2. How to **bind tools** with an LLM
3. The concept of **tool calling**

Now comes the **fourth and probably the most interesting piece of the puzzle**, which is **Tool Execution**.

---

## What Is Tool Execution?

Once the LLM suggests that:
- *“To answer this query, you should use this specific tool”*, and
- *“While using this tool, pass these exact inputs”*

Then the next responsibility comes to **you as a programmer**.

At this stage:
- The LLM has already done its job
- It has decided **which tool to use**
- It has also decided **what inputs should be passed**

Now **you manually execute the tool** using those inputs.

This entire process is called **Tool Execution**.

---

## Formal Definition

**Tool Execution** is the step where:
- The **actual Python function or tool is run**
- Using the **input arguments suggested by the LLM**
- During the **tool calling phase**

---

## Understanding Tool Execution with an Example

Suppose:
- The LLM suggests using a **multiply tool**
- It clearly specifies:
  - Value of `a` = 8
  - Value of `b` = 7

Now:
- You, as a programmer, call the multiply tool
- You pass exactly the same inputs (`a = 8`, `b = 7`)
- The tool executes
- You receive the result back

This act of **running the tool with LLM-provided inputs** is known as **Tool Execution**.

---

## Key Responsibility Split

- **LLM’s role**
  - Understand the query
  - Decide whether a tool is needed
  - Select the correct tool
  - Generate structured inputs for the tool

- **Programmer’s role**
  - Take the tool name and inputs from the LLM
  - Execute the tool
  - Collect the result

---

## Big Picture Flow (Till Now)

1. Tool Creation → Define what actions are possible  
2. Tool Binding → Make the LLM aware of available tools  
3. Tool Calling → LLM decides *which tool to use and with what inputs*  
4. Tool Execution → Programmer runs the tool using those inputs  

---

## Final Takeaway

Tool Execution is the step where:
- Thinking ends
- Action begins

The LLM **plans the action**,  
but the **actual execution happens in your code**.

This is what finally allows LLM-based agents to perform **real-world tasks**.

# A More Meaningful Example: Real-Time Currency Conversion Using Tools

Now, let’s look at a **more meaningful and practical example**.

Instead of using a simple multiplication example, we will build a **small but proper application**.  
The idea is to give our **LLM the power to perform real-time currency conversion**.

---

## Problem Context: Currency Conversion

- The world has many countries
- Every country has its **own currency**
- Currency conversion rates between countries **change continuously**
  - Daily
  - Hourly
  - Sometimes even minute by minute

For example:
- Indian Rupee (INR)
- US Dollar (USD)

Roughly:
- `1 USD ≈ 85 INR`

But this value:
- Is **not fixed**
- Keeps fluctuating throughout the day

---

## Why LLM Alone Is Not Enough

Suppose you ask an LLM:

> “How much is 80 USD in INR?”

The LLM **will give you an answer**, but:
- That answer is based on **historical or approximate data**
- The LLM **does not have access to real-time exchange rates**

So:
- The answer may sound correct
- But it is **not guaranteed to be accurate in real time**

This is a **serious limitation** when building real-world applications.

---

## Solution: Using a Tool for Real-Time Data

To solve this problem, we will **create a tool**.

### What Will This Tool Do?

- It will be a **Python function**
- Inside the function:
  - We will call a **currency exchange API**
- The API will:
  - Take a **source currency**
  - Take a **target currency**
  - Return the **current conversion factor**

Example:
- Source currency: USD
- Target currency: INR
- API returns the **latest exchange rate**

---

## Exchange Rate API

We will use an **Exchange Rate API** that:
- Accepts:
  - Base (source) currency
  - Target currency
- Instantly returns:
  - The **current conversion factor** between the two currencies

This factor represents the **real-time exchange rate**, not an old or estimated one.

---

## What This Enables

Once we have:
- The real-time conversion factor

We can:
- Convert **any amount**
- From **any currency**
- To **any other currency**
- In **real time**

This means:
- The LLM no longer relies on outdated knowledge
- Instead, it uses a **tool to fetch live data**
- And then performs accurate calculations

---

## Key Takeaway

- LLMs are great at reasoning and explaining
- But they **cannot access real-time information on their own**
- Tools bridge this gap
- By combining:
  - LLM reasoning
  - External APIs
- We can build **practical, real-world AI applications**

This currency converter is a perfect example of how **tool calling + tool execution** turns an LLM into a truly powerful agent.


## Tool Creation

```python
from langchain_core.tools import tool, InjectedToolArg
from typing import Annotated, Dict
import requests


# Tool 1: Fetch conversion rate from external API
@tool
def get_conversion_factor(base_currency: str, target_currency: str) -> Dict:
    """
    Fetch conversion rate between two currencies.

    Args:
        base_currency (str): Currency code to convert from (e.g., 'USD')
        target_currency (str): Currency code to convert to (e.g., 'INR')

    Returns:
        Dict: Contains conversion_rate and metadata
    """

    try:
        # Normalize input to uppercase (handles 'usd' → 'USD')
        base_currency = base_currency.upper()
        target_currency = target_currency.upper()

        # Construct API endpoint dynamically
        url = (
            f"https://v6.exchangerate-api.com/v6/"
            f"api_key/pair/{base_currency}/{target_currency}"
        )

        # Make HTTP request with timeout to avoid hanging
        response = requests.get(url, timeout=10)

        # Raise error for bad HTTP status (4xx / 5xx)
        response.raise_for_status()

        # Parse JSON response
        data = response.json()

        # Validate API-level success (not just HTTP success)
        if data.get("result") != "success":
            return {
                "error": "API failed",
                "details": data
            }

        # Return only relevant fields (agent-friendly response)
        return {
            "conversion_rate": data.get("conversion_rate"),
            "base_currency": base_currency,
            "target_currency": target_currency
        }

    except Exception as e:
        # Catch all errors (network, parsing, etc.)
        return {
            "error": str(e)
        }
```

```python
# Tool 2: Convert amount using injected conversion rate
@tool
def convert(
    base_currency_value: float,
    conversion_rate: Annotated[float, InjectedToolArg]
) -> float:
    """
    Convert amount using injected conversion rate.

    Args:
        base_currency_value (float): Amount in base currency
        conversion_rate (float): Injected conversion rate (from another tool)

    Returns:
        float: Converted value
    """

    try:
        # Core conversion logic: Amount × Rate
        converted_value = base_currency_value * conversion_rate

        # Round to 2 decimal places (standard for currency)
        return round(converted_value, 2)

    except Exception as e:
        # Handle unexpected calculation errors
        return {
            "error": str(e)
        }
```

## Tool Binding

```python
# Initialize the ChatOpenAI language model instance
llm = ChatOpenAI()

# Bind the currency-related tools to the LLM
# This enables the model to decide when and how to call these tools
# during a conversation based on the user's query
llm_with_tools = llm.bind_tools([
    get_conversion_factor,
    convert
])
```

```python
# Initialize the conversation history with the user's query
# This message acts as the starting point for the LLM interaction
messages = [
    HumanMessage(
        content=(
            "What is the conversion factor between INR and USD, "
            "and based on that can you convert 10 inr to usd"
        )
    )
]
```

```python
# Invoke the LLM with the current conversation history
# The model analyzes the user query and decides whether any tools
# need to be called based on the bound tools
ai_message = llm_with_tools.invoke(messages)

# Append the AI's response to the message history
# This response may contain tool_calls that will be executed next
messages.append(ai_message)
```

```python
from langchain_core.messages import ToolMessage
import json

# This variable will store the conversion rate retrieved from the first tool
# It is later reused when calling the 'convert' tool
conversion_rate = None


# Iterate over all tool calls generated by the LLM
# The LLM may decide to call one or multiple tools in sequence
for tool_call in ai_message.tool_calls:

    # --- Case 1: Fetch conversion factor ---
    # The model first calls this tool to get the exchange rate
    if tool_call["name"] == "get_conversion_factor":

        # Execute the tool with arguments provided by the LLM
        result = get_conversion_factor.invoke(tool_call["args"])
        
        # Extract and store conversion rate for downstream usage
        conversion_rate = result.get("conversion_rate")

        # Append the tool response back to the conversation history
        # This allows the LLM to "see" the tool output in the next step
        messages.append(
            ToolMessage(
                content=json.dumps(result),  # Tool output must be a string
                tool_call_id=tool_call["id"]  # Links response to the correct tool call
            )
        )


    # --- Case 2: Perform currency conversion ---
    # This step depends on the output of the previous tool
    elif tool_call["name"] == "convert":

        # Inject the conversion rate dynamically into tool arguments
        # (LLM does not automatically carry forward intermediate results)
        tool_call["args"]["conversion_rate"] = conversion_rate

        # Execute the conversion tool
        result = convert.invoke(tool_call["args"])

        # Append the result back to the message history
        # This completes the tool interaction cycle
        messages.append(
            ToolMessage(
                content=json.dumps(result),
                tool_call_id=tool_call["id"]
            )
        )
```

```python
# Invoke the LLM with full conversation history
final_response = llm_with_tools.invoke(messages)

# Safely handle different response formats
if isinstance(final_response.content, str):
    print(final_response.content)
elif isinstance(final_response.content, list):
    print(" ".join([str(part) for part in final_response.content]))
else:
    print(final_response)
```

```python
from langchain_openai import ChatOpenAI
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate


# --- Step 1: Initialize the LLM ---
# This is the core language model responsible for:
# - Understanding user queries
# - Deciding when to call tools
# - Generating the final response
llm = ChatOpenAI()


# --- Step 2: Define the Prompt Template ---
# This controls how input is structured before being sent to the LLM
# Components:
# - system: Sets behavior/personality of the assistant
# - human: Placeholder for user input
# - agent_scratchpad: Stores intermediate reasoning + tool interactions
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    ("human", "{input}"),
    ("placeholder", "{agent_scratchpad}")
])


# --- Step 3: Create Tool-Calling Agent ---
# This agent:
# - Uses the LLM to interpret queries
# - Automatically decides which tool to call (if needed)
# - Handles multi-step reasoning internally (ReAct-style)
agent = create_tool_calling_agent(
    llm=llm,
    tools=[get_conversion_factor, convert],
    prompt=prompt
)


# --- Step 4: Wrap Agent in Executor ---
# AgentExecutor manages the full execution lifecycle:
# - Runs the agent
# - Executes selected tools
# - Feeds tool outputs back into the agent
# - Produces the final response
#
# verbose=True → shows internal reasoning steps (debug + learning)
agent_executor = AgentExecutor(
    agent=agent,
    tools=[get_conversion_factor, convert],
    verbose=True
)
```

```python
# --- Step 6: Run the Agent ---

# Define the user input that will be passed to the agent
user_query = "Hi how are you?"

# Invoke the agent executor with the user input
# The agent will:
# - Analyze the query
# - Decide whether any tools are needed
# - Generate a final response accordingly
response = agent_executor.invoke({"input": user_query})
```

```python
from langchain.tools import tool

@tool
def convert(base_currency: str, target_currency: str, base_currency_value: float):
    """
    Convert a given amount from one currency to another.

    Args:
        base_currency: Source currency (e.g., INR)
        target_currency: Target currency (e.g., USD)
        base_currency_value: Amount to convert
    """

    # Fetch conversion rate internally
    rate_result = get_conversion_factor.invoke({
        "base_currency": base_currency,
        "target_currency": target_currency
    })

    conversion_rate = rate_result["conversion_rate"]

    converted_value = base_currency_value * conversion_rate

    return {
        "converted_value": converted_value,
        "conversion_rate": conversion_rate
    }
```

```python
# --- Step 6: Run the Agent with a meaningful query ---

# This query forces the agent to:
# 1. Call 'get_conversion_factor' to fetch exchange rate
# 2. Use that rate to call 'convert'
# 3. Return a final computed answer
user_query = "What is the conversion factor between INR and USD, and convert 500 INR to USD"

# Invoke the agent
response = agent_executor.invoke({"input": user_query})

# Print final result
print(response["output"])
```

```python
# %%writefile inr_to_usd.py
# -------------------- Imports --------------------

import requests
import json
from typing import Annotated

from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.tools import tool, InjectedToolArg


# -------------------- Tool 1: Fetch Conversion Factor --------------------

@tool
def get_conversion_factor(base_currency: str, target_currency: str) -> dict:
    """
    Fetches the currency conversion data between a base currency
    and a target currency using an external exchange rate API.
    """

    url = (
        f"https://v6.exchangerate-api.com/v6/"
        f"api_key/pair/{base_currency}/{target_currency}"
    )

    response = requests.get(url)
    return response.json()


# -------------------- Tool 2: Convert Currency --------------------

@tool
def convert(
    base_currency_value: int,
    conversion_rate: Annotated[float, InjectedToolArg]
) -> float:
    """
    Converts a base currency amount into a target currency amount
    using the provided conversion rate.
    """

    return base_currency_value * conversion_rate


# -------------------- Initialize LLM --------------------

llm = ChatOpenAI()

llm_with_tools = llm.bind_tools([
    get_conversion_factor,
    convert
])


# -------------------- Dynamic User Input --------------------

amount = int(input("Enter amount in INR: "))

messages = [
    SystemMessage(
        content=(
            "You are an intelligent financial assistant.\n"
            "When answering:\n"
            "1. First explain what you are doing\n"
            "2. Then show the conversion factor\n"
            "3. Then show the final converted value clearly\n"
            "4. Keep it structured and easy to read\n"
        )
    ),
    HumanMessage(
        content=(
            f"What is the conversion factor between INR and USD, "
            f"and convert {amount} INR to USD"
        )
    )
]


# -------------------- Step 1: LLM Decides Tool Calls --------------------

ai_message = llm_with_tools.invoke(messages)
messages.append(ai_message)


# -------------------- Step 2: Execute Tool Calls --------------------

conversion_rate = None

for tool_call in ai_message.tool_calls:

    if tool_call["name"] == "get_conversion_factor":

        tool_message = get_conversion_factor.invoke(tool_call)

        conversion_rate = json.loads(tool_message.content)["conversion_rate"]

        messages.append(tool_message)


# -------------------- Step 3: Deterministic Conversion (NO LLM) --------------------

if conversion_rate is not None:
    converted_value = amount * conversion_rate
else:
    raise ValueError("Conversion rate not fetched. Tool call failed.")


# -------------------- Step 4: Final Answer --------------------

messages.append(
    HumanMessage(
        content=(
            f"The conversion rate is {conversion_rate}. "
            f"I have already calculated that {amount} INR = {round(converted_value, 2)} USD. "
            f"Explain this properly."
        )
    )
)

final_response = llm.invoke(messages)

print("\n--- AI Response ---\n")
print(final_response.content)


# -------------------- Step 5: Structured Output --------------------

print("\n--- Structured Output ---\n")
print(f"Conversion Rate (INR → USD): {conversion_rate}")
print(f"Converted Amount: {amount} INR = {round(converted_value, 2)} USD")
```

```python
# -------------------- OPTIONAL: Agent-Based Version --------------------

from langchain_classic.agents import create_tool_calling_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate

# -------------------- Prompt --------------------

prompt = ChatPromptTemplate.from_messages([
    ("system",
     "You are a helpful financial assistant.\n"
     "Always use tools for currency conversion.\n"
     "Explain step-by-step:\n"
     "1. Fetch conversion rate\n"
     "2. Perform calculation\n"
     "3. Give final answer clearly\n"
    ),
    ("human", "{input}"),
    ("placeholder", "{agent_scratchpad}")
])

# -------------------- Create Agent --------------------

agent = create_tool_calling_agent(
    llm=llm,
    tools=[get_conversion_factor, convert],
    prompt=prompt
)

agent_executor = AgentExecutor(
    agent=agent,
    tools=[get_conversion_factor, convert],
    verbose=True   # shows reasoning steps
)

# -------------------- Run Agent --------------------

agent_response = agent_executor.invoke(
    {
        "input": f"What is the conversion factor between INR and USD and convert {amount} INR to USD"
    }
)

# -------------------- Output --------------------

print("\n--- Agent Final Output ---\n")
print(agent_response["output"])
```
