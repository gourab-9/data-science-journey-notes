# Long Short-Term Memory (LSTM) Networks

## 1. Core Idea of LSTM
- LSTMs are designed to solve one of the biggest limitations of vanilla RNNs:  
  **difficulty in maintaining long-term dependencies**.  
- In a standard RNN, there is only a **single hidden state** that tries to capture both **short-term context** and **long-term context**.  
- Mathematically, this often fails because **short-term information dominates** and important long-term context is lost (vanishing gradient problem).  

**LSTM introduces two types of memory:**
1. **Short-Term Memory (hidden state, hₜ)** – captures immediate context.
2. **Long-Term Memory (cell state, Cₜ)** – preserves important information across longer sequences.

---

## 2. Story Analogy to Understand LSTM
Imagine you are watching a long movie and at the end, you must decide:
- Was the movie *good*? → **Yes**
- Was the movie *bad*? → **No**

Your brain processes the movie **scene by scene** and decides what is **important enough to remember** for the final judgment.

### Example Flow
- Scene 1: *"The movie starts in a village where a kind doctor helps poor people."*  
  → You remember: “Village setting” (long-term context), “Doctor” (main character).  

- Scene 2: *"The doctor falls seriously ill."*  
  → You realize the doctor may no longer drive the story, so you **update** your memory.  

- Scene 3: *"The doctor’s daughter grows up and continues his work, even better than him."*  
  → You add “Daughter” as the new hero in long-term memory.  

- Scene 4: *"She opens a hospital and saves the village during a pandemic."*  
  → You keep her as the central character in memory for your final decision.

**Conclusion:**  
Like your brain in this example, LSTMs **decide dynamically what to keep, what to forget, and what to update** in memory.  

---

## 3. How LSTM Differs from RNN
- **RNN:**  
  - Maintains a single state → struggles with long-term dependencies.  
  - Example: remembers recent scenes but forgets important details from earlier in the movie.  

- **LSTM:**  
  - Maintains **two states**:  
    - Short-term memory (hidden state)  
    - Long-term memory (cell state)  
  - Introduces a **communication mechanism** between the two.  
  - Uses **gates** (special components) to decide what to keep, forget, or update.

---

## 4. Visual Intuition
- **RNN Architecture:** simple loop with a single state line.  
- **LSTM Architecture:**  
  - Two parallel lines (short-term and long-term memory).  
  - Extra components (gates) that regulate communication between them.  

At first glance, LSTM looks more complex, but the complexity is necessary to handle **long-term dependencies** effectively.

---

## 5. Key Takeaways
- LSTM introduces a **separate long-term memory** to overcome the limitations of RNNs.  
- It maintains **short-term + long-term states** simultaneously.  
- The core idea: **decide what information is important enough to keep, and what to discard**.  
- This is implemented using **three gates** (forget, input, and output gates) which will be explained in the next part.

# LSTM Gates and Their Roles

## 1. Why Gates in LSTM?
- Standard RNNs cannot effectively decide what information to **keep**, **forget**, or **use** at different time steps.  
- LSTM introduces **gates** — special neural network units that control the flow of information.  
- These gates allow LSTMs to dynamically update their **long-term memory (cell state)** and **short-term memory (hidden state)**.

---

## 2. The Three Gates in LSTM

### 🔹 Forget Gate
- **Role:** Decides what information from the **long-term memory** should be removed.  
- Example (fresh analogy):  
  - Imagine you are studying for exams and making notes.  
  - You initially keep every detail, but later realize some points are irrelevant.  
  - The forget gate is like your **eraser** – it removes the less useful information from your notes.  

---

### 🔹 Input Gate
- **Role:** Decides what new information should be **added to long-term memory**.  
- Continuing the analogy:  
  - While studying, you come across a very important formula.  
  - You decide to highlight and store it in your notes for the future.  
  - The input gate acts like a **highlighter/pen** – it ensures important new details are permanently written into memory.  

---

### 🔹 Output Gate
- **Role:** Decides what part of the memory (long-term + short-term) should be used to produce the **current output**.  
- Analogy:  
  - When answering a test question, you don’t use everything from your notes.  
  - You select only the **most relevant points** and write the final answer.  
  - The output gate acts like your **decision filter** – it controls what information influences the final prediction.  

---

## 3. Summary of Gate Functions
- **Forget Gate →** Removes unimportant information from long-term memory.  
- **Input Gate →** Adds new important information into long-term memory.  
- **Output Gate →** Uses current input + memory to decide the final output.  

---

## 4. Simplified View of LSTM as a Computer
- LSTM can be compared to a simple computer system:
  - **Input:** Current data (xₜ), previous short-term memory (hₜ₋₁), and previous long-term memory (Cₜ₋₁).  
  - **Processing:**  
    1. Update long-term memory → remove old info, add new info.  
    2. Create a new short-term memory for the next step.  
  - **Output:** The prediction (yₜ) and updated states (hₜ, Cₜ).  

Thus, at each step, LSTM:
1. Updates **long-term memory (Cₜ)**.  
2. Produces a new **short-term memory (hₜ)** for the next step.  

This is the **core functionality** of an LSTM cell.

## RNN vs LSTM
- In a simple **RNN**, only one flow of information exists: the **hidden state**.  
- In an **LSTM**, there are two flows:
  - **Cell State (C)** → represents long-term memory.  
  - **Hidden State (h)** → represents short-term memory.  

This additional cell state makes the LSTM architecture look more complex than RNN.

---

## Inputs to LSTM Cell
At time step \(t\), an LSTM cell receives three inputs:
1. **Previous Cell State** ($(C_{t-1}$))  
2. **Previous Hidden State** ($(h_{t-1}$))  
3. **Current Input** ($(x_t$))  

---

## Outputs from LSTM Cell
The LSTM cell produces two outputs:
1. **Updated Cell State** ($(C_t$))  
2. **Current Hidden State** ($(h_t$))  

---

## Internal Working
Inside the LSTM cell, two main tasks are performed:

1. **Update Long-Term Memory (Cell State):**  
   - Based on the current input and previous hidden state, the model decides:  
     - What to **remove** from memory.  
     - What to **add** into memory.  
   - Hence, the cell state is updated step by step.

2. **Generate Short-Term Memory (Hidden State):**  
   - After updating the cell state, the hidden state is calculated.  
   - This hidden state serves as the output for the current step and is passed forward.

---

## Key Point
The complex-looking LSTM design mainly exists to perform these two functions:  
1. **Update the cell state (long-term memory).**  
2. **Calculate the hidden state (short-term memory) for the next time step.**


# LSTM – Gates Concept

The most important concept in **LSTM** is the use of **gates**.  
Every detailed explanation of LSTM eventually revolves around these gates.

---

## Why Gates?
The LSTM architecture may look complex, but its main strength lies in these gates.  
They control the flow of information:  
- What to keep,  
- What to remove,  
- And what to output.  

---

## Types of Gates
1. **Forget Gate**  
   - Decides what part of the cell state (long-term memory) should be removed.  
   - If certain information is no longer useful, the forget gate drops it.

    
2. **Input Gate**  
   - Decides what new information should be added to the cell state.  
   - This ensures the long-term memory gets updated with relevant data.



3. **Output Gate**  
   - Decides what part of the memory should be exposed as the **hidden state** (short-term memory).  
   - This hidden state becomes the output of the current step.

---


# LSTM – Mathematical Representation of Components




Before we dive deeper into the three gates, let us first simplify how the different components of an LSTM cell look **mathematically**.

---

## 1. Cell State and Hidden State
- **Cell State ($(C_t$))** → represents long-term memory.  
- **Hidden State ($(h_t$))** → represents short-term memory.  
- **Mathematically:** Both are **vectors** (a collection of numbers).  

👉 At every step:  
- Input is a vector.  
- Output is also a vector.  
- The **dimension (size)** of these vectors always remains the same throughout the LSTM.  
This is a strict rule in LSTMs.

---

## 2. Input Representation ($(x_t$))
- At time step $(t$), the input $(x_t$) is also a **vector**.  
- Example: In a sentiment analysis task, we convert words into vectors (via one-hot encoding, embeddings, etc.).  
  - Suppose we have three words: "good", "bad", "night".  
  - One-hot encoding example:  
    - "good" → [1, 0, 0]  
    - "bad" → [0, 1, 0]  
    - "night" → [0, 0, 1]  
  - Here, each word becomes a vector, so the LSTM can process it numerically.

Thus, **$(x_t$) is simply the current input word/feature converted into a vector**.

---

## 3. Gate Vectors
Inside an LSTM cell, we calculate several intermediate vectors:

- **Forget Gate ($(f_t$))** → vector  
- **Input Gate ($(i_t$))** → vector  
- **Candidate Cell State ($(\tilde{C}_t$))** → vector  
- **Output Gate (\(o_t\))** → vector  

👉 All of these ($(f_t, i_t, \tilde{C}_t, o_t$)) are **vectors of the same dimension** as the hidden state.  
This ensures smooth mathematical operations inside the LSTM.

---

## 4. Key Rule
So far, we have **six important vectors**:
1. Cell State ($(C_t$))  
2. Hidden State ($(h_t$))  
3. Input ($(x_t$))  
4. Forget Gate ($(f_t$))  
5. Input Gate ($(i_t$))  
6. Output Gate ($(o_t$))  
7. Candidate Cell State ($(\tilde{C}_t$))  

**All of these vectors share the same shape (dimension).**  
This consistency is what allows the LSTM to update states and pass them forward without mismatch.

---


# LSTM – Mathematical Operations Inside the Cell

Before studying each gate in detail, let us understand the **basic mathematical operations** used inside an LSTM cell.  
In diagrams, these are often shown as **red-colored symbols**.

---

## 1. Pointwise Multiplication (Element-wise Multiplication)
- Requires two vectors of the same dimension.  
- Each element in one vector is multiplied with the corresponding element in the other.  

**Example:**  
$$
C_{t-1} = [4, 5, 6], \quad f_t = [1, 2, 3]
$$

Performing element-wise multiplication:  
$$
C_{t-1} \odot f_t = [4 \times 1, \; 5 \times 2, \; 6 \times 3] = [4, 10, 18]
$$

This is how **forget gate** removes or keeps parts of the cell state.

---

## 2. Pointwise Addition
- Again, requires vectors of the same size.  
- Each element is added with the corresponding element from another vector.  

**Example:**  
$$
[4, 5, 6] + [1, 2, 3] = [5, 7, 9]
$$

This is used when combining updated information with the previous memory.

---

## 3. Activation Functions
Inside gates, activation functions are applied to control the flow of information.

- **Sigmoid (σ):**  
  - Outputs values between **0 and 1**.  
  - Decides “how much to keep” or “how much to forget.”  
  - Example: In forget gate and input gate.  

- **Tanh (t):**  
  - Outputs values between **-1 and 1**.  
  - Helps regulate the scale of information (whether to add positive/negative influence).  
  - Example: In candidate cell state and hidden state updates.  

---

## 4. Neural Network Units in Gates
- Each gate in LSTM (forget, input, output) behaves like a **small neural network layer**.  
- Each has a set of neurons (units), and each neuron applies:
  - A weighted sum of inputs.  
  - Followed by an activation function (sigmoid or tanh).  

👉 The **number of units** in a gate determines the **dimension of the vectors** produced.  
This dimension is always consistent with the hidden state dimension of the LSTM.

---


# LSTM – Forget Gate






Now let us focus on the **Forget Gate**, one of the most important components of the LSTM architecture.  
Its role is to decide **what part of the past cell state should be kept** and **what should be removed**.

---

## 1. Inputs to Forget Gate
The forget gate takes **two inputs**:
1. **Current input** $(x_t$)  
2. **Previous hidden state** $(h_{t-1}$)

Both are combined and passed through a **neural network layer** with:
- Weights $(W_f$)  
- Bias $(b_f$)  
- Activation function = **Sigmoid**

Thus, the forget gate generates a vector $(f_t$), where each value lies between **0 and 1**.

---

## 2. Forget Gate Equation
$$
f_t = \sigma(W_f \cdot [h_{t-1}, x_t] + b_f)
$$

Where:
- $(\sigma$) is the sigmoid activation.  
- $([h_{t-1}, x_t]$) denotes concatenation of previous hidden state and current input.  
- $(f_t$) is a vector (same size as the cell state).

---

## 3. Updating Cell State
The forget gate decides how much of the previous memory $(C_{t-1}$) should continue.  
It does this via **pointwise multiplication**:

$$
C_t = f_t \odot C_{t-1}
$$

- If a value in $(f_t$) is **1**, that information is fully kept.  
- If it is **0**, that information is completely forgotten.  
- Values between 0 and 1 represent **partial forgetting**.

---

## 4. Example
Suppose:
$$
C_{t-1} = [4, 5, 6], \quad f_t = [0.5, 0.5, 0.5]
$$

Then:
$$
C_t = [4 \times 0.5,\; 5 \times 0.5,\; 6 \times 0.5] = [2, 2.5, 3]
$$

This means the forget gate **allowed 50% of the past memory to flow forward**.

---

## 5. Why “Gate”?
- The forget gate **controls the flow of information**—like a door that can let memory pass or block it completely.  
- Depending on the situation:
  - **100% pass** → $(f_t = 1$)  
  - **100% block** → $(f_t = 0$)  
  - **Partial pass** → $(0 < f_t < 1$)

👉 This selective control ensures that the LSTM does not carry unnecessary information for long, making it effective in handling **long-term dependencies**.

---



# LSTM – Input Gate





After the Forget Gate, the next important component in LSTM is the **Input Gate**.  
Its role is to **add new relevant information into the cell state** based on the current input and the previous hidden state.

---

## 1. Function of Input Gate
- **Forget Gate**: decides *what to remove* from the memory.  
- **Input Gate**: decides *what new information to add* to the memory.  

Thus, the input gate updates the long-term context by introducing important features extracted from the new input.

---

## 2. Steps in Input Gate Operation

The input gate works in **two stages**:

### (a) Candidate Cell State ($(\tilde{C}_t$))
We first calculate possible new values for the cell state using the current input $(x_t$) and previous hidden state $(h_{t-1}$):

$$
\tilde{C}_t = \tanh(W_c \cdot [h_{t-1}, x_t] + b_c)
$$

- Here, $(\tilde{C}_t$) represents **candidate values** that might be added.  
- $(\tanh$) ensures the values remain between $(-1$) and $(+1$).  
- It acts like a “proposal” for what could be stored.

---

### (b) Input Gate Vector ($(i_t$))
Next, we calculate a filter vector $(i_t$) that decides *how much of the candidate should actually be added*:

$$
i_t = \sigma(W_i \cdot [h_{t-1}, x_t] + b_i)
$$

- $(\sigma$) is the **sigmoid activation** that gives values between 0 and 1.  
- This acts as a **filter**:  
  - If close to 1 → information is added.  
  - If close to 0 → information is ignored.  

---

## 3. Updating the Cell State
Finally, the cell state is updated by combining:
- The **retained old information** from the forget gate, and  
- The **filtered new candidate values** from the input gate.

$$
C_t = f_t \odot C_{t-1} + i_t \odot \tilde{C}_t
$$

Where:
- $(f_t \odot C_{t-1}$) = old memory carried forward.  
- $(i_t \odot \tilde{C}_t$) = new information added.  

---

## 4. Example
Suppose:
- Previous cell state: $(C_{t-1} = [4, 5, 6]$)  
- Forget gate output: $(f_t = [0.5, 0.5, 0.5]$)  
- Candidate state: $(\tilde{C}_t = [1, 2, 3]$)  
- Input gate output: $(i_t = [0.7, 0.7, 0.7]$)  

Step 1: Forget part  
$$
f_t \odot C_{t-1} = [2, 2.5, 3]
$$

Step 2: Add new part  
$$
i_t \odot \tilde{C}_t = [0.7, 1.4, 2.1]
$$

Final cell state:  
$$
C_t = [2 + 0.7,\; 2.5 + 1.4,\; 3 + 2.1] = [2.7, 3.9, 5.1]
$$

---

## 5. Why is Input Gate Important?
- It ensures that **only meaningful new information enters the memory**.  
- Prevents random noise or irrelevant details from polluting the long-term state.  
- Solves the **vanishing gradient problem** by allowing controlled updates to memory.  

---

👉 At this stage, the LSTM has the ability to **retain old useful information** (via Forget Gate) and **add new relevant knowledge** (via Input Gate).  

Next, we will explore the **Output Gate**, which decides what part of the cell state is exposed as the hidden state.


# LSTM – Output Gate





We have now reached the final stage of the LSTM architecture: the **Output Gate**.  
Its role is to decide the **hidden state ($(h_t$))** at the current time step, which also serves as the output of the LSTM cell.

---

## 1. Function of Output Gate
- The **cell state ($(C_t$))** carries the long-term memory.  
- The **output gate** decides which part of this memory should be revealed as the hidden state.  
- Hidden state is important because it is passed to the next time step and often used for prediction.

---

## 2. Output Gate Equations

### (a) Gate Vector ($(o_t$))  
We first calculate the output gate filter vector $(o_t$) using the current input and the previous hidden state:

$$
o_t = \sigma(W_o \cdot [h_{t-1}, x_t] + b_o)
$$

- $(\sigma$) (sigmoid) outputs values between 0 and 1.  
- This decides how much of the memory should be passed on.  

---

### (b) Hidden State ($(h_t$))  
Then we combine the filtered memory with the transformed cell state:

$$
h_t = o_t \odot \tanh(C_t)
$$

- $(\tanh(C_t)$) scales the cell state values between $(-1$) and $(+1$).  
- The output gate vector $(o_t$) filters this information.  
- The result is the hidden state $(h_t$), which acts as **short-term memory + current output**.

---

## 3. Example

Suppose:
- Cell state: $(C_t = [2.7, 3.9, 5.1]$)  
- Output gate vector: $(o_t = [0.6, 0.8, 0.5]$)  

Step 1: Apply $(\tanh$) to cell state:  
$$
\tanh(C_t) \approx [0.99, 0.999, 0.9999]
$$

Step 2: Multiply with output gate:  
$$
h_t = o_t \odot \tanh(C_t) \approx [0.6, 0.8, 0.5]
$$

Thus, the hidden state exposes a **controlled portion** of the memory as output.

---

## 4. Why Output Gate is Important?
- Prevents the full cell state from being exposed (avoids noise).  
- Ensures **only relevant and filtered memory** is used at each step.  
- Provides the final output used in tasks like text prediction, sentiment analysis, or sequence classification.  

---

## 5. Putting It All Together
1. **Forget Gate**: removes unnecessary old information.  
2. **Input Gate**: adds new relevant information.  
3. **Output Gate**: decides what to reveal as the hidden state.  

This combination allows LSTMs to effectively handle **long-term dependencies** while avoiding the **vanishing gradient problem** that affects simple RNNs.

# Next-Word Predictor with LSTM 

## What is a “next-word predictor”?
A model that suggests the **next word** while you’re typing. You’ve seen it in:
- Smartphone keyboards (suggested words as you type)
- Email editors (e.g., “smart compose” style suggestions)
- Code editors (auto-completion)

These save effort by predicting likely continuations of your text.

---

## What we’ll build
A **Next-Word Predictor** using an **LSTM**.  
Plan:
1. Take a small text document as our dataset (enough for a demo).
2. Convert the raw text into a **supervised learning** dataset of *(input sequence → next word)* pairs.
3. Train an LSTM to predict the next token.
4. Use the trained model to generate words.

> Note: For strong results you’d normally use **much larger** text. Here we keep it small to demonstrate the full pipeline end-to-end.

---

## Why LSTM here?
LSTMs are designed for **sequences**. They remember important context over time and decide what to **keep**, **add**, or **expose** (via forget, input, and output gates), which is crucial for predicting the next word correctly.

---

## Turning text generation into supervised learning
We transform free-form text into labelled examples.  
Given a sentence:

> “hi my name is **Gourab**”

we create training pairs by **sliding** through the sentence:

- Input: `["hi"]` → Output: `"my"`
- Input: `["hi", "my"]` → Output: `"name"`
- Input: `["hi", "my", "name"]` → Output: `"is"`
- Input: `["hi", "my", "name", "is"]` → Output: `"Gourab"`

Repeat this for **every sentence** in the corpus to build a large set of (input, next-word) pairs.

---

## From words to numbers (tokenization)
Neural nets work with numbers, not raw text.  
Assign each distinct word an **integer ID** (a vocabulary):
---
```text
hi → 1
my → 2
name → 3
is → 4
Gourab → 5
```
---


Now the earlier pairs become numeric:

- Input: `[1]` → Output: `2`
- Input: `[1, 2]` → Output: `3`
- Input: `[1, 2, 3]` → Output: `4`
- Input: `[1, 2, 3, 4]` → Output: `5`

You’ll store inputs as **sequences of integers** (often padded to a fixed length) and outputs as the **next token’s ID** (or one-hot/softmax target).

---

## What the LSTM learns to do
- Read an input sequence of word IDs.
- Produce a probability distribution over the vocabulary for the **next word**.
- During inference, you can:
  - pick the top-1 word (greedy), or
  - sample from the distribution (for more variety),
  - then append that word to the input and **repeat** to generate longer text.

---

## Minimal, implementation-level checklist

1. **Collect text**
   - Use your chosen document as the dataset (demo-sized is fine for now).

2. **Tokenize**
   - Split into words/tokens consistently (lowercasing, handling punctuation/spaces).
   - Build a vocabulary (word → integer ID).
   - Convert the entire corpus to a list of IDs.

3. **Build training pairs**
   - For each sentence, generate sliding windows:
     - `X`: sequences (e.g., length `n`)  
     - `y`: the next token ID after each sequence
   - Pad sequences to fixed length if needed.

4. **Model outline (concept)**
   - Embedding layer (maps IDs → dense vectors)
   - LSTM layer(s)
   - Dense layer with softmax over the vocabulary

5. **Train**
   - Loss: categorical cross-entropy
   - Target: the next word’s class (ID)
   - Evaluate with accuracy or top-k accuracy.

6. **Predict / Generate**
   - Feed a seed text (e.g., “hi my name is”)
   - Get next word, append, and iterate.

---

````python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def sample_with_temperature(preds, temperature=1.0):
    preds = np.asarray(preds).astype("float64")
    preds = np.log(preds + 1e-9) / temperature
    exp_preds = np.exp(preds)
    preds = exp_preds / np.sum(exp_preds)
    return np.random.choice(len(preds), p=preds)

def generate_text(seed_text, next_words=50, max_sequence_len=56, temperature=0.7):
    output_text = seed_text
    
    for _ in range(next_words):
        # tokenize current text
        token_list = tokenizer.texts_to_sequences([output_text])[0]
        # pad sequence
        token_list = pad_sequences([token_list], maxlen=max_sequence_len, padding='pre')
        # predict probabilities
        predicted_probs = model.predict(token_list, verbose=0)[0]
        # sample word index with temperature
        predicted_index = sample_with_temperature(predicted_probs, temperature)
        
        # get word from index
        predicted_word = None
        for word, index in tokenizer.word_index.items():
            if index == predicted_index:
                predicted_word = word
                break
        
        if predicted_word is None:
            break
        
        # append predicted word
        output_text += " " + predicted_word
        
        # optional: stop if punctuation appears
        if predicted_word in [".", "!", "?"]:
            break
    
    return output_text
````

````python
seed = "Prince Orion and his companions enter the Forest of Whispers"
generated = generate_text(seed_text=seed, next_words=50, temperature=0.7)
print(generated)
````
