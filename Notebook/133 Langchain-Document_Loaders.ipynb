# Introduction to RAG-Based Applications

RAG (Retrieval-Augmented Generation) combines **information retrieval** with **language generation** to handle queries where the LLM alone may not have sufficient knowledge.

### Why RAG is useful
- LLMs cannot answer questions about **real-time events** or **personal/private data**.
- Large documents or specialized company data may not fit within standard LLM context windows.
- RAG allows integrating an **external knowledge base** with the LLM to provide accurate and grounded answers.

---

# Key Components of RAG Applications

A RAG-based LLM application generally consists of four essential components:

1. **Document Loaders**  
   Load documents from various sources such as PDFs, text files, web pages, or CSVs.

2. **Text Splitters**  
   Break large documents into smaller, manageable chunks for processing.

3. **Vector Databases**  
   Store embeddings for the document chunks, enabling fast similarity searches.

4. **Retrievers**  
   Query the vector database to fetch relevant documents that can guide the LLM’s response.

> These components form the backbone of most RAG architectures.

---

# Focus: Document Loaders

- LangChain provides a wide variety of document loaders.
- The underlying principle of document loaders is **consistent** across formats.
- Four commonly used loaders:
  - **Text Loader**
  - **PDF Loader**
  - **Web Loader**
  - **CSV Loader**
- Once one loader is understood, the others are easy to work with.

---

# How RAG Works

1. A **knowledge base** is connected to the LLM.
2. The user sends a **query**.
3. The system uses **retrievers** to find relevant documents.
4. The LLM generates a **grounded response** using the retrieved documents as context.

---

# Benefits of RAG

- **Up-to-date information**: Access latest data without retraining the LLM.
- **Privacy**: Use sensitive data locally without uploading to public LLMs.
- **Handling large documents**: Split and process large datasets efficiently.

---

# Summary

- RAG integrates external knowledge with LLMs for accurate, contextual answers.
- Document loaders are the **first and critical step** in a RAG workflow.
- Understanding loaders, splitters, vector databases, and retrievers enables building full RAG-based applications.


# Document Loaders in LangChain

## What are Document Loaders?

- **Definition:** Document loaders are components in LangChain used to **load data from various sources into a standardized format**, usually as **Document objects**.  
- These Document objects can then be used for **chunking, embeddings, retrieval, and generation**.

## Why are they important?

- In RAG (Retrieval-Augmented Generation) applications, data can come from **multiple sources**, such as:  
  - PDFs  
  - Text files  
  - Databases  
  - Cloud storage  
- LangChain ensures that data from **any source** can be converted into a **common format**, so it can work seamlessly with other components like chunkers, vector stores, and LLMs.

## Standardized Format: Document Object

Each `Document` object in LangChain has two main parts:

1. **Page content** – The actual textual data.  
2. **Metadata** – Information about the source, such as:  
   - File path or source URL  
   - Creation date  
   - Last modified date  
   - Author or origin  

**Key point:** Regardless of the loader used, the output is always in this standardized `Document` format.

---

## Types of Document Loaders

### 1. Text Loader

- **Purpose:** Simplest document loader in LangChain.  
- **Function:** Reads **text files** and converts them into `Document` objects.  
- **Use cases:**  
  - Processing **log files**  
  - Extracting **code snippets**  
  - Reading **transcripts** (e.g., meeting or video transcripts)  

**Example Intuition:**  
If you have a transcript of a meeting saved as `meeting.txt`, using a Text Loader will fetch the text and return a `Document` object with both page content and metadata. This object can then be directly passed to chunking, embedding, or retrieval steps in a RAG pipeline.

---

## Notes

- All Document Loaders in LangChain are available in the **`langchain_community` package**.  
- Once you understand **one loader**, most others work similarly because the **output is always a standardized `Document`**.


````python
from langchain_community.document_loaders import TextLoader
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
import os

# -------------------------------
# Load environment variables (like API keys)
# -------------------------------
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# -------------------------------
# Initialize OpenAI Chat model
# -------------------------------
model = ChatOpenAI(api_key=api_key)

# -------------------------------
# Define a prompt template
# -------------------------------
# This template will ask the model to summarize a given poem
prompt = PromptTemplate(
    template='Write a summary for the following poem - \n {poem}',
    input_variables=['poem']
)

# -------------------------------
# Initialize output parser
# -------------------------------
# Ensures the model output is returned as a simple string
parser = StrOutputParser()

# -------------------------------
# Load text data using TextLoader
# -------------------------------
# TextLoader reads a local text file and converts it into a standardized Document object
loader = TextLoader('cricket.txt', encoding='utf-8')

# Load the document(s) into a list
docs = loader.load()
````

````python
# -------------------------------
# Inspect the loaded documents
# -------------------------------
print(type(docs))          # Shows that docs is a list of Document objects
````

````python
print(docs)
````

````python
print(len(docs))           # Number of documents loaded
````

````python
print(docs[0].page_content) # Actual content of the first document
````

````python
print(docs[0].metadata)    # Metadata associated with the document (like source, creation date)
````

````python
# -------------------------------
# Create a chain combining PromptTemplate, model, and parser
# -------------------------------
# Using the pipe syntax (LangChain Expression Language) to create a sequential chain
chain = prompt | model | parser
````

````python
# -------------------------------
# Invoke the chain
# -------------------------------
# Pass the page content of the first document as input to get a summary
print(chain.invoke({'poem': docs[0].page_content}))
````

## PyPDFLoader (PDF Document Loader)

### What is PyPDFLoader?

`PyPDFLoader` is one of the **most commonly used document loaders** in LangChain.  
Its responsibility is simple and focused:

- Read **PDF files**
- Convert them into LangChain’s **standard `Document` objects**

This makes PDF data usable in downstream steps like:
- Text splitting
- Embedding generation
- Retrieval
- LLM-based question answering

---

### Page-by-Page Processing (Key Concept)

The most important characteristic of `PyPDFLoader` is:

> **It works on a page-by-page basis**

This means:

- If a PDF has **25 pages**
- The loader will return **25 `Document` objects**
- Each `Document` corresponds to **one page of the PDF**

So after loading:
- You don’t get one large document
- You get a **list of documents**, one per page

This is extremely useful because:
- Pages are already naturally chunked
- You get better control over embeddings and retrieval
- Long PDFs become easier to manage

---

### Structure of Each Document Object

Each page-level `Document` contains:

#### 1. `page_content`
- The actual **text extracted from that page**
- This is what gets passed to text splitters, embeddings, or LLMs

#### 2. `metadata`
- Additional information about the page, such as:
  - Source file path
  - Page number
  - Other loader-specific details

This metadata becomes very useful later during:
- Debugging
- Source attribution
- Showing references in answers

---

### Internal Working

- `PyPDFLoader` internally uses the **PyPDF library**
- It extracts text directly from the PDF structure

Because of this, it works best when:
- The PDF is **text-based**
- The content is selectable (not images)

---

### Limitations

`PyPDFLoader` is **not ideal** for:

- Scanned PDFs (image-based PDFs)
- PDFs created from photographs
- PDFs with very complex layouts (tables, multi-column designs)

In such cases:
- Text extraction quality degrades
- You may need OCR-based loaders instead

---

### When Should You Use PyPDFLoader?

Use `PyPDFLoader` when:

- Your PDFs are **clean, textual documents**
- Examples:
  - Research papers
  - Reports
  - Documentation
  - Manuals
  - Articles

Avoid it when:
- PDFs are scanned
- Text is embedded inside images

---

### Mental Model (Simple Intuition)

Think of `PyPDFLoader` like this:

- **Input:** One PDF file  
- **Processing:** Split by pages  
- **Output:**  

```text
[
Document(page 1),
Document(page 2),
Document(page 3),
...
]
```

Each page becomes an independent unit that LangChain can easily process.

---

This page-wise standardization is exactly why `PyPDFLoader` fits naturally into
**RAG pipelines** and large-document workflows.

## Directory Loader

So far, we have seen how to load **a single text file** or **a single PDF file** into LangChain.  
However, in real-world RAG applications, data rarely comes as just one file.

Most of the time:
- You have a **folder**
- Inside that folder, there are **multiple files**
  - multiple PDFs  
  - multiple text files  
  - or even a mix of both  

Manually loading each file one by one is neither scalable nor practical.

---

## What is a Directory Loader?

A **Directory Loader** is a document loader designed to handle this exact scenario.

**Core idea:**
- Instead of loading one file at a time
- You point the loader to a **directory (folder)**
- It automatically loads **all matching files inside that directory**
- Each file is converted into one or more **Document objects**

---

## Why Directory Loader Exists

LangChain introduces the Directory Loader to solve three problems:

- **Scalability**  
  Easily handle tens, hundreds, or thousands of documents.

- **Automation**  
  No need to manually specify file paths for every document.

- **Consistency**  
  All documents are converted into the same standardized `Document` format, making them ready for:
  - text splitting  
  - embedding  
  - vector storage  
  - retrieval  

---

## How It Works Conceptually

Think of the Directory Loader as a **wrapper around other loaders**.

High-level flow:
1. You provide a directory path
2. You specify which loader to use (e.g., TextLoader, PyPDFLoader)
3. The Directory Loader:
   - iterates over all files in the directory
   - applies the chosen loader to each file
4. You receive a **single list of Document objects**

---

## Example Use Cases

### When Directory Loader is the Right Choice

- You have:
  - a folder of PDFs (reports, manuals, research papers)
  - a folder of text files (logs, transcripts, notes)
- You are building:
  - a RAG system
  - a document-based chatbot
  - an internal knowledge assistant

### Typical Scenarios

- Company documentation stored as multiple PDFs
- Logs or transcripts saved as `.txt` files
- Knowledge bases organized by folders

---

## Key Takeaway

The **Directory Loader** allows you to:
- Load **multiple documents at once**
- From a **single directory**
- Using a **single loader configuration**

This makes it a foundational tool for building **real-world RAG pipelines**, where data almost always exists in bulk rather than as a single file.


````python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader

loader = DirectoryLoader(
    path='books',
    glob='*.pdf',
    loader_cls=PyPDFLoader
)

docs = loader.lazy_load()
````

## `load()` vs `lazy_load()` in LangChain Document Loaders

When working with document loaders in LangChain, you will often see two methods:

- `load()`
- `lazy_load()`

Both are used to read data from a source and convert it into `Document` objects, but **they differ in how and when the data is loaded into memory**.

Understanding this difference is important for performance and scalability.

---

## `load()`

### What it does

- Reads **all documents at once**
- Loads the **entire dataset into memory**
- Returns a **list of `Document` objects**

### Mental model

Think of `load()` as:
> “Load everything first, then start working.”

---

### Behavior

- Blocking operation
- All files/pages are processed before you get any output
- Memory usage grows with dataset size

---

### When to use `load()`

Use `load()` when:

- Dataset is **small or medium**
- You are working locally
- You want simplicity and quick experimentation
- You need random access to documents

---

### Simple example (conceptual)

- A directory with 5 PDFs
- Each PDF has 10 pages  
→ `load()` returns **50 Document objects at once**

---

### Pros

- Simple and easy to use
- Predictable behavior
- Works well for small datasets

### Cons

- High memory usage for large datasets
- Slow startup for big folders
- Not suitable for streaming or very large corpora

---

## `lazy_load()`

### What it does

- Loads documents **one at a time**
- Returns an **iterator / generator**
- Documents are created **only when needed**

### Mental model

Think of `lazy_load()` as:
> “Load documents only when I ask for the next one.”

---

### Behavior

- Non-blocking style iteration
- Lower memory usage
- Ideal for large datasets

---

### When to use `lazy_load()`

Use `lazy_load()` when:

- Dataset is **large**
- Files are heavy (large PDFs, many pages)
- You want to process documents incrementally
- You are embedding or indexing documents in batches

---

### Simple example (conceptual)

- A directory with 1,000 PDFs
- Each PDF has 20 pages  
→ Documents are yielded **page by page**, not all at once

---

### Pros

- Memory-efficient
- Scales to large datasets
- Works well with batch processing

### Cons

- Slightly more complex usage
- No direct random access
- You must iterate to consume documents

---

## Key Differences at a Glance

| Aspect | `load()` | `lazy_load()` |
|------|--------|--------------|
| Loading style | Eager | Lazy |
| Memory usage | High | Low |
| Return type | List of Documents | Iterator / Generator |
| Startup time | Slower for large data | Faster |
| Best for | Small datasets | Large datasets |
| Streaming support | ❌ | ✅ |

---

## Practical Rule of Thumb

- **Exploration, demos, small data** → `load()`
- **Production RAG pipelines, large corpora** → `lazy_load()`

---

## How This Fits in RAG Pipelines

In real-world RAG systems:

- `lazy_load()` is commonly used with:
  - text splitters
  - embedding models
  - vector databases
- Documents are processed **in chunks**, not all at once

This makes pipelines:
- faster
- more memory-efficient
- more production-ready

---

## Final Intuition

- `load()` = **eager, simple, memory-heavy**
- `lazy_load()` = **lazy, scalable, production-friendly**


## Web-Based Document Loaders in LangChain

Web-based loaders are used when your data **lives on the internet** instead of local files.  
They fetch content from URLs and convert it into LangChain’s standard **`Document` objects**, so the data can be chunked, embedded, retrieved, and used in RAG pipelines.

---

## Why Web Loaders Exist

Real-world knowledge often comes from:
- Documentation websites
- Blogs and articles
- Public web pages
- Internal company portals

Web loaders solve one core problem:

> **Convert unstructured web content into structured `Document` objects.**

---

## Common Web-Based Loaders

LangChain provides multiple loaders depending on **how complex the web page is**.

---

## 1. `WebBaseLoader`

### What it does

- Loads content from one or more URLs
- Extracts **main text** from HTML
- Removes most navigation, scripts, and styling

### Mental model

> “Give me the readable text from this webpage.”

---

### When to use

- Blogs
- Articles
- Documentation pages
- Clean, text-heavy websites

---

### Strengths

- Simple
- Works for most standard pages
- Good default choice

---

### Limitations

- Not ideal for:
  - JavaScript-heavy pages
  - Complex layouts
  - Websites requiring authentication

---

### Typical use case

- Indexing documentation pages for a chatbot
- Loading blog posts into a knowledge base

---

## 2. `BSHTMLLoader` (BeautifulSoup-based)

### What it does

- Parses HTML using BeautifulSoup
- Gives you **more control** over extraction

### Mental model

> “Let me decide which HTML elements I want.”

---

### When to use

- You need specific tags (e.g., `<article>`, `<main>`)
- Page structure is known
- You want cleaner content than generic extraction

---

### Strengths

- Precise control
- Cleaner output for structured pages

---

### Limitations

- Requires understanding of HTML structure
- Slightly more setup

---

## 3. `SitemapLoader`

### What it does

- Reads a **website sitemap**
- Automatically loads **multiple URLs**

### Mental model

> “Load an entire website by following its sitemap.”

---

### When to use

- Large documentation sites
- Blogs with many pages
- Knowledge bases with a sitemap.xml

---

### Strengths

- Scales well
- Automated multi-page loading

---

### Limitations

- Sitemap must exist and be accessible
- Not all pages may be relevant

---

## 4. `RecursiveUrlLoader`

### What it does

- Crawls links starting from a root URL
- Follows links recursively

### Mental model

> “Start here and keep following links.”

---

### When to use

- Internal documentation
- Knowledge portals
- Controlled websites

---

### Strengths

- Flexible crawling
- Good for structured internal sites

---

### Limitations

- Risk of crawling unwanted pages
- Needs depth and domain limits

---

## Choosing the Right Web Loader

| Scenario | Recommended Loader |
|--------|-------------------|
| Single article or blog | `WebBaseLoader` |
| Structured HTML pages | `BSHTMLLoader` |
| Full documentation site | `SitemapLoader` |
| Internal site crawling | `RecursiveUrlLoader` |

---

## How Web Loaders Fit in RAG

Typical flow:

1. Load web content using a web loader  
2. Convert to `Document` objects  
3. Chunk the text  
4. Generate embeddings  
5. Store in vector database  
6. Retrieve and generate answers

---

## Key Takeaway

Web loaders allow LangChain to **treat the internet as a data source**, just like PDFs or text files.

- Simple pages → `WebBaseLoader`
- Structured control → `BSHTMLLoader`
- Large sites → `SitemapLoader`
- Crawling → `RecursiveUrlLoader`

---

## Practical Insight

In production systems:
- Always clean and filter web content
- Avoid loading navigation-heavy pages
- Combine with text splitters for best results

## CSV Loader in LangChain

The **CSV Loader** is used when your data is stored in **tabular form** (rows and columns) and you want to bring it into LangChain as standardized `Document` objects for RAG-based applications.

---

## Why CSV Loader Is Needed

Many real-world datasets exist as:
- CSV exports from databases
- Logs and reports
- Product catalogs
- User records
- Analytical datasets

LLMs cannot directly understand tables.  
The CSV loader **converts each row into text**, making it usable for embeddings and retrieval.

---

## What CSV Loader Does Internally

- Reads a `.csv` file
- Converts each **row** into a `Document`
- Stores:
  - **Row content** → `page_content`
  - **Row metadata** → column values, row index, source file

---

## Mental Model

> **One row = one Document**

Each row becomes a small, searchable knowledge unit.

---

## When to Use CSV Loader

Use CSV loader when:
- Data is structured in rows
- Each row represents a meaningful entity
- You want row-level retrieval

### Examples
- Employee records
- Customer transactions
- Product descriptions
- FAQ datasets

---

## When NOT to Use CSV Loader

Avoid CSV loader when:
- CSV contains large unstructured text blocks
- You need column-wise reasoning
- Relationships between rows are critical

In such cases, preprocessing may be required.

---

## Basic Workflow

1. Load CSV file
2. Convert rows into `Document` objects
3. Chunk (if needed)
4. Create embeddings
5. Store in vector database
6. Retrieve relevant rows during query time

---

## Metadata Handling

Each `Document` includes:
- Source file path
- Row number
- Column-value mappings (optional)

This helps:
- Trace answers back to rows
- Filter documents during retrieval

---

## Common Scenarios & Best Practices

### Small CSV (hundreds to thousands of rows)
- Load directly
- Use row-level documents

### Large CSV (millions of rows)
- Use lazy loading
- Pre-filter rows
- Consider database loaders instead

---

## CSV Loader vs Text Loader

| Feature | CSV Loader | Text Loader |
|------|-----------|------------|
| Structure awareness | ✅ Yes | ❌ No |
| Row-based documents | ✅ Yes | ❌ No |
| Metadata support | Strong | Basic |
| Best for tables | ✅ | ❌ |

---

## CSV Loader vs Database Loader

| Scenario | Preferred |
|--------|----------|
| Static CSV file | CSV Loader |
| Frequently changing data | Database Loader |
| Very large datasets | Database Loader |

---

## Key Takeaways

- CSV Loader is ideal for **structured, tabular data**
- Each row becomes a retrievable knowledge unit
- Best used when rows are independent and meaningful
- Scales well for small to medium datasets

---

## Intuition to Remember

> **Text → paragraphs**  
> **PDF → pages**  
> **CSV → rows**

That mental mapping helps you choose the right loader quickly.


````python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(file_path='Social_Network_Ads.csv')

docs = loader.load()

print(len(docs))
print(docs[1])
````

## Document Loader Categories in LangChain

LangChain supports document loading from multiple sources.  
At a high level, all document loaders fall into the following **categories**:

### 1. Web-based Loaders
- Load content from URLs, websites, sitemaps, and crawled web pages.

### 2. PDF Loaders
- Load and parse PDF documents using different parsing strategies.

### 3. Cloud Storage Loaders
- Load documents from cloud storage providers (object storage, drives, buckets).

### 4. Social Media Loaders
- Load data from social platforms and community sources.

### 5. Messaging / Chat Loaders
- Load conversation data from chat and messaging platforms.

### 6. Productivity & Collaboration Tool Loaders
- Load data from tools used for documentation, communication, and project management.

### 7. Common File Type Loaders
- Load standard file formats such as text, CSV, JSON, HTML, etc.

---

### Quick Summary
- **Total loader categories:** 7  
- **Purpose:** Bring data from different sources into LangChain  
- **Output format:** Standard `Document` objects (content + metadata)
