## What Are Chains and Why Do We Need Them?

Let’s understand this in very simple terms.

Whenever you build an **LLM-based application**, it is never a single step.  
Your application is made up of **multiple small steps**.

### A Simple Example Application

Even the simplest LLM application has at least **three steps**:

1. Take input (prompt) from the user  
2. Send the prompt to the LLM  
3. Process the LLM response and show it to the user  

So even in a basic app, you already have multiple steps.

---

## The Problem with Manual Execution

If you execute each step **individually**, the process becomes very manual:

- Design the prompt using a prompt template  
- Manually invoke the prompt  
- Manually send it to the LLM  
- Receive the response  
- Extract the content from metadata  
- Display it to the user  

This approach:
- Works for small demos  
- Becomes **hard to manage** for large or complex applications  

As complexity increases, this manual handling becomes a problem.

---

## How Chains Solve This Problem

**Chains provide a pipeline-based approach.**

A chain allows you to **connect multiple steps together** so that:

- Output of Step 1 → becomes Input of Step 2  
- Output of Step 2 → becomes Input of Step 3  
- And so on…

You define the pipeline **once**, and after that:

- You only provide input to the **first step**
- The entire pipeline executes automatically
- You directly receive the **final output**

No need to manually manage intermediate steps.

---

## Key Idea: Automatic Data Flow

The biggest advantage of chains is **automatic data flow**:

- First component runs
- Its output automatically feeds into the next component
- This continues until the final step

This makes your application:
- Cleaner
- More modular
- Easier to maintain
- Easier to scale

---

## Why Chains Are So Powerful

Chains are not limited to simple linear pipelines.

Using chains, you can build:

### 1. Sequential (Linear) Chains
- One step after another
- Most common and easiest to understand

### 2. Parallel Chains
- Multiple chains running at the same time
- Useful when you want parallel processing

### 3. Conditional Chains
- Execution depends on conditions
- Different inputs can trigger different chains

This flexibility allows you to build **very complex applications** in a clean and structured way.

---

## Final Takeaway

- LLM applications are made of multiple small steps  
- Managing these steps manually does not scale well  
- **Chains help you connect these steps into pipelines**  
- Once built, chains automate execution from start to end  
- This is the core reason why chains are one of the most important components in LangChain  

In the **next video**, we will go deeper into:

```python


from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os


load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')


prompt = PromptTemplate(
    template='Generate 5 interesting facts about {topic}',
    input_variables=['topic']
)

model = ChatOpenAI(api_key = api_key)

parser = StrOutputParser()

chain = prompt | model | parser

result = chain.invoke({'topic':'Career in Data Science'})

print(result)
```

## Sequential Chain with Multiple LLM Calls (Two-Step LLM Pipeline)

Now, let’s build something **slightly more complex** than the previous example.

This will still be a **sequential chain**, but in this case, we will **call the LLM twice** within the same chain.

---

## Problem Statement

We want to build an application with the following flow:

1. **User provides a topic**  
   - Example: *Cricket*

2. **First LLM Call – Detailed Report Generation**
   - The user-provided topic is sent to the LLM
   - The prompt asks the LLM to generate a **detailed report** on that topic
   - The LLM processes the prompt and returns a **long, detailed report**

3. **Second LLM Call – Key Points Extraction**
   - The detailed report generated in Step 2 is sent **again** to the same LLM
   - A second prompt instructs the LLM to:
     - Extract the **top 5 most important points** from the report
   - This step effectively **summarizes** the detailed report

---

## What Is Happening Conceptually?

- The output of the **first LLM call** (detailed report)  
  becomes the **input of the second LLM call**
- Each step runs **one after the other**
- This makes it a **sequential chain**

---

## Why This Example Is More Complex

Compared to the previous example:

- We are no longer calling the LLM just once
- We are chaining **two LLM interactions**
- The chain is **longer**
- Each step depends on the output of the previous step

However, the execution logic is still **step-by-step (sequential)**:
- Step 1 executes first
- Step 2 executes only after Step 1 is completed

---

## Key Takeaway

- Sequential chains can include **multiple LLM calls**
- Outputs can be reused and fed back into the same or another LLM
- This pattern is extremely common in real-world applications:
  - Report generation → summarization
  - Answer generation → refinement
  - Long response → concise insights

This example shows how chains help manage **multi-step LLM workflows** cleanly and efficiently.


```python
# Sequential Chain.py

from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os


load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

prompt1 = PromptTemplate(
    template='Generate a detailed report on {topic}',
    input_variables=['topic']
)

prompt2 = PromptTemplate(
    template='Generate a 5 pointer summary from the following text \n {text}',
    input_variables=['text']
)

model = ChatOpenAI(api_key= api_key)

parser = StrOutputParser()

chain = prompt1 | model | parser | prompt2 | model | parser

result = chain.invoke({'topic': 'Unemployment in India'})

print(result)
```

```python
chain.get_graph().print_ascii()
```

## Parallel Chains in LangChain (Conceptual Explanation)

Now, let’s build a **parallel chain**, which is a slightly advanced but very interesting concept.

---

## Problem Statement

We want to build an application where:

- The **user provides a large text document**
  - Example: A detailed document explaining *Linear Regression* in Machine Learning

From this **same document**, we want to generate **two different outputs**:

1. **Notes** from the document  
2. **A Quiz** based on the document  

Finally, we want to **combine both outputs** and show them to the user.

---

## Why Parallel Chains Are Needed

Both tasks:
- Generating **notes**
- Generating a **quiz**

are **independent of each other**.

That means:
- Notes generation does not depend on quiz generation
- Quiz generation does not depend on notes generation

So instead of doing them one after the other (sequentially), we can do them **at the same time (in parallel)**.

This is where **parallel chains** come into the picture.

---

## High-Level Architecture

1. **User Input**
   - User provides a large document as input

2. **Parallel Processing**
   - The same document is sent to:
     - **Chain 1** → Generates structured notes
     - **Chain 2** → Generates a quiz (questions based on the document)
   - Both chains run **simultaneously**

3. **Merging Outputs**
   - Outputs from both chains (notes + quiz) are collected
   - A final step combines them into a single response

4. **Final Output**
   - User sees:
     - Notes derived from the document
     - Quiz generated from the same document

---

## Key Idea Behind Parallel Chains

- One input can be processed by **multiple chains at the same time**
- Each chain focuses on a **different task**
- This improves:
  - Performance
  - Scalability
  - Clean separation of responsibilities

---

## Why This Pattern Is Powerful

Parallel chains are very useful when:

- Multiple outputs are required from the same input
- Tasks are independent
- You want faster and more efficient execution

---

## Key Takeaway

- **Sequential chains** → Steps depend on each other  
- **Parallel chains** → Steps run independently and simultaneously  

Parallel chains allow you to build **more advanced, real-world LLM applications** like:
- Notes + Quiz generation
- Summary + Keywords extraction
- Translation + Sentiment analysis

This makes them an essential concept in LangChain.


```python
from langchain_openai import ChatOpenAI
from langchain_anthropic import ChatAnthropic
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.schema.runnable import RunnableParallel
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

import os

load_dotenv(override=True)
openai_key = os.getenv("OPENAI_API_KEY")
model1 = ChatOpenAI(api_key=openai_key)

model2 = ChatOpenAI(api_key=openai_key,model='gpt-4.1')


prompt1 = PromptTemplate(
    template='Generate short and simple notes from the following text \n {text}',
    input_variables=['text']
)

prompt2 = PromptTemplate(
    template='Generate 5 short question answers from the following text \n {text}',
    input_variables=['text']
)

prompt3 = PromptTemplate(
    template='Merge the provided notes and quiz into a single document \n notes -> {notes} and quiz -> {quiz}',
    input_variables=['notes', 'quiz']
)

parser = StrOutputParser()

parallel_chain = RunnableParallel({
    'notes': prompt1 | model1 | parser,
    'quiz': prompt2 | model2 | parser
})

merge_chain = prompt3 | model1 | parser

chain = parallel_chain | merge_chain

text = """
                Support vector machines (SVMs) are a set of supervised learning methods used for classification, regression and outliers detection.
                
                The advantages of support vector machines are:
                
                Effective in high dimensional spaces.
                
                Still effective in cases where number of dimensions is greater than the number of samples.
                
                Uses a subset of training points in the decision function (called support vectors), so it is also memory efficient.
                
                Versatile: different Kernel functions can be specified for the decision function. Common kernels are provided, but it is also possible to specify custom kernels.
                
                The disadvantages of support vector machines include:
                
                If the number of features is much greater than the number of samples, avoid over-fitting in choosing Kernel functions and regularization term is crucial.
                
                SVMs do not directly provide probability estimates, these are calculated using an expensive five-fold cross-validation (see Scores and probabilities, below).
                
                The support vector machines in scikit-learn support both dense (numpy.ndarray and convertible to that by numpy.asarray) and sparse (any scipy.sparse) sample vectors as input. However, to use an SVM to make predictions for sparse data, it must have been fit on such data. For optimal performance, use C-ordered numpy.ndarray (dense) or scipy.sparse.csr_matrix (sparse) with dtype=float64.
                """

result = chain.invoke({'text':text})

print(result)

```

```python
chain.get_graph().print_ascii()
```

## Conditional Chains in LangChain (Conceptual Explanation)

Now, let’s discuss the **last important type of chain**, called **Conditional Chains**.

---

## Problem Statement

We want to build an application where:

- A **user provides feedback** about a product  
  - Example: Product feedback on an e-commerce platform like Amazon

Our goal is to:
1. **Analyze the sentiment** of the feedback  
   - Positive  
   - Negative  

2. **Generate a response based on the sentiment**
   - If the sentiment is **positive**, return a polite and appreciative reply  
     - Example: *"Thank you for your kind words!"*
   - If the sentiment is **negative**, return an appropriate apologetic or corrective reply  

---

## Why Conditional Chains Are Needed

In this application:
- The **next step depends on a condition**
- Only **one path should execute**, based on the sentiment result

This is different from:
- **Sequential chains** → every step always runs
- **Parallel chains** → multiple paths run together

Here, we want:
- **Either Path A**
- **Or Path B**
- **Never both at the same time**

This behavior is exactly what **conditional chains** are designed for.

---

## High-Level Flow of the Conditional Chain

1. **User Input**
   - User submits product feedback

2. **Sentiment Analysis Step**
   - Feedback is sent to an LLM
   - Model classifies sentiment as:
     - Positive  
     - Negative  

3. **Conditional Routing**
   - If sentiment is **positive**:
     - Execute the *positive response chain*
   - If sentiment is **negative**:
     - Execute the *negative response chain*

4. **Final Output**
   - User receives a response appropriate to their sentiment

---

## Important Characteristic of Conditional Chains

- Only **one branch executes at a time**
- Decision is made based on:
  - Model output
  - A condition (if / else logic)

This is the key difference from **parallel chains**, where multiple branches run together.

---

## Real-World Extension (Conceptual)

Although we are keeping this example simple, the same idea can be extended:

- If sentiment is **positive**:
  - Ask the user to rate the product (e.g., 5-star review)
- If sentiment is **negative**:
  - Automatically notify customer support
  - Trigger follow-up communication

Such advanced behavior is usually implemented using **agents and tools**, which will be covered later.

---

## Key Takeaway

- **Conditional chains** allow decision-based execution
- Flow depends on model output
- Only one execution path is active at a time

They are extremely useful for building:
- Feedback systems
- Moderation workflows
- Decision-driven LLM applications

This concept completes the core chain patterns in LangChain.


```python
from langchain_openai import ChatOpenAI
from langchain_anthropic import ChatAnthropic
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.schema.runnable import RunnableParallel, RunnableBranch, RunnableLambda
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
from typing import Literal
import os


load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

model = ChatOpenAI(api_key= api_key)

parser = StrOutputParser()

class Feedback(BaseModel):

    sentiment: Literal['positive', 'negative'] = Field(description='Give the sentiment of the feedback')

parser2 = PydanticOutputParser(pydantic_object=Feedback)

prompt1 = PromptTemplate(
    template='Classify the sentiment of the following feedback text into postive or negative \n {feedback} \n {format_instruction}',
    input_variables=['feedback'],
    partial_variables={'format_instruction':parser2.get_format_instructions()}
)

classifier_chain = prompt1 | model | parser2

prompt2 = PromptTemplate(
    template='Write an appropriate response to this positive feedback \n {feedback}',
    input_variables=['feedback']
)

prompt3 = PromptTemplate(
    template='Write an appropriate response to this negative feedback \n {feedback}',
    input_variables=['feedback']
)

branch_chain = RunnableBranch(
    (lambda x:x.sentiment == 'positive', prompt2 | model | parser),
    (lambda x:x.sentiment == 'negative', prompt3 | model | parser),
    RunnableLambda(lambda x: "could not find sentiment")
)

chain = classifier_chain | branch_chain

print(chain.invoke({'feedback': "I didn't expect it to turn out like this."}))
```
