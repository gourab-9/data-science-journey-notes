## Output Parsers in LangChain

### What are Output Parsers?

Before using output parsers, it is important to understand **what they actually are**.

**Definition:**  
Output parsers in LangChain help convert **raw LLM textual responses** into **structured formats** such as:
- JSON
- CSV
- Pydantic models
- Other structured representations  

They ensure:
- Consistency in output  
- Basic validation  
- Ease of usage inside applications  

---

### Core Idea (Simple Explanation)

The basic idea is very straightforward:

- Output parsers are **pre-written classes in LangChain**
- They allow you to work with **any type of LLM**
- They help you **derive structured output** even when the model itself does not support structured output natively

This means:
- You can use output parsers with models that **already support structured output**
- You can also use them with models that **do NOT support structured output**

So, output parsers act like a **bridge** between:
- Unstructured LLM text output  
- Structured, application-ready data  

---

### Important Clarification (No Confusion)

You should **not get confused** here:

- Output parsers can be used with:
  - Models that *can* generate structured output (e.g., GPT models)
  - Models that *cannot* generate structured output (e.g., some open-source models)

In both cases, output parsers work **seamlessly**.

That is why they are extremely powerful and flexible.

---

### Models Covered Using Output Parsers

Using output parsers, you can:
- Work with **OpenAI / GPT models**
- Work with **open-source models**
- Get consistent structured output across all of them

This makes your application **model-agnostic**.

---

### Types of Output Parsers in LangChain

LangChain provides **many types of output parsers**, each designed for a specific data format.

However, in most real-world use cases, **four output parsers are used most frequently**.

These are the ones we focus on.

---

### Most Commonly Used Output Parsers

#### 1. String Output Parser
- Used when you want:
  - Clean
  - Plain
  - Final string output
- Often used to strip unnecessary formatting

---

#### 2. JSON Output Parser
- Used when:
  - You want output in **JSON format**
- Very common for APIs and structured responses

---

#### 3. Structured Output Parser
- Used when:
  - You want a predefined structured format
- Helps enforce structure without heavy validation

---

#### 4. Pydantic Output Parser
- Used when:
  - You want **strict validation**
  - You want guaranteed data types
- Most powerful parser for production-grade systems

---

### Other Available Output Parsers (For Awareness)

LangChain also provides:
- CSV Output Parser
- List Output Parser
- Other specialized parsers for different data formats

Each parser exists to handle a **specific type of structured output**.

---

### Why We Focus on These Four Parsers

In most practical applications:
- You will almost always use **one of these four**
- They cover:
  - Simple text
  - JSON
  - Structured schemas
  - Fully validated models

That is why they are considered the **core output parsers** in LangChain.

---

### Key Takeaway

- Output parsers make **structured output possible with any LLM**
- They remove dependency on model-specific capabilities
- They simplify application development
- They are essential when working with open-source or non-structured models

In upcoming sections, each of these output parsers will be explored **one by one** with clear use cases.


## String Output Parser

### Introduction

The **String Output Parser** is the **simplest output parser** available in LangChain.

Its job is very simple:

- It takes the LLM’s response  
- Extracts only the **actual textual content**  
- Returns it as a **plain Python string**

That’s it. No schema, no validation, no structure.

---

### Why Do We Need a String Output Parser?

When you interact with a **chat-based LLM**, the response you receive is **not just plain text**.

It usually contains:
- Token usage information  
- Completion tokens  
- Audio tokens (in some cases)  
- Other metadata  

The actual answer is buried inside a field like:

- `result.content`

That is why, till now, we always had to do:
- `result.content` to access the real response

---

### What String Output Parser Solves

When you use the **String Output Parser**:

- You do **not** need to manually access `result.content`
- You directly get the **clean textual output**
- The output is already a **string**, ready to be reused

This becomes extremely useful in **multi-step LLM workflows**.

---

### Important Question (Very Common Doubt)

> “If I can already print `result.content`, why do I need a String Output Parser?”

This is a **very valid question**.

The real value of the String Output Parser is **not printing** —  
The real value is **chaining LLM calls cleanly**.

---

### Key Use Case: Chaining Multiple LLM Calls

Consider this scenario:

1. You send a **topic** to the LLM  
2. The LLM generates a **detailed report** on that topic  
3. You send this **entire report back** to the same LLM  
4. Now you ask the LLM to **summarize the report in 5 lines**

---

### Why String Output Parser Is Perfect Here

In this workflow:

- The output of the **first LLM call**
- Becomes the **input of the second LLM call**

Using the String Output Parser ensures that:
- You get **only clean text**
- No metadata leaks into the next prompt
- The output is immediately reusable as input

This makes the pipeline:
- Cleaner  
- More readable  
- Less error-prone  

---

### When to Use String Output Parser

Use the **String Output Parser** when:

- You only need **plain text**
- You are chaining **multiple LLM calls**
- You want to avoid manually accessing `.content`
- You want cleaner and more maintainable code

---

### When NOT to Use It

Do **not** use it when:
- You need structured data (JSON, dict, schema)
- You need validation
- You need field-level access

In those cases, use:
- JSON Output Parser
- Structured Output Parser
- Pydantic Output Parser

---

### Key Takeaway

- String Output Parser = **clean text only**
- Ideal for **LLM-to-LLM workflows**
- Removes unnecessary metadata
- Makes chaining and reuse effortless

It may look simple, but it is **extremely powerful** in real-world pipelines.


```python
# without parser
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate

load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="google/gemma-2-2b-it",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)

# 1st prompt -> detailed report
template1 = PromptTemplate(
    template='Write a detailed report on {topic}',
    input_variables=['topic']
)

# 2nd prompt -> summary
template2 = PromptTemplate(
    template='Write a 5 line summary on the following text. /n {text}',
    input_variables=['text']
)

prompt1 = template1.invoke({'topic':'black hole'})

result = model.invoke(prompt1)

prompt2 = template2.invoke({'text':result.content})

result1 = model.invoke(prompt2)

print(result1.content)
```

```python
# with StrOutputParser

from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser



load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="google/gemma-2-2b-it",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)



# 1st prompt -> detailed report
template1 = PromptTemplate(
    template='Write a detailed report on {topic}',
    input_variables=['topic']
)

# 2nd prompt -> summary
template2 = PromptTemplate(
    template='Write a 5 line summary on the following text. /n {text}',
    input_variables=['text']
)


parser = StrOutputParser()

chain = template1 | model | parser | template2 | model | parser

result = chain.invoke({'topic':'black hole'})

print(result)
```

## JSON Output Parser

### Introduction

As the name suggests, the **JSON Output Parser** is used when you want your LLM to return its response strictly in **JSON format**.

This parser **forces the LLM** to structure its output as JSON, instead of returning free-form text.

---

### What Does JSON Output Parser Do?

The JSON Output Parser:

- Tells the LLM that the **expected output format is JSON**
- Parses the LLM response into a **valid JSON object**
- Ensures the output is **machine-readable and structured**

In simple words:
> If you are expecting JSON as output, this parser helps you get it cleanly and reliably.

---

### Why Is JSON Output Parser Important?

Most real-world applications do **not** work well with raw text.

They need:
- Key–value pairs  
- Structured fields  
- Predictable output format  

JSON is the **most common format** used to:
- Store data
- Send data via APIs
- Integrate with databases and frontend systems

That’s where JSON Output Parser becomes extremely useful.

---

### Biggest Advantage

The **biggest advantage** of JSON Output Parser is:

> It is the **quickest and simplest way** to get JSON output from an LLM.

You don’t need:
- Complex schemas
- Validation logic
- Heavy configurations

Just tell the LLM:
- “Respond in JSON”
- And parse it using the JSON Output Parser

---

### Typical Use Cases

JSON Output Parser is commonly used when:

- You are building **APIs**
- You want to send LLM output to a **frontend**
- You want to store LLM output in a **database**
- You want structured data but **don’t need strict validation**

Examples include:
- Extracting summary and sentiment from reviews
- Generating structured responses for dashboards
- Passing LLM output to another system or service

---

### When to Use JSON Output Parser

Use JSON Output Parser when:

- You want **structured output**
- JSON format is sufficient
- You don’t need strict type validation
- You want a fast and simple solution

---

### When NOT to Use It

Avoid JSON Output Parser when:

- You need **strong validation**
- You want type safety
- You want default values or constraints

In those cases, prefer:
- **Pydantic Output Parser**
- **Structured Output Parser**

---

### Key Takeaway

- JSON Output Parser = **force JSON output**
- Fastest way to get structured data from an LLM
- Ideal for APIs, databases, and integrations
- Lightweight and easy to use

If your goal is:
> “I just want clean JSON from my LLM”  
then **JSON Output Parser** is the perfect choice.


```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser

load_dotenv()

# Define the model
llm = HuggingFaceEndpoint(
    repo_id="google/gemma-2-2b-it",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)

parser = JsonOutputParser()

template = PromptTemplate(
    template='Give me 5 facts about {topic} \n {format_instruction}',
    input_variables=['topic'],
    partial_variables={'format_instruction': parser.get_format_instructions()}
)

chain = template | model | parser

result = chain.invoke({'topic':'black hole'})

print(result)
```

## Limitation of JSON Output Parser

Now let’s understand an important limitation.

Suppose the LLM returns JSON like this:

- A single key: **"facts_about_black_holes"**
- All facts are stored inside a **list**

This structure is valid, but it is **not flexible or strictly controlled**.

---

### Problem Scenario

Now assume you want the output in this exact structure:

- **fact_1** → first fact  
- **fact_2** → second fact  
- **fact_3** → third fact  

That means:
- Keys should be `fact_1`, `fact_2`, `fact_3`
- Values should be the actual facts

If you want JSON **strictly in this format**, then:

> ❌ JSON Output Parser cannot guarantee this structure

You can try to specify it in the prompt,  
but there is **no guarantee** the LLM will always follow it exactly.

---

### Biggest Flaw of JSON Output Parser

**The biggest drawback of JSON Output Parser is:**

> ❌ It does NOT enforce a schema

This means:
- Keys may change  
- Structure may vary  
- Output may be inconsistent  

---

## Structured Output Parser

To solve this problem, we use **Structured Output Parser**.

---

### Definition

A **Structured Output Parser** is an output parser in LangChain that:

- Extracts structured JSON data from LLM responses
- Based on a **pre-defined field schema**

---

### JSON Output Parser vs Structured Output Parser

| Feature | JSON Output Parser | Structured Output Parser |
|------|-------------------|-------------------------|
| Returns JSON | ✅ Yes | ✅ Yes |
| Schema enforcement | ❌ No | ✅ Yes |
| Fixed keys | ❌ Not guaranteed | ✅ Guaranteed |
| Consistent structure | ❌ No | ✅ Yes |

---

### Key Difference (Most Important)

With **Structured Output Parser**:

- You define a schema in advance
- You clearly tell the LLM:
  > “Return the output strictly according to this schema”

The LLM:
- Understands the schema  
- Generates output following the defined structure  

---

### Biggest Benefit

**The main benefit is:**

> ✅ Schema enforcement

This ensures:
- Keys are always the same  
- Output format is predictable  
- Data can be safely passed to databases, APIs, or tools  

---

### When to Use Structured Output Parser

Use it when:
- You need an **exact JSON structure**
- Output will be consumed by another system
- You are building agents or tool-based workflows
- Consistency and reliability are important  

---

### Final Takeaway

- **JSON Output Parser** → Simple and fast, but no structure guarantee  
- **Structured Output Parser** → Slightly advanced, but highly reliable  

If structure matters,  
**Structured Output Parser is the right choice.**


```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain.output_parsers import StructuredOutputParser, ResponseSchema

load_dotenv()

# Define the model
llm = HuggingFaceEndpoint(
    repo_id="google/gemma-2-2b-it",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)

schema = [
    ResponseSchema(name='fact_1', description='Fact 1 about the topic'),
    ResponseSchema(name='fact_2', description='Fact 2 about the topic'),
    ResponseSchema(name='fact_3', description='Fact 3 about the topic'),
]

parser = StructuredOutputParser.from_response_schemas(schema)

template = PromptTemplate(
    template='Give 3 fact about {topic} \n {format_instruction}',
    input_variables=['topic'],
    partial_variables={'format_instruction':parser.get_format_instructions()}
)

chain = template | model | parser

result = chain.invoke({'topic':'black hole'})

print(result)
```

## Limitation of Structured Output Parser

Now let’s talk about the **downside (disadvantage)** of the Structured Output Parser.

The **biggest disadvantage** of the Structured Output Parser is:

> ❌ **It cannot perform data validation**

---

### What Structured Output Parser Can Do
- It can **enforce structure**
- It can force the LLM to return output in a **specific JSON format**

---

### What It Cannot Do (Important)

It **cannot validate data types or constraints**.

---

### Example Problem

Assume your schema expects:
- `name` → string  
- `age` → integer  
- `city` → string  

Now suppose the LLM returns:
- `name`: `"Rahul"`
- `age`: `"35 years"` ❌ (string instead of integer)
- `city`: `"Delhi"`

Even though:
- `age` should be an integer  

The Structured Output Parser:
- ❌ **will NOT stop this**
- ❌ **will NOT throw an error**
- ❌ **will NOT auto-correct it**

You can mention rules in the prompt,  
but **if the LLM still sends incorrect data**, you must:
- Accept it  
- Or manually process and fix it  

---

### Core Limitation Summary

| Capability | Structured Output Parser |
|--------|--------------------------|
| Enforces JSON structure | ✅ Yes |
| Enforces data types | ❌ No |
| Validates constraints | ❌ No |
| Prevents wrong formats | ❌ No |

So:
- It **forces structure**
- But **does NOT validate data**

---

## Why We Need Pydantic Output Parser

Because of this limitation, we use the **fourth output parser**:

> ✅ **Pydantic Output Parser**

---

## Pydantic Output Parser

### Definition

**Pydantic Output Parser** is a structured output parser in LangChain that:

- Uses **Pydantic models**
- Enforces **schema + data validation**
- Validates LLM responses during parsing

---

### How It Works (Conceptually)

- Instead of passing a simple schema
- You pass a **Pydantic model**
- Since it is a Pydantic object, it supports:
  - Type validation
  - Constraints
  - Default values
  - Automatic type conversion

---

### Key Features of Pydantic Output Parser

#### 1. Strict Schema Enforcement
- You can strictly define:
  - Required fields
  - Data types
  - Constraints

#### 2. Data Validation
- If the data does not match the schema:
  - Validation errors can be raised
  - Invalid outputs are caught early

#### 3. Type Safety
- If the LLM returns slightly incorrect data:
  - Pydantic can apply **type coercion**
  - Example: `"35"` → `35`

#### 4. Easy Validation Logic
- You can add:
  - Custom validators
  - Field constraints
  - Default values

#### 5. Seamless Integration
- Works smoothly with:
  - Databases
  - APIs
  - Agents
  - Other LangChain components

---

### Final Comparison

| Feature | Structured Output Parser | Pydantic Output Parser |
|------|-------------------------|------------------------|
| JSON structure enforcement | ✅ Yes | ✅ Yes |
| Data validation | ❌ No | ✅ Yes |
| Type safety | ❌ No | ✅ Yes |
| Error handling | ❌ Limited | ✅ Strong |
| Best for production systems | ❌ | ✅ |

---

### Final Takeaway

- **Structured Output Parser**  
  → Use when structure matters, but validation is not critical  

- **Pydantic Output Parser**  
  → Use when you need **reliable, validated, production-ready data**

This is why **Pydantic Output Parser is the most powerful and preferred option** in real-world applications.


```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field

load_dotenv()

# Define the model
llm = HuggingFaceEndpoint(
    repo_id="google/gemma-2-2b-it",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)

class Person(BaseModel):

    name: str = Field(description='Name of the person')
    age: int = Field(gt=18, description='Age of the person')
    city: str = Field(description='Name of the city the person belongs to')

parser = PydanticOutputParser(pydantic_object=Person)

template = PromptTemplate(
    template='Generate the name, age and city of a fictional {place} person \n {format_instruction}',
    input_variables=['place'],
    partial_variables={'format_instruction':parser.get_format_instructions()}
)

chain = template | model | parser

final_result = chain.invoke({'place':'sri lankan'})

print(final_result)
```
