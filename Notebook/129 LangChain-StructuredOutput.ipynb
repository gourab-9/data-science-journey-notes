# Structured Output in LLMs — Conceptual Notes (Single Cell)

In earlier lessons, we learned **how to send inputs (prompts) to an LLM**.  
This section focuses on **how to work with the output generated by an LLM**, especially when we want to use that output inside real applications.

The core topic here is **Structured Output**.

---

## 1. What is Unstructured Output?

### Definition
Most LLMs, by default, return **plain text responses**.  
Plain text does not follow any strict format or schema.

### Example
**Prompt:**  
> What is the capital of India?

**Output:**  
> New Delhi is the capital of India.

This output is:
- Human-readable  
- Easy to understand  
- ❌ Difficult for programs to reliably parse  

This type of output is called **Unstructured Output**.

---

## 2. Why Unstructured Output is a Problem

Unstructured text:
- Cannot be directly inserted into databases
- Is hard to integrate with APIs
- Is unreliable for automation
- Requires extra parsing logic

LLMs work well with humans,  
but **machines prefer structured data**.

---

## 3. What is Structured Output?

### Definition
**Structured output** means forcing the LLM to return responses in a **well-defined data format** such as:
- JSON
- Key–value pairs
- Lists of dictionaries

This makes the output:
- Predictable
- Machine-readable
- Easy to integrate with other systems

---

## 4. Unstructured vs Structured (Conceptual Example)

### Unstructured Output
**Prompt:**  
> Create a one-day travel itinerary for Paris.

**Output (text):**
- Morning: Visit Eiffel Tower  
- Afternoon: Visit a museum  
- Evening: Dinner at a café  

Readable, but not program-friendly.

---

### Structured Output
**Same idea, structured response:**

- Day Plan:
  - Time: Morning → Activity: Eiffel Tower
  - Time: Afternoon → Activity: Museum
  - Time: Evening → Activity: Dinner

Here:
- Each item has fixed fields
- Data follows a predictable structure

This is **Structured Output**.

---

## 5. Formal Definition (LangChain Context)

In LangChain, **Structured Output** refers to the practice of making LLMs return responses in a **well-defined data format**, instead of free-form text.

This allows the output to be:
- Parsed
- Validated
- Stored
- Sent to other systems

---

## 6. Why Structured Output is Powerful

Structured output makes LLMs usable beyond chat:

- Databases
- APIs
- Agents
- Automation pipelines
- Tool-based systems

It transforms LLMs from **chat systems** into **backend components**.

---

## 7. Use Case 1: Data Extraction

### Scenario
A job portal where users upload resumes.

### Problem
Resumes are unstructured text.

### Solution using Structured Output
The LLM extracts structured fields such as:
- Candidate name

### Benefit

- No manual parsing
- Scalable for thousands of resumes

## 8. Use Case 2: Building APIs from Text

### Scenario
E-commerce product reviews (long, unstructured text)

### Structured information extracted:
- Review topic (battery, display, performance)
- Pros
- Cons
- Overall sentiment

### Result
You can build an API that:
- Takes raw reviews
- Returns clean structured data
- Can be consumed by any frontend or service

---

## 9. Use Case 3: Building AI Agents

### Key Insight
**Tools do not understand text — they understand data.**

### Example
User prompt:
> Find the square root of 2

A calculator tool needs:
- Operation: square_root
- Number: 2

### Role of Structured Output
The LLM converts text into structured fields:
- What operation to perform
- On which number

Then:
- The tool executes
- The result is returned

This is **critical for agent systems**.

---

## 10. Structured Output Enables Machine-to-Machine Communication

Earlier:
- Human ↔ LLM (text)

Now:
- LLM ↔ Databases
- LLM ↔ APIs
- LLM ↔ Tools
- LLM ↔ Other machines

All of this becomes possible **only because of structured output**.

## 11. Key Takeaway

  - Agents

# Structured Output in LLMs — Models, LangChain Support & TypedDict (Conceptual Notes)

---

## 1. Two Types of LLMs (Important Clarification)

In practice, you will encounter **two categories of LLMs** when working with structured output:

### Type 1: LLMs that Support Structured Output Natively
These models are **trained** to follow structured instructions.

**Examples:**
- OpenAI GPT models
- Some enterprise-grade chat models

**Capabilities:**
- Can directly return JSON or structured formats
- Follow schemas accurately
- Require minimal prompt engineering

---

### Type 2: LLMs that Do NOT Support Structured Output Natively

These models:
- Always return free-form text
- Cannot reliably follow strict formats
- Often hallucinate structure

**Examples:**
- Many open-source or smaller LLMs

---

## 2. How LangChain Handles Both Types

LangChain provides **two different mechanisms** so you can generate structured output **from any LLM**.

### Case 1: Model Supports Structured Output
→ Use **`with_structured_output`**

### Case 2: Model Does NOT Support Structured Output
→ Use **Output Parsers**

This separation ensures:
- Clean architecture
- Model-agnostic design
- Same application logic for different LLMs

---

## 3. `with_structured_output` — Conceptual Overview

When an LLM **already understands structure**, LangChain simplifies everything.

### Core Idea
Before invoking the model:
- You attach a **data format**
- The model is instructed to respond **only in that format**

### Key Benefit
- No manual parsing
- No regex
- No brittle prompt tricks

You simply say:
> “Here is the structure I want. Follow it.”

---

## 4. Ways to Define the Output Structure in LangChain

LangChain allows **three ways** to specify the desired data format:

1. **TypedDict**
2. **Pydantic Models**
3. **JSON Schema**

Each has a specific use case.

---

## 5. TypedDict — What It Is

### Definition
A **TypedDict** is a way to define the **expected structure of a dictionary** in Python.

It specifies:
- What keys must exist
- What type of value each key should hold

---

## 6. Why TypedDict Exists

Normally in Python:
- Dictionaries are flexible
- No enforcement of structure
- Easy to misuse in large projects

### Problem Scenario
Two developers working on the same dictionary:
- One treats `age` as an integer
- Another treats `age` as a string

This leads to:
- Runtime bugs
- Inconsistent behavior
- Hard-to-debug errors

---

## 7. What TypedDict Solves

TypedDict acts as a **contract**.

It communicates:
- Which keys are expected
- What data type each key should have

This helps:
- Code editors (autocomplete & warnings)
- Developers collaborating on the same code
- LLMs understanding the expected structure

---

## 8. What TypedDict Does *NOT* Do (Very Important)

TypedDict:
- ❌ Does NOT enforce validation at runtime
- ❌ Does NOT stop wrong data types
- ❌ Does NOT raise errors automatically

If you:
- Define `age` as integer
- But assign a string

➡️ Python will still run the code.

TypedDict is about **clarity**, not enforcement.

---

## 9. TypedDict in Structured Output Context

When used with `with_structured_output`:

- The TypedDict becomes a **schema**
- The LLM is guided to match that structure
- The output becomes predictable and machine-friendly

This is especially useful when:
- You trust the model’s structured capabilities
- You want lightweight schemas
- You don’t need strict runtime validation

---

## 10. When Should You Use TypedDict?

Use TypedDict when:
- You want simple structured output
- You want editor support
- You are prototyping or building internal tools
- You trust the LLM to behave reasonably

Avoid TypedDict when:
- You need strict validation
- Data correctness is critical
- You are exposing APIs publicly  
(→ Pydantic is better there)

---

## 11. Big Picture Mental Model

- **LLM supports structure** → `with_structured_output`
- **LLM doesn’t support structure** → Output Parsers
- **TypedDict** → Light schema, developer guidance
- **Pydantic / JSON Schema** → Stronger contracts

---

## 12. Key Takeaway

- Not all LLMs are equal in structured output capability
- LangChain abstracts this difference cleanly
- TypedDict is a **schema description tool**, not a validator
- It improves clarity, collaboration, and reliability

> Structured output is not about formatting text.  
> It is about making LLMs usable inside real systems.

---

### Problem Statement

We are considering a simple problem where we have **phone reviews** available as text data.  
These reviews are sent to an **LLM**, and in return, we expect a **structured output**.

The required structured output should be a **dictionary** containing:
- A **`summary`** key, which holds a generated summary of the review.
- A **`sentiment`** key, which represents the overall sentiment of the review and can be **Positive**, **Negative**, or **Neutral**.

Once this dictionary is generated, we should be able to easily extract:
- the **summary** from the `summary` field, and  
- the **sentiment** from the `sentiment` field,  

and use them independently wherever required.


```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict, Annotated, Optional, Literal
import os


load_dotenv(override=True)


# Read the API key
api_key = os.getenv("OPENAI_API_KEY")

model = ChatOpenAI(openai_api_key=api_key)


# schema
class Review(TypedDict):

    key_themes: Annotated[list[str], "Write down all the key themes discussed in the review in a list"]
    summary: Annotated[str, "A brief summary of the review"]
    sentiment: Annotated[Literal["pos", "neg"], "Return sentiment of the review either negative, positive or neutral"]
    pros: Annotated[Optional[list[str]], "Write down all the pros inside a list"]
    cons: Annotated[Optional[list[str]], "Write down all the cons inside a list"]
    name: Annotated[Optional[str], "Write the name of the reviewer"]
    

structured_model = model.with_structured_output(Review)

result = structured_model.invoke("""I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! 
                                    The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, 
                                    or editing photos. The 5000mAh battery easily lasts a full day even with heavy use, 
                                    and the 45W fast charging is a lifesaver.

                            The S-Pen integration is a great touch for note-taking and quick sketches, though I don't use it often. 
                            What really blew me away is the 200MP camera—the night mode is stunning, capturing crisp, 
                            vibrant images even in low light. Zooming up to 100x actually works well for distant objects, 
                            but anything beyond 30x loses quality.
                            
                            However, the weight and size make it a bit uncomfortable for one-handed use. Also, 
                            Samsung’s One UI still comes with bloatware—why do I need five different Samsung apps for things Google already provides?
                            The $1,300 price tag is also a hard pill to swallow.
                            
                            Pros:
                            Insanely powerful processor (great for gaming and productivity)
                            Stunning 200MP camera with incredible zoom capabilities
                            Long battery life with fast charging
                            S-Pen support is unique and useful
                                                             
                            Review by Gourab Singh
                            """)

print(result['name'])
```

### Limitations of TypedDict and Need for Pydantic

This is how **TypedDict** works and how we can extract **structured output** from an LLM response using it.  
This approach works correctly, but it comes with an important limitation.

The main problem is that **there is no guarantee of data correctness**.

For example:
- Even if we specify that **`summary` should be a string**, the LLM might still return it in a different format.
- If we instruct the model to include a **rating only when it is greater than 3**, the LLM might still include the rating even when the condition is not met.

This happens because **TypedDict does not enforce validation**.  
It only defines how the data *should look*, but it **does not check or restrict** the actual values returned by the model.

So, TypedDict is useful mainly for **representation and structure clarity**, not for enforcing rules or constraints.

---

### Introducing Pydantic

If we need **data validation along with structure**, then TypedDict is not sufficient.  
For such cases, we use **Pydantic**.

**Pydantic** is a **data validation and data parsing library for Python**.

In simple terms:
- If you want to apply **strict checks** on incoming data,
- If you want to ensure the data follows specific **rules, types, and constraints**,
- And if you want the system to **raise errors when the data is invalid**,

then **Pydantic** is the right choice.

Using Pydantic, we can create **schemas** that not only define the structure of the output but also **validate** whether the data returned by the LLM satisfies all required conditions.

This makes Pydantic a much better option when **reliability and correctness of structured output** is critical.


```python
# 

from pydantic import BaseModel

class Student(BaseModel):
    
    name:str

new_student = {"name":"Gourab"}
stu = Student(**new_student)
print(stu)
print(type(stu))
```

```python
from pydantic import BaseModel, EmailStr, Field
from typing import Optional

class Student(BaseModel):

    name: str = 'Gourab Kumar'
    age: Optional[int] = None
    email: EmailStr
    cgpa: float = Field(gt=0, lt=10, default=5, description='A decimal value representing the cgpa of the student')


new_student = {'age':'28', 'email':'gourabkumarsingh09@gmail.com'}

student = Student(**new_student)

student_dict = dict(student)

print(student_dict['age'])

student_json = student.model_dump_json()
```

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict, Annotated, Optional, Literal
from pydantic import BaseModel, Field
import os


load_dotenv(override=True)


# Read the API key
api_key = os.getenv("OPENAI_API_KEY")

model = ChatOpenAI(openai_api_key=api_key)

# schema
class Review(BaseModel):

    key_themes: list[str] = Field(description="Write down all the key themes discussed in the review in a list")
    summary: str = Field(description="A brief summary of the review")
    sentiment: Literal["pos", "neg"] = Field(description="Return sentiment of the review either negative, positive or neutral")
    pros: Optional[list[str]] = Field(default=None, description="Write down all the pros inside a list")
    cons: Optional[list[str]] = Field(default=None, description="Write down all the cons inside a list")
    name: Optional[str] = Field(default=None, description="Write the name of the reviewer")
    

structured_model = model.with_structured_output(Review)

result = structured_model.invoke("""I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! 
                                    The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, or editing photos. 
                                    The 5000mAh battery easily lasts a full day even with heavy use, and the 45W fast charging is a lifesaver.

                                The S-Pen integration is a great touch for note-taking and quick sketches, though I don't use it often. 
                                What really blew me away is the 200MP camera—the night mode is stunning, capturing crisp, 
                                vibrant images even in low light. Zooming up to 100x actually works well for distant objects, 
                                but anything beyond 30x loses quality.
                                
                                However, the weight and size make it a bit uncomfortable for one-handed use. Also, 
                                Samsung’s One UI still comes with bloatware—why do I need five different Samsung apps for things Google already 
                                provides? The $1,300 price tag is also a hard pill to swallow.
                                
                                Pros:
                                Insanely powerful processor (great for gaming and productivity)
                                Stunning 200MP camera with incredible zoom capabilities
                                Long battery life with fast charging
                                S-Pen support is unique and useful
                                                                 
                                Review by Gourab Singh
                                """)

print(result)
```

## JSON Schema for Structured Output in LLMs

### When to Use JSON Schema

JSON Schema is used when your project is **not limited to a single programming language**.  
If your system is built using **multiple languages** (for example, Python in the backend and JavaScript in the frontend), then **TypedDict** or **Pydantic** are not ideal choices.

This is because:
- TypedDict and Pydantic are **Python-specific**
- JSON is a **universal data format** understood by almost every programming language

So, whenever the same schema needs to be **shared across different technologies**, JSON Schema becomes the best option.

---

### What is a JSON Schema?

A JSON Schema defines:
- The **structure** of the data
- The **data types**
- Which fields are **required** and which are optional

A typical JSON Schema contains the following key components:

1. **Title**  
   Describes the name of the schema.

2. **Description (Optional but Recommended)**  
   Explains what the schema represents.  
   This is especially useful when interacting with LLMs.

3. **Type**  
   Defines the overall data type.  
   - For dictionary-like data, this is usually `object`.

4. **Properties**  
   Lists all attributes (fields) in the schema along with:
   - Their data types
   - Optional descriptions
   - Nested structures if needed

5. **Required**  
   Specifies which fields must always be present in the output.

---

### Example: Student Schema (Conceptual)

- Schema represents a **Student**
- `name` is mandatory
- `age` is optional
- The overall structure is an **object**
- This is a valid JSON Schema and can be understood by any language

---

### Smartphone Review JSON Schema (Conceptual Overview)

For a structured smartphone review, the schema may contain:

- **themes**  
  - Type: Array of strings  
  - Represents key topics discussed in the review

- **summary**  
  - Type: String  
  - A concise summary of the review

- **sentiment**  
  - Type: String  
  - Allowed values defined using **enum**  
  - Example values: Positive, Negative, Neutral

- **pros**  
  - Type: Array of strings or null  
  - Optional field

- **cons**  
  - Type: Array of strings or null  
  - Optional field

- **name**  
  - Type: String or null  
  - Optional field

- **required fields**  
  - themes  
  - summary  
  - sentiment  

This schema ensures clarity while still being flexible and cross-platform.

---

### Output Behavior with JSON Schema

- The output returned by the LLM is a **Python dictionary**
- Similar to TypedDict in usage
- Unlike Pydantic, the output is not a class instance
- Easy to extract fields like `summary` or `sentiment`

---

### Comparison: TypedDict vs Pydantic vs JSON Schema

#### TypedDict
- Use when:
  - Entire project is in Python
  - You only need **type hints**
- Limitations:
  - No validation
  - No default values
  - Not cross-language

#### Pydantic
- Use when:
  - You need **data validation**
  - Default values are required
  - Automatic type conversion is useful
- Limitations:
  - Python-only
  - Requires extra dependency

#### JSON Schema
- Use when:
  - Project involves **multiple languages**
  - Schema needs to be shared across systems
- Strength:
  - Universal compatibility
- Limitation:
  - No automatic type conversion like Pydantic

> In most Python-based LLM projects, **Pydantic becomes the default choice**.

---

### Structured Output Methods in `with_structured_output`

When using `with_structured_output`, you can specify **how** the structured output should be generated.

There are two main methods:

#### 1. JSON Mode
- Output is returned in JSON format
- Commonly used
- Works well with most models

#### 2. Function Calling
- Output is designed to trigger a function call
- Used heavily in **agents and tool-based workflows**
- Example: Calling a calculator or database function

**Rule of Thumb**
- OpenAI models → Prefer **Function Calling** (default)
- Other models (Claude, Gemini, etc.) → Prefer **JSON Mode**

---

### Models That Do NOT Support Structured Output

Some models **do not support structured output at all**, neither:
- JSON Mode
- Nor Function Calling

Example:
- Certain open-source models like **TinyLlama**

In such cases:
- `with_structured_output` will fail
- You must manually structure the output using **Output Parsers**

This is covered separately in the next topic.
