## Introduction to Natural Language Processing (NLP)

### Starting Point
The field of modern AI is filled with **abbreviations and concepts** that can feel overwhelming at first. The key goal when learning NLP is to gradually build a **mental map**—understanding what each term refers to and how different ideas connect.  
To do this effectively, it helps to start from a **high-level perspective** and then go deeper step by step.

---

## What is NLP?
**Natural Language Processing (NLP)** is the area of AI and machine learning that focuses on:
- Working with **text as data**
- Performing computation and learning over human language

At a high level, most NLP problems can be grouped into **three major task categories**.

---

## 1. Classification Tasks

**Input:** Text  
**Output:** A single label  

The goal is to assign one category to an entire piece of text.

### Common Examples
- **Sentiment analysis**: positive, negative, neutral  
- **Intent detection**: understanding what the user wants to do  
- **Language detection**: identifying the language of a text  
- **Topic classification**

### Typical Data Sources
- Movie reviews  
- Product reviews  
- Social media posts  

### Evaluation Metrics
- **Accuracy**: Percentage of correct predictions  
- **Precision**: Of all predicted positives, how many are correct  
- **Recall**: Of all actual positives, how many were identified  
- **F1 Score**: Harmonic mean of precision and recall  

### Why Multiple Metrics Matter
In many real datasets, classes are **imbalanced** (e.g., 99% positive, 1% negative).  
In such cases:
- Accuracy can be misleading  
- Precision and recall provide a clearer picture of model quality  

---

## 2. Multi-Classification (Sequence Labeling)

**Input:** Text  
**Output:** Multiple labels within the same text  

Instead of labeling the entire sentence, the model labels **individual words or tokens**.

### Common Tasks
- **Named Entity Recognition (NER)**: identifying entities like locations, dates, or names  
- **Part-of-Speech (POS) tagging**: noun, verb, adjective, etc.  
- **Syntactic parsing**: dependency or constituency parsing  

### Evaluation
- Metrics are computed at the **token level** or **entity-type level**
- Performance is often analyzed separately for each entity category (e.g., locations only)

---

## 3. Text Generation Tasks

**Input:** Text  
**Output:** Text (variable length)  

Here, both input and output are text, and the output length is not fixed in advance.

### Common Examples
- **Machine translation**
- **Question answering**
- **Summarization**
- **General text generation** (stories, code, poems)

This category has become especially prominent in recent years.

---

## Evaluating Text Generation

### Reference-Based Metrics
These metrics compare model output against a reference text.

- **BLEU (Bilingual Evaluation Understudy)**  
- **ROUGE** (a family of recall-oriented metrics)

**Limitations:**
- Require labeled reference outputs
- Many correct answers may exist for the same input
- Label creation is expensive and time-consuming

### Probability-Based Metric
- **Perplexity**: Measures how surprised a model is by its own predictions  

**Interpretation:**
- BLEU / ROUGE → higher is better  
- Perplexity → lower is better  

---

## Brief Historical Context

- Early ideas for language models existed as far back as the **1980s**
- **LSTMs (1990s)** improved sequence modeling but were limited by data and compute
- Lack of large-scale data and computing power slowed progress
- **Word2Vec** later enabled meaningful word embeddings
- **Transformers (2017)** introduced a new foundation for language modeling
- Scaling data and compute in the **2020s** led to what we now call **Large Language Models (LLMs)**

---

## Key Takeaway
NLP can be understood through:
- A small number of **core task categories**
- Clear **evaluation principles**
- An appreciation of how the field evolved over time

This high-level foundation helps build the mental framework needed to understand more advanced NLP and LLM concepts later.


## Tokenization: Converting Text into Model-Readable Form

### Why Tokenization Is Needed
Machine learning models **do not understand text directly**.  
They operate on **numbers**, so raw text must first be converted into a structured, quantifiable form.

The first fundamental question in NLP is:
**How do we break text into pieces that a model can process?**

This process is called **tokenization**.

---

## What Is Tokenization?
**Tokenization** is the process of splitting text into smaller units called **tokens**.  
These tokens are the basic inputs that later get converted into numerical representations.

Example sentence:
> *“a cute teddy bear is reading”*

The way this sentence is split depends on the tokenization strategy.

---

## Tokenization Strategies

### 1. Word-Level Tokenization
Text is split into full words.

**Example:**
- `a | cute | teddy | bear | is | reading`

#### Pros
- Simple and intuitive
- Easy to interpret

#### Cons
- Does **not capture word roots**
- Similar words are treated as completely different tokens  
  - `bear` vs `bears`
  - `run` vs `runs`
- Leads to **Out-Of-Vocabulary (OOV)** issues  
  - If a word was never seen during training, it must be marked as *unknown*

---

### 2. Subword-Level Tokenization
Words are split into **meaningful sub-units**, often based on shared roots.

**Example:**
- `bear` + `s`
- `read` + `ing`

#### Pros
- Leverages **common roots and prefixes**
- Reduces the **OOV problem**
- Enables better generalization across word variations

#### Cons
- Produces **longer token sequences**
- Longer sequences increase:
  - Computational cost
  - Inference time  
- Model complexity often grows with sequence length

---

### 3. Character-Level Tokenization
Text is split into individual characters.

**Example:**
- `a | c | u | t | e | ...`

#### Pros
- Extremely robust to:
  - Misspellings
  - Casing variations
- Very low risk of OOV

#### Cons
- Sequences become **very long**
- Computation becomes much slower
- Hard to learn meaningful representations  
  - Individual characters have little semantic meaning on their own

---

## Out-Of-Vocabulary (OOV) Problem
A token is **OOV** when:
- It appears during inference
- But was **never seen during training**

**Impact:**
- The model cannot assign a meaningful representation
- Word-level tokenization suffers the most
- Subword tokenization significantly reduces this risk

---

## Summary of Trade-offs

- **Word-level**
  - ✅ Simple and interpretable  
  - ❌ Poor generalization, high OOV risk  

- **Subword-level**
  - ✅ Best balance of efficiency and meaning  
  - ❌ Longer sequences  

- **Character-level**
  - ✅ Robust to noise and spelling errors  
  - ❌ Slow and semantically weak  

---

## Key Takeaway
Tokenization is the **foundation of NLP pipelines**.  
The choice of tokenizer directly affects:
- Model efficiency
- Generalization ability
- Computational cost

Most modern NLP systems rely on **subword tokenization** as the best trade-off between linguistic structure and computational efficiency.


## Token Representation: From One-Hot Vectors to Learned Embeddings

### From Tokens to Numbers
After tokenization, we have **tokens**, but models still need **numerical representations** to work with them.  
This step is called **token (or word) representation**.

The goal is to map each token to a vector that:
- Can be processed by a model
- Captures **meaning and similarity** between tokens

---

## One-Hot Encoding (OHE)

### What Is One-Hot Encoding?
Each token is represented as a vector where:
- The vector length = vocabulary size
- Only one position is `1`, all others are `0`

**Example vocabulary:**  
`[soft, teddy_bear, book]`

| Token       | Vector      |
|------------|-------------|
| soft       | [1, 0, 0]   |
| teddy_bear | [0, 1, 0]   |
| book       | [0, 0, 1]   |

---

### Why One-Hot Encoding Fails
To compare token meanings, we often use **cosine similarity**, which measures the **angle between vectors**.

**Problem:**
- All one-hot vectors are **orthogonal**
- Cosine similarity between any two different tokens = 0
- No notion of semantic similarity

**Example:**
- `teddy_bear` and `soft` should be similar
- One-hot encoding treats them as completely unrelated

---

## What We Actually Want
- Similar words → vectors pointing in similar directions
- Unrelated words → vectors closer to orthogonal
- Opposite meanings → vectors pointing away from each other

This is **not possible** with one-hot vectors.

---

## Vocabulary Size and OOV
- Vocabulary size (`V`) = number of unique tokens seen during training
- If a token appears at inference time but not during training, it becomes **OOV (Out Of Vocabulary)**

**Handling OOV:**
- Reserve a special **[UNK] (unknown)** token
- Word-level tokenization has a **high OOV risk**
- Subword tokenization reduces this risk
- Character-level tokenization almost eliminates it

---

## Learning Token Embeddings from Data

### Core Idea
Instead of manually defining vectors, **learn embeddings from data** using a training task.

A good embedding should:
- Encode semantic relationships
- Reflect how words are used in context

---

## Word2Vec: Learning Meaningful Embeddings

Word2Vec introduced a simple but powerful idea:
> *Words that appear in similar contexts should have similar embeddings.*

It uses **proxy tasks**—tasks that are not the final goal, but help learn good representations.

---

### Proxy Tasks in Word2Vec

#### 1. Continuous Bag of Words (CBOW)
- Input: surrounding words
- Output: target word

**Goal:** Predict a word using its context

---

#### 2. Skip-Gram
- Input: target word
- Output: surrounding words

**Goal:** Predict the context using a word

Both approaches force the model to **understand language structure**.

---

## Simple Neural Network for Embedding Learning

### Architecture (Conceptual)
- **Input layer**: one-hot vector of size `V`
- **Hidden layer**: dense vector of size `d` (embedding dimension)
- **Output layer**: probability distribution over vocabulary

Key properties:
- `V` → large (tens or hundreds of thousands)
- `d` → much smaller (hundreds or thousands)

---

### Training Process
1. Take a token as input (one-hot encoded)
2. Pass it through the network
3. Predict a related token (next word or context word)
4. Compute loss (typically cross-entropy)
5. Backpropagate error and update weights
6. Repeat across data until loss converges

---

## Where the Embedding Comes From
After training:
- The **hidden layer activations** become the token embeddings
- Each token’s embedding is obtained by:
  - One-hot vector × learned weight matrix

These vectors encode:
- Semantic similarity
- Linguistic relationships  
  - *king − man + woman ≈ queen*
  - *Paris − France + Germany ≈ Berlin*

---

## Embedding Size: A Practical Trade-off
Choosing the embedding dimension (`d`) depends on:
- Complexity of downstream tasks
- Latency and compute constraints
- Storage and inference cost

**Typical ranges:**
- Hundreds to thousands
- Common values like `300`, `512`, `768`

There is no perfect value—this choice is **empirical**.

---

## When to Stop Training
Since the task is a proxy:
- Monitor training loss across epochs
- Stop when loss **converges**
- Embeddings are then reused for downstream tasks

---

## Key Takeaways
- One-hot encoding cannot represent semantic similarity
- Learned embeddings solve this by encoding meaning geometrically
- Word2Vec introduced proxy tasks to learn embeddings from raw text
- Embedding quality depends on data, task, and dimensionality
- Token representation is a **critical foundation** for all modern NLP systems


## From Token Embeddings to Sentence Representations

### Why Token Embeddings Are Not Enough
So far, we learned how to represent **individual tokens** using embeddings.  
However, many NLP tasks require understanding **sentences or longer pieces of text**.

A naive approach is to **average token embeddings**, but this fails because:
- Word **order is lost**
- Context is ignored
- The same token always has the same representation, regardless of where it appears

To address this, models must capture the **sequential nature of language**.

---

## Recurrent Neural Networks (RNNs)

### Core Idea
A **Recurrent Neural Network (RNN)** processes text **one token at a time**, while maintaining a **hidden state** that summarizes the sequence seen so far.

Instead of treating tokens independently:
- The model keeps memory of previous tokens
- Word order directly affects the representation

RNNs were first introduced in the **1980s**.

---

### How an RNN Works (High Level)
At each time step:
- Input: current token + previous hidden state
- Output: updated hidden state (and optionally a prediction)

The **hidden state** (also called activation or context vector):
- Encodes the meaning of the sequence so far
- Evolves as more tokens are processed

---

### Example Intuition
For the sentence:
> *“a cute teddy bear is reading”*

The RNN:
1. Processes `"a"` → updates hidden state  
2. Uses that hidden state + `"cute"` → updates again  
3. Continues token by token  

Each hidden state represents **everything the model has read up to that point**.

---

## Using RNNs for Different NLP Tasks

### Classification
- Use the **final hidden state**
- Project it into a label space (e.g., positive vs negative sentiment)

### Multi-Classification (Sequence Labeling)
- Use the hidden state at **each token**
- Predict labels per token (e.g., Named Entity Recognition)

### Generation
- Encode the input sequence into a final hidden state
- Use that state to **generate output text**

---

## Limitations of RNNs

### 1. Long-Range Dependency Problem
All information about the past must be stored in **one hidden state**.

As sequences grow longer:
- Early information is gradually forgotten
- The model struggles to “remember” distant tokens

---

### 2. Vanishing Gradient Problem
During training:
- Errors are backpropagated through time
- Gradients become products of many small values

Result:
- Gradients shrink toward zero
- Early layers stop learning effectively
- The model cannot learn long-term dependencies

This issue is known as **vanishing gradients**.

---

### 3. Computational Inefficiency
RNNs are **inherently sequential**:
- Each token depends on the previous one
- No parallel computation across tokens

This makes training **slow**, especially for long sequences.

---

## Long Short-Term Memory (LSTM)

### Motivation
**LSTMs** were introduced to mitigate RNN limitations, especially long-term memory issues.

### Key Idea
In addition to the hidden state, LSTMs maintain a **cell state**:
- Acts as a long-term memory
- Explicitly controls what information to keep or forget

This helps preserve important information across longer sequences.

---

### Reality Check
While LSTMs improve memory:
- They do not fully solve long-range dependency issues
- They are still sequential
- Training remains slow

---

## Why Attention Emerged

### Core Problem
RNN-based models:
- Compress all past information into a single vector
- Struggle with long sequences
- Cannot directly access earlier tokens

### Key Insight
When predicting the next token, it would be useful to:
- **Directly look back** at relevant parts of the input
- Not rely solely on a compressed hidden state

---

## Intuition Behind Attention
**Attention** introduces direct connections between:
- What the model is predicting now
- Specific parts of the input sequence

Instead of remembering everything indirectly:
- The model can **focus on relevant tokens**
- Past information becomes directly accessible

This idea lays the foundation for **modern Transformer models**.

---

## Recap
- Token embeddings alone cannot represent full meaning
- RNNs introduce order and sequence awareness
- RNNs suffer from memory, gradient, and speed issues
- LSTMs partially address memory limitations
- Attention was introduced to directly connect predictions with relevant past information

This transition sets the stage for understanding **attention mechanisms and Transformers**.


## Attention and Self-Attention: The Core Idea Behind Transformers

### Why Attention Was Introduced
Earlier sequence models struggled with **long-range dependencies**:
- Information from far earlier tokens was hard to remember
- Everything had to be compressed into a single hidden state

**Attention**, introduced in **2014**, was designed to solve this problem by allowing models to:
- Directly reference relevant parts of the input
- Avoid relying on a single compressed memory

This idea becomes **central** to modern NLP architectures.

---

## From Sequential Processing to Direct Access
Traditional models process text **step by step**.
Attention changes this by allowing the model to:
- Look at **all tokens at once**
- Create **direct connections** between tokens, regardless of distance

This removes the need to process information strictly in order.

---

## Self-Attention
**Self-attention** means:
> Each token builds its representation by looking at **all other tokens in the same sequence**.

Example sentence:
> *“a cute teddy bear is reading”*

To compute the representation of **“teddy bear”**:
- The model considers **every other token** in the sentence
- Each token contributes differently depending on relevance

As a result:
- Token representations become **context-aware**
- The same word can have different representations in different contexts

---

## Contextual Meaning
Self-attention enables **context-dependent representations**:
- A word’s meaning is shaped by surrounding words
- The same word can have different vectors in different sentences

This directly addresses earlier limitations where:
- One word always had the same embedding
- Context was ignored

---

## Transformers and Self-Attention
In **2017**, the Transformer architecture was introduced with the key idea:
> **“Attention is All You Need”**

The core change:
- Remove sequential recurrence
- Rely entirely on **self-attention** to model relationships in text

This led to:
- Better handling of long-range dependencies
- Faster training due to parallel computation
- Strong performance on tasks like translation

---

## Query, Key, and Value (Q, K, V)

To formalize attention, three concepts are introduced:

- **Query (Q)**: What we are currently focusing on
- **Key (K)**: What we compare the query against
- **Value (V)**: The information carried by each token

---

### High-Level Intuition
To represent a token (e.g., *“teddy bear”*):
1. Treat it as a **query**
2. Compare it against **keys** of all tokens
3. Measure similarity between the query and each key
4. Use those similarities as **weights**
5. Compute a weighted combination of the corresponding **values**

Tokens that are more relevant:
- Receive higher weights
- Contribute more to the final representation

---

## Learned Representations
Important points:
- Queries, keys, and values are **not fixed**
- They are produced via **learned linear projections**
- The model learns what to compare and what information to extract

Interpretation:
- **Keys** decide *relevance*
- **Values** carry *content*
- **Queries** decide *what to look for*

---

## Efficient Computation
Self-attention can be computed:
- Using **matrix operations**
- Across the entire sequence at once

Benefits:
- Highly efficient on GPUs
- Enables parallel processing
- Scales well with modern hardware

Weights are typically computed using:
- A similarity function between queries and keys
- A **softmax** to normalize importance across tokens

---

## Key Takeaways
- Attention was introduced to solve long-range dependency issues
- Self-attention allows each token to directly access all others
- Token representations become context-aware
- Query, Key, and Value formalize how relevance is computed
- This mechanism is the foundation of the Transformer architecture

This idea forms the backbone of all modern Large Language Models.


## Transformer Architecture: From Attention to End-to-End Translation

### High-Level Motivation
Self-attention enables models to **directly connect any token with any other token**, solving long-range dependency and memory issues seen in sequential models.  
The **Transformer architecture** is built entirely around this idea.

---

## Encoder–Decoder Structure

The Transformer consists of **two main components**:

### 1. Encoder
- Processes the **input (source language) text**
- Produces **context-rich representations** for each input token
- All input tokens can attend to each other

### 2. Decoder
- Generates the **output (target language) text**
- Predicts tokens **one step at a time**
- Uses information from:
  - Previously generated output tokens
  - Encoded input representations

This structure was originally designed for **machine translation**, but generalizes to many tasks.

---

## Encoder: Building Contextual Input Representations

Each encoder block contains:

### Self-Attention Layer
- All input tokens attend to **each other**
- Each token representation becomes a function of the entire input sequence

### Feed Forward Neural Network (FFN)
- Applies a learned projection independently to each token
- Adds expressive power and non-linearity

**Output of the Encoder:**
- A sequence of **context-aware token embeddings**
- These embeddings encode meaning relative to the full input sentence

---

## Decoder: Generating Output Tokens

Each decoder block contains **two attention layers** and a feedforward layer.

### 1. Masked Self-Attention
- Tokens attend only to **previously generated tokens**
- Future tokens are masked to prevent information leakage
- Ensures autoregressive generation

### 2. Cross-Attention (Encoder–Decoder Attention)
- Queries come from the decoder
- Keys and values come from the encoder outputs
- Allows the decoder to focus on **relevant parts of the input**

### Feed Forward Neural Network
- Further transforms token representations

---

## Query, Key, and Value in the Decoder

- **Query (Q):** comes from the decoder (what we are trying to predict)
- **Keys (K) & Values (V):** come from the encoder (input sentence)

**Interpretation:**
> “Given what I’m generating now, which parts of the input matter most?”

---

## Multi-Head Attention

### What Is a Head?
A **head** is one set of learned projections used to compute:
- Queries
- Keys
- Values

### Why Multiple Heads?
- Allows the model to learn **different types of relationships in parallel**
- Each head focuses on different aspects of the sequence

**Analogy:**
- Similar to using multiple filters in a convolutional neural network

There are typically **no constraints** forcing heads to be different—the model learns useful diversity naturally.

---

## Positional Encoding (Why It’s Needed)
Self-attention alone has **no sense of order**.
To encode word position:
- Positional information is **added to token embeddings**
- Enables the model to understand sequence structure

This restores the notion of word order without recurrence.

---

## Output Layer
At each decoding step:
- The decoder produces a vector
- This vector is projected to vocabulary size
- A **softmax** converts it into probabilities
- The most likely token is selected as the next output

---

## Label Smoothing

### Motivation
In language generation:
- Multiple outputs may be valid
- Hard one-hot targets can make models **overconfident**

### What Label Smoothing Does
- Replaces strict one-hot labels with a softened distribution
- Assigns:
  - High probability to the correct token
  - Small probabilities to other tokens

### Benefits
- Reduces overconfidence
- Improves generalization
- Improves translation metrics like BLEU

This is a **general and widely used technique** in NLP.

---

## Key Takeaways
- Transformers rely entirely on **attention**, not recurrence
- Encoder builds rich, context-aware input representations
- Decoder generates outputs using masked self-attention and cross-attention
- Multi-head attention increases representational power
- Positional encoding restores word order
- Label smoothing improves robustness in generation tasks

This architecture forms the **foundation of all modern Large Language Models**.


## Transformer: End-to-End Flow with a Concrete Example

We walk through the full Transformer pipeline using the example sentence:

> **“a cute teddy bear is reading”**

This section connects all concepts into a single, coherent flow.

---

## Step 1: Tokenization and Special Tokens
- The sentence is split into **tokens** using a chosen tokenization strategy
- Two special tokens are added:
  - **BOS (Beginning of Sequence)** → marks the start
  - **EOS (End of Sequence)** → marks the end

This allows the model to:
- Know when generation should start
- Know when generation should stop

---

## Step 2: Input Embeddings + Positional Information

Each token representation is composed of **two parts**:

### 1. Token Embedding
- A learned vector representation
- Encodes semantic meaning of the token

### 2. Positional Encoding
- Adds information about **where the token appears in the sequence**
- Original Transformer uses **sine and cosine functions**
- Added **element-wise** to the token embedding

**Result:**  
Each token becomes a **position-aware embedding**.

All embeddings together form a matrix of shape:
- `n × d_model`
  - `n` → sequence length
  - `d_model` → embedding dimension

---

## Step 3: Encoder – Self-Attention over Input Tokens

### Projection into Q, K, V
The same input embeddings are projected into three spaces using learned matrices:
- **Queries (Q)**
- **Keys (K)**
- **Values (V)**

These projections are learned during training.

---

### Scaled Dot-Product Self-Attention

The core attention computation is:

$$
\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V
$$

**Intuition by steps:**
1. Each **query** is compared with all **keys**
2. Dot products measure similarity
3. Scaling by `√d_k` stabilizes gradients
4. Softmax converts similarities into weights
5. Weights are applied to **values**
6. Each token becomes a weighted combination of all tokens

**Outcome:**
- Each input token representation is updated based on **all other tokens**
- Representations become **context-aware**

---

## Step 4: Multi-Head Attention

- Self-attention is performed **multiple times in parallel**
- Each repetition is called a **head**
- Each head uses different projection matrices

### Why Multiple Heads?
- Enables the model to learn **different relationships simultaneously**
- Each head can focus on different aspects of the sequence

### Aggregation
- Outputs from all heads are **concatenated**
- A final projection matrix **Wₒ** maps them back to `d_model`

There are **no constraints** forcing heads to be different, but training naturally leads to diversity.

---

## Step 5: Feed Forward Network (FFN)

After attention:
- Each token is passed through a **position-wise feed forward network**
- Hidden dimension is **larger than `d_model`**

**Purpose:**
- Increase representational capacity
- Allow more complex feature transformations

---

## Step 6: Stacked Encoders
- The encoder block is repeated **N times**
- Output of one encoder becomes input to the next
- Final encoder output = **deep, context-rich representations**

These encoded embeddings are passed to the decoder.

---

## Step 7: Decoder – Generating Output Tokens

### Initialization
- Decoding starts with the **BOS token**
- Goal: predict the **next token**

---

### Decoder Block Components

#### 1. Masked Self-Attention
- Tokens attend only to **themselves and previous tokens**
- Prevents looking at future tokens
- Ensures autoregressive generation

#### 2. Cross-Attention
- **Queries** → from decoder
- **Keys & Values** → from encoder outputs

This allows the decoder to:
- Focus on relevant parts of the input sentence
- Align output tokens with input tokens

#### 3. Feed Forward Network
- Same role as in the encoder
- Adds expressive power

---

## Step 8: Output Projection and Prediction
- Decoder output is projected to **vocabulary size**
- Softmax produces a probability distribution
- Highest-probability token is selected as the next word

---

## Step 9: Autoregressive Generation Loop
1. Predict next token
2. Feed predicted token back into the decoder
3. Repeat decoding steps
4. Stop when **EOS token** is generated

---

## Big Picture Recap
- Token embeddings + positional encodings form the input
- Encoder builds **context-aware input representations**
- Decoder generates output tokens step by step
- Self-attention removes sequential bottlenecks
- Multi-head attention increases modeling power
- The process continues until EOS is produced

This complete pipeline is the foundation of **Transformer-based language models**.
